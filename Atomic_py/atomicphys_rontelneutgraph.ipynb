За допомогою звичайного світла, довжина хвилі якого сягає декількох тисяч нанометрів, вивчати структри атомних масштабів одиниць та долей нанометрів неможливо. Тому зрозуміло, що відкриття Х-променів було важливим для прогресу експериментальних методів дослідження структури атомів та молекул. Виникла __РЕНТГЕНОГРАФІЯ__. Зменшення довжини хвилі дозволяє збільшити роздільну здатність. Тому факт навяності дифракції та інтерференції для атомних частинок (електронів, нейтронів, мюонів та інш) дозволив побудувати потужні методи досліджень атомної структури речовини (електронографія та нейтронографія). Першим і головним звісно є рентгеноструктурний аналіз - група методів досліджень атомної будови речовини по розподілу інтенсивності розсіяного рентгенівського випромінювання. Спостереження характерних максимумів інтенсивності дозволяє встановлювати координати атомів з точністю до 0,1-0,01 нм, визначати характеристики теплових коливань цих атомів, розподіл валентних електронів на хім. зв'язках у молекулах. Серед дифракційних методів дослідження рентгеноструктурний аналіз є найбільш поширеним та розвиненим. Його можливості доповнюють методи нейтронографії та електронографії.

Дифракція рентг. променів на кристалах була відкрита в 1912 М. Лауе та його співробітниками В. Фрідріхом та П. Кніппінгом. Розроблена Лауе теорія дифракції рентг. променів на кристалах дозволила зв'язати довжину хвилі випромінювання $\lambda$, лінійні розміри елементарної ячійки кристала $а, b, с$, кути падаючого $\alpha_0, \beta_0, \gamma_0$ і дифракційного $\alpha_0, \beta_0, \gamma_0$ промінів 
співвідношеннями (рівняння Лауе):
$$
a\,(\cos{\alpha} - \cos{\alpha_0}) = n_1\\
b\,(\cos{\beta} - \cos{\beta_0}) = n_2\\
c\,(\cos{\gamma} - \cos{\gamma_0}) = n_3
$$
де цілі числа $n_{i}\, i=1,2,3$ - кристалографічні індекси (Miller indices). У 1913 Л. Брегг  і Г. В. Вульф показали, що дифракцію рентг. пучка можна розглядати як відбиття падаючого променя від деякої системи кришталографічних площин з міжплощинною відстанню $d$:
$$2\,d\sin\theta = n\,\lambda$$

[Експерименти Девіссона-Джермера](https://ru.wikipedia.org/wiki/Опыт_Дэвиссона_—_Джермера) продемонстрували дифракційні ефекти при відбитті електронів від кристалів. Одна із старих назв квантової механіки - хвильова механіка (картина Шредінгера) -

У дослідах Девіссона та Джермера максимум інтенсивності реєстрації електронів при відбитті від кристалу нікеля спостерігалося при прискорюючій різниці потенціалів $U = 54\,V$, що відповідає дебройлівській довжині хвилі
$$
\lambda_{max} = \frac{h}{\sqrt{2\,m\,e\,U}} \approx 1.67\cdot 10^{-10}\,m
$$
що майже співпало з значенням сталої гратки отриманої з дифракції рентгенівських променів $\lambda = 0.165\,\text{nm}$. 

```python
from scipy import constants
import numpy as np
lmbd= constants.value('Planck constant')/np.sqrt(2*constants.value('electron mass')*constants.value('elementary charge')*54)
print('lambda = ',"{:.3}".format(lmbd),'m')
```

__ЕЛЕКТОРНОГРАФІЯ__ -метод вивчення структури речовини, заснований на дослідженні розсіювання зразком прискорених електронів. Застосовується для вивчення атомної структури кристалів, аморфних тіл та рідин, молекул газів. В електронографах та електронних мікроскопах формуються формується вузький пучок прискорених електронів. Він прямує на об'єкт і розсіюється ним. Дифракційна картина (електронограма) або фотографується, або реєструється електронним пристроєм. Основними варіантами методу є дифракція швидких електронів (прискорювальна напруга від 30-50 keV і більше) та дифракція повільних електронів ($\sim 0.1\,$keV). Інтенсивна взаємодія електронів з речовиною обмежує товщину просвічуваних ними зразків десятими частками мкм. 

```python
from scipy import constants
constants.find('neutron mass')
```

```python
enrg=1.6e-19*50000
print('lambdadB(E=50keV) = ',"{:.2e}".format(
constants.value('Planck constant')/np.sqrt(2*constants.value('electron mass')*enrg)),'m')
```

[__НЕЙТРОНОГРАФІЯ__](https://neutron.ethz.ch/neutron-scattering.html) — сукупність методів дослідження структури та властивостей речовини за допомогою розсіювання нейтронів низьких енергій ($<1 eV$). Довжина хвилі де Бройля повільних нейтронів можна порівняти з міжатомними відстанями в конденсованих середовищах, що дозволяє вивчати розташування атомів. Цей метод не потребує вакууму, оскільки нейтрони вільно проходять газове середовище і розсіюються тільки на ядрах.

```python
import numpy as np
enrg=1.6e-19
print('lambdadB(E=1eV) = ',"{:.2e}".format(
constants.value('Planck constant')/np.sqrt(2*constants.value('neutron mass')*enrg)),'m')
```

Маса та кінетич. енергія нейтрону співмірна з масою атома і енергією міжатомних взаємодій у речовині, що дозволяє за допомогою __непружного__ розсіювання нейтронів досліджувати динамічні властивості окремих атомів та молекул у середовищі. Магнітний момент нейтрону взаємодіє із магнітними моментами атомів, що дозволяє за інтенсивністю і поляризації магнітного розсіювання визначати магнітні моменти атомів та відповідні магнітні структури. Малокутове розсіяння нейтронів [SANS](https://en.wikipedia.org/wiki/Small-angle_neutron_scattering) застосовується для дослідження структурних, динамічних та магнітних властивостей практично всіх відомих форм конденсованого стану речовини, від простих рідин і кристалів до біологічних макромолекул.
