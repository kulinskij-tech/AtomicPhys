# Рівняння Шредінгера вільної частинки та густина потоку ймовірності
<!--  -->
Хвильова функція квантової системи в даний момент часу несе всю інформацію про її поведінку між процесами вимірювання. Рівняння яке визначає швидкість зміни х.ф. з часом називається [рівнянням Шредінгера](https://en.wikipedia.org/wiki/Schrödinger_equation).
Для вільної частинки х.ф. має вигляд суперпозиції хвиль де Бройля амплітуди яких вожна вважати незалежними від часу:
<!--  -->
\begin{equation}\label{eq:fp} 
\psi (\mathbf{r},t)=\int\limits_{\Lambda}\,c(\mathbf{p})\,\psi_{\mathbf{p}} (\mathbf{r},t)d\,\mathbf{p}\\
\psi_{\mathbf{p}} ({\rm r},t) =\frac{1}{(2\pi \hbar)^{3/2}}\,\exp \left( \frac{i\mathbf{p} \cdot \mathbf{r}-i\,E\,t}{\hbar}\right)\,,\,\,\,\,E = \frac{\mathbf{p}^2}{2\,m}
\end{equation}
<!--  -->
Звідси отримаємо:
<!--  -->
\begin{equation}\label{eq:fpsh1} 
i\,\hbar\,\frac{\partial }{\partial\,t}\psi (\mathbf{r},t) = 
\int\limits_{\Lambda}\,c(\mathbf{p})\,\frac{\mathbf{p}^2}{2\,m}\,\psi_{\mathbf{p}} (\mathbf{r},t)\,d\,\mathbf{p} = -\frac{\hbar^2}{2\,m}\,\Delta\,\left(\int\limits_{\Lambda}\,c(\mathbf{p})\,\psi_{\mathbf{p}} (\mathbf{r},t)\,d\,\mathbf{p}\right)
\end{equation}
<!--  -->
Отже:
<!--  -->
\begin{equation}
\label{eq:schr} i\,\hbar \frac{\partial}{\partial\, t}
\psi(\mathbf{r},t) =\hat{H}\, \psi(\mathbf{r},t)\\
\end{equation}
<!--  -->
тобто для вільної частиники швидкість зміни хвильової функції з часом визначається дією оператора (кінетичної) енергії:
<!--  -->
$$\hat{H} = -\frac{\hbar^2}{2\,m}\,\Delta$$
<!--  -->
на хвильову функцію.

Узагальнюючи цей результат можна вважати, що швидкість зміни хвильової функції з часом визначається оператором енергії - гамільтоніаном, який є аналогом функції Гамільтона класичної механіки. Для частинки у зовнішньому полі з потенціальною енергією $V$ можна вважати що:
<!--  -->
\begin{equation}\label{eq:ham}
\hat{H} = - \frac{\hbar^2}{2\,m}\,\Delta + \hat{V}(\mathbf{r})
\end{equation}
<!--  -->
є природнім виразом для оператора енергії.  

Факт береження повної ймовірності і те що її локальна зміна відбуається за рахунок потоку частинок крізь стінки, дозволяє отримати диференціальну форму закона збереження аналогічну до відомих аналогів з механіки суцільних середовищ та ввести густину струму ймовірності:
<!--  -->
\begin{equation}
\frac{d}{d \,t}\int\limits_{V}\rho\, dV = 
\int\left(\frac{\partial \psi^*}{\partial t}\,\psi +
\psi^*\frac{\partial \psi}{\partial t}
\right)\,dV = \frac{\hbar}{i\,2\,m}\int \left(\psi\,\Delta \psi^* -
\psi^* \,\Delta \psi\right)dV=\\
-\frac{\hbar}{m}\int  {\rm div}\,\left(\frac{\psi^*\,\nabla \psi -
\psi \,\nabla \psi^*}{2\,i}\right)dV = - \int \mathbf{j}\cdot \mathbf{n}\,dS\\
\mathbf{j} = \frac{\hbar}{m}{\rm Im}\, \psi^*\nabla \psi
\end{equation}
$$\frac{\partial \rho}{\partial t}+{\rm div} \mathbf{j} =0$$
***

## Стаціонарне рівняння Шредінгера

Якщо гамільтониан системи не залежить від часу, у системи є т.з. стаціонарні стани які визначаються стаціонарним рівнянням Шредінгера: 
<!--  -->
$$
\hat{H}\,\psi_{E}(\mathbf{r},t) = E\,\psi_{E}(\mathbf{r},t)
$$
<!--  -->
Вони характеризуються тим, що їх часова залежність может бути виділена явним чином: 
<!--  -->
\begin{equation}
\label{stac} \psi_{E}(\mathbf{r},t) =\exp\left(\frac{E\,
t}{i\,\hbar}\right)\phi_{E}(\mathbf{r})
\end{equation}
<!--  -->
Густина ймовірності $\rho$ в таких станах не залежить від часу, тобто $\rm{div} \mathbf{j} =0$ у кожній внутрішній точці області руха. Зокрема, для хвиль де Бройля :
<!--  -->
$$
\psi_{\mathbf{k}}(\mathbf{r})=C\exp(i\,\mathbf{k}\cdot \mathbf{r}-
i\,\omega\, t)
$$ 
<!--  -->
\begin{equation}\label{dBrl}
  \mathbf{j}=\frac{\hbar\, \mathbf{k}}{m}\,|C|^2\,,
\end{equation}
<!--  -->
що відповідає змісту густини потока як числа частинок, який проходить через одиничную площадку в одницю часу. Величина
<!--  -->
$$\mathbf{v} = \frac{\hbar \mathbf{k}}{m}$$
<!--  -->
природньо ототожнюється із швидкістю потоку який складається з вільних частинок.

Спектр гамільтоніана є найважливішою характеристикою квантової системи. Як витікає з властивостей власних векторів для операторів фізичних величин, стаціонарні стани утворюють повний набір векторів так що довільний стан
квантовой системи може бути представлено у вигляді їх суперпозиции:
<!--  -->
\begin{equation}
\label{stat1} \psi
=\sum\limits_{n,\ldots}c_{n}\phi_{n}+\int\limits_{cont}a(E)\,\phi_{E}\,dE\,,\ldots
\end{equation}
<!--  -->
тут крапки позначають інші змінні, які можуть бути задані в стані з певною енергією у випадку виродженя. Зазвичай дикретний спектр енергії виникає, коли квантовий рух відбувається у обмеженій області або ймовірність спостереження швидко зникає на нескінченності - т.з. __зв'язані стани__ ("електрон в атомі"). Стани з неперервними значенням енергії відповідають можливості спостереження цієї системи на некінченності - т.з. __стани розсіяння__ ("електрон вибитий з атома іонізуючим випромінюванням"). Якщо стаціонарні стани відомі то еволюція довільного стана $\psi$ визначається суперпозицією еволюцій стаціонарних станів за \eqref{stat1}.
