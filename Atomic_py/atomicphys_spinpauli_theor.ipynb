## Спін як внутрішня характеристика частинки.
<!--  -->
$\newcommand{\ket}[1]{\left.\left\vert\, #1\,\right. \right \rangle}$
$\newcommand{\bra}[1]{\left. \langle \, #1\, \right\vert}$
$\newcommand{\braket}[2]{\langle \,#1 \vert #2 \, \rangle}$
$\newcommand{\av}[3]{\left\langle \,#1 \left \vert\, #2 \,\right\vert #3 \,\right \rangle}$
В рамках картини борівських орбіт зрозуміло, що наявність момента імпульсу $\mathbf{L}$ у електрона на орбіті призводить до появи магнітного моменту $\mathbf{M}_{L}$ величина якого повязана із орбітальним моментом відомим співвідношенням:
<!--  -->
$$
\mathbf{M}_{L} = \frac{q}{2\,m_{e}\,c}\,\mathbf{L}
$$
коефіцієнт $\frac{q}{2\,m_{e}\,c}$ називається гіромагнітним відношенням. Зважаючи на квантування проекції момента імпульсу, магнітний момент також є квантованим і його квант носить назву (атомного) магнетона Бора:
$$
\mu_{B} = \frac{\hbar\,|e|}{2\,m_e\,c}
$$

У зовнішньому неоднорідному магнітному полі на магнітний момент діє сила (нагадаємо, що потенціальна енергія $U_{m} = -\mathbf{M}\cdot \mathbf{B}$):
<!--  -->
$$
\mathbf{F} = - \nabla\,U_{m} =  \nabla \left(\mathbf{M}_L\cdot \mathbf{B}\right)
$$
<!--  -->
тому можна очікувати, що пучок атомів буде розділятись на відповідні підпучки з різними значеннями проекції момента на напрямок градієнта поля. Таких пучків повинно бути $2\,l+1$ ($l=0,1,2\ldots$) тобто непарна кількість. Але для атомів водня, лужних металів та срібла де зовнішній електрон перебуває у воднеподібному стані з $l=0$ в [дослідах Штерна-Герлаха](https://en.wikipedia.org/wiki/Stern–Gerlach_experiment) неоднорідне магнітне поле призводило до розділення пучка тільки на дві(!) компоненти. Для пояснення цього факту, [Уленбек](https://en.wikipedia.org/wiki/George_Uhlenbeck) та [Гаудсміт](https://en.wikipedia.org/wiki/Samuel_Goudsmit) припустили що це може бути пояснено наявністю власного магнітного і отже власного кутового момента у електрона, який отримав назву спіна (spin). Значення проекції спіна електрона таким чином може бути $\pm \frac{\hbar}{2}$. Згодом такий власний момент був відкритий і для всіх інших частинок мікросвіту. Тому разом із орбітальним квантовим числом $l$ вводять також спінове квантове число $s$ яке визначає максимальне значення власного (магнітного) моменту на вісь. Проекція спіна характеризується магнітним спіновим числом $-s\le m_s\le s$.

<!--  -->
Спін $\mathbf{S}$ є додатковою внутрішньою характеристикою частинки яка проявляється через наявність власного магнітного моменту який взамодіє з магнітним полем. Відповідне гіромагнітне відношення для електрону:
<!--  -->
\begin{equation}\label{eq:gyrospin}
\mathbf{M}_{S} = \frac{e}{m_{e}\,c}\,\mathbf{S}
\end{equation}
<!--  -->

Експеримент вказує, що магнітний момент $\mathbf{M}_{s}$ пов'язаний із спіном існує навіть якщо частинка нейтральна (наприклад <a href='https://en.wikipedia.org/wiki/Neutron_magnetic_moment'>нейтрон</a>), що свідчить про її складну (кваркову) внутрішню структуру. Спін є прикладом внутрішньої ступені вільності аналогічної до заряду чи маси покою. Прикладом інших внутрішніх квантових чисел, якими характеризується стан частинок є ізотопічний, баріонний, лептоний заряди, а також дивність та чарівність. Вони проявляються при аналізі процесів зіткнення та взаємоперетворення частинок. Приклад спіна надає розуміня загальнгого механізму введення такого типу квантових чисел та звязок із властивостями симетрії.

Спін та гіромагнітне відношення проявляються у [ефекті Айнштайна-деХааса](https://en.wikipedia.org/wiki/Einstein%E2%80%93de_Haas_effect). Він полягає у появі обертального моменту зразка (феромагнетика або парамагнетика) при його перемагничені (завдяки закону збереження момента імпульсу зразок при перемагниченні набуває обетального моменту який протилежний до внутрішнього моменту електронів які впорядковуються в напрямку поля).

Послідовне введення поняття спіну засноване на релятивістській інваріантості рівняннь квантової теорії і це було показано [Діраком](https://en.wikipedia.org/wiki/Paul_Dirac). Однак, завдяки тісному зв'язку цього поняття із властивостями інваріантості системи відносно поворотів, спін також природнім чином виникає у нерелятивістській квантовій механіці. Оскільки спінова ступінь свободи не залежить від положення частинки у просторі (виключаючи релятивістські ефекти), то можна розглядати тільки залежність хвильової функції від спінових ступенів свободи. Таким чином, простір спінових ступенів свободи для частинки із спіном $s = \frac{1}{2}$ являє собою 2-вимірний (комплексний) гільбертовий простір. В загальному випадку розмірність $2s+1$.
<br>
***
Тобто хвильова функція має структуру $2s+1$ вектора:
<!--  -->
$$
\Psi(\mathbf{r})=
\begin{pmatrix} \psi_{-s}(\mathbf{r})
   \\
  \ldots \\
\psi_{+s}(\mathbf{r})
\end{pmatrix}
$$
компоненти якого відповідають станам з проекціями спінового моменту $S_{z} = -\hbar\,s,-\hbar\,(s-1),\ldots, +\hbar\,s$
<!--  -->
У випадку $s=1/2$ оператор проекції спіну на виділену вісь $\mathbf{n}\cdot \hat{\mathbf{S}}$  у базис своїх власних станів має вигляд:
<!--  -->
$$
\mathbf{n}\cdot \hat{\mathbf{S}} = \frac{\hbar}{2}
\begin{pmatrix}
1&0\\
0&-1
\end{pmatrix}
$$
<!--  -->

Момент імпульсу системи, яка складається з підсистем $\Sigma_1, \Sigma_2, \ldots$ визначається за правилом суми моментів: якщо підсистема $\Sigma_1$ є у власному стані обертального моменту з $l_1$, і підсистема $\Sigma_2$ має стан обертального моменту з $l_2$, то комбінована система $\Sigma_{1}\otimes \Sigma_{2}$ перебуває у суперпозиційному стані оператору обертального моменту. Максимальна кількість компонент в такій суперпозиції  дорівнює $(2l_1+1)(2l_2+1)$. Максимальне значення магнітного квантового числа такої системи сягає $M_{max} = (l_1+l_2)$. Наприклад для системи двох частинок такий стан відповідає хвильовій функції:
<!--  -->
$$\Psi(\mathbf{r}_1,\mathbf{r}_2) = \psi_{l_1,l_1}(\mathbf{r}_1)\,\psi_{l_2,l_2}(\mathbf{r}_2)$$
<!--  -->
В загальному випадку до суперпозиції входять стани із наступними значеннями обертального квантового числа $|l_1-l_2|, |l_1-l_2|+1,\ldots, l_1+l_2$. Доведення цього факту буде преставлено у курсі квантової механіки. Наприклад, якщо два електрони знаходяться на $p$-орбіталях ($l=1$), то можливими значеннями орбітального моменту є $L = 0,1,2$, а спіна $S = 0,1$.<br>
<!--  -->
Виявляється, що за математичними властивостями (комутаційні співвідношення) оператори спіна $\hat{\mathbf{S}}$ та повного моменту
<!--  -->
$$\hat{\mathbf{J}} = \hat{\mathbf{S}}+\hat{\mathbf{L}}$$ 
<!--  -->
аналогічні оператору кутового момента $\hat{\mathbf{L}}$ (звісно за виключенням того що $s$ і відповідно $j$ можуть бути полуцілими). Тому для них має місце також наступне
<!--  -->
- оператор квадрата спіна (повного момента) і проекції на певну вісь є одночасно спостережувальними величинами;
- власні значення оператора $\mathbf{S}^{\,2}$ дорівнюють $\hbar^{2} s\,(s + 1)$, $\mathbf{J}^{\,2} 
\to \hbar^{\,2}j\,(j + 1)$ відповідно
- власні значення оператора проекції моменту імпульсу на певну вісь
$\hbar \,m_{s|j}\,\,,\,\, - s|j \le m_{s|j} \le s|j\,\,,$
тобто проекція моменту спіна (повного момента) на цю вісь є квантованою. 

## Бозони та Ферміони. Принцип Паулі.
<!--  -->
Спін є важливою характеристикою елементарних частинок. Всі частинки підпадають під два класи: частинки з цілим спіном - бозони і частинки з полуцілим спіном - ферміони. 

Бозони - колективісти вони можуть всі знаходитись у одному стані. Ферміони - індивідуалісти два ферміона не можуть знаходитись в одному стані це __принцип заборони Паулі__. Електрони, протони та нейтрони є ферміонами ($s=1/2$). Ця обставина є вирішальною при пояснені стійкості речовини, періодичного закону зміни хімічних властивостей речовин та багато інших явищ. 

Важливу роль відіграє суттєва відмінність поняття однаковості частинок у квантовій теорії від такої у класичній фізиці. В класичній фізиці немає справжньої тотожності частинок оскільки незважаючи на співпадіння їх фізичних характеристик (маса, заряд та інш.) вважається, що спостерігач завжди тоже надати їм мітки-номери за якими він відстежує еволюцію кожної з цих частинок. Фактично це еквівалентно присвєню кожної частинкі індивідуальної спостережної величини (номер) яка притамана тільки їй. Тому, строго кажучи, у класичній фізиці всі частинки є принципово різними. Зрозуміло що це відображає припущення про існування траєкторії. Тому, зрозуміло, що відсутність такого поняття як траєкторія у квантовій механіці призводить до неможливості спостереження нумерації частинок. 
Математичне вираження факту тотожності (незалежності стану від обраного способу нумерации тотожніх частинок) полягає в тому, що гамільтоніан системи тотожніх частинок є інвариантним відносно довільної перестановки цих частинок. Таким чином, гамільтоніан системи комутує зі всіма операторами перестановок пар тотожніх частинок (такі перестановки називаються транспозиціями). З таких транспозицій можна скласти довільну перестановку. Таким чином:
<!--  -->
\begin{equation}
\label{19eq2} \left[ H,T_{mn}  \right] = 0\,\,.
\end{equation}
<!--  -->
Слід зауважити, що оператори перестановок між собою, взагалі кажучи, не комутують.
<!--  -->
Оскільки
<!--  -->
$$T_{mn}^{2} = 1\,\,,$$
<!--  -->
то власні значення такого оператора дорівнюють $\pm 1$, тому власні функції такого оператора є либо парними, або непарними відносно транспозиції довільної пари тотожніх
частинок. У квантовій теорії поля доводиться теорема про зв'язок спіна і статистики: частинки з цілим спіном є бозонами (статистика Бозе-Айнштайна), частинки з полуцілим спіном є ферміонами (статистика Фермі-Дирака).

Врахування статистики тотожніх частинок при розвязанні рівняння Шредінгера, навіть якщо воно не включає спіну, полягає у виборі з всього простору тільки тих розв'язків, які відповідають хвильовим функціям з правильною симетрією відносно перестановки тотожніх частинок. Для системи невзаємодіючих частинок у випадку статистики
Бозе-Айнштайна, наприклад, хвильова функція має вигляд:
<!--  -->
\begin{equation}
\label{19eq5}
\begin{array}{l}
 \psi (1,2,\ldots,N) = \left( \frac{1}{N!} \right)^{1 /
2}
\sum\limits_{\{P\}}
\prod\limits_{k = 1}^{N} \varphi_{i_{k}(k)} =
\left( \frac{\prod\limits_{m = 1}^{N} {n_{m}
!}}{N!} \right)^{1 / 2}
\sum\limits_{\{P\,'\}}
\prod\limits_{k = 1}^{N} \varphi _{i_{k} (k)} \\
 \sum\limits_{k = 1}^{N} n_{k}   = N \\
 \end{array}
\end{equation}
<!--  -->
де $\varphi $ - хвильові функції одночастинкових станів, $n_{k}$ - числа заповнення станів, тобто кількість частинок в даному одночастинковому стані. Перша сума у (\ref{19eq5}) відбувається за всіма перестановками, друге - тільки за різними перестановками, тобто перестановкам частинок між різними станами, оскільки перестановка двох частинок в одному і тому ж стані не призводить до нової конфігурації. Тому
у (\ref{19eq5}) з'являється поліноміальний коефіцієнт, який визначає
число сполучень з $N$ елементів по групах (одночастинковим станам), $n_{m} $ елементів у групі з номером $m$. Енергія такого стану:
<!--  -->
\begin{equation}
\label{19eq6} E = \sum\limits_{k = 1}^{N} n_{k} E_{i_{k}}\,\,.
\end{equation}
<!--  -->
Для системи невзаємодіючих ферміонів хвильова функція будується за допомогою, відомої з тензорної алгебри, операції альтернування індексів:
<!--  -->
\begin{equation}\label{19eq7}
 \psi (1,2,\ldots,N) = \left( \frac{1}{N!} \right)^{1/2}
 \varepsilon^{i_{1} ,i_{2} ,\ldots,i_{N}} \varphi_{i_{1}}(1)\varphi_{i_{2}}(2)\ldots\varphi_{i_{N}}(N) =
\end{equation}
<!--  -->
$$= \frac{1}{\sqrt{N!}}\begin{vmatrix}\varphi_{1}(1) & \ldots & \varphi_{1}(N)
\\
\ldots & \ldots & \ldots \\ \varphi_{N}(1) & \ldots
&\varphi_{N}(N)
\end{vmatrix}\,,\,\,\,\,  i_{k} = 1,\ldots, N
$$
<!--  -->
Вираз у \eqref{19eq7}, у вигляді детермінанта, називають __детермінантом Слеттера__. За будовою детермінанта зрозуміло що він дорівнює 0 якщо два ферміона опиняються у однаковому стані (принцип Паулі). Наприклад, основний стан атома гелію наближено мона уявляти як два елекрона у стані $n=1,l=0$ звісно з різними проекціями спіна (принцип Паулі). Відповідна хвильова функція (припускаємо що електрони не взаємодють):
<!--  -->
$$
\Psi = \frac{1}{\sqrt{2}}\left(\psi_{1,0,0,\uparrow}(r_1)\,\psi_{1,0,0,\downarrow}(r_2) -\psi_{1,0,0,\uparrow}(r_2)\,\psi_{1,0,0,\downarrow}(r_1)\right) = \frac{1}{\sqrt{2}}
\begin{vmatrix}\psi_{1,0,0,\uparrow}(r_1) &  \psi_{1,0,0,\uparrow}(r_2)
\\
\psi_{1,0,0,\downarrow}(r_1) &  \psi_{1,0,0,\downarrow}(r_2)
\end{vmatrix}
$$
<!--  -->
Для атома водню врахування спіна призводить до $2\,n^2$ - кратного виродження енеретичних рівнів ($n$ - головне квантове число). Ця обставина визначає механізм заповнення електронних оболоньок у багатоелектронних атомах (періодична система).
