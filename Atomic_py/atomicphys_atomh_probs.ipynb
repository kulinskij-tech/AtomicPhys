__Розвязок 6.91__ Спочатку визначимо сталу нормування $A$:
$$
4\,\pi\,\int\limits_{0}^{+\infty}\,\psi(r)^2\,r^2\,dr = 1 = 4\,\pi\,A^2\,\int \limits_{0}^{+\infty}\,(1+a\,r)^2\,e^{-2\alpha\,r}\,r^2\,dr$$

```python
e,g,r,a,A,b=symbols('e g r a A b',positive=True)
psi = (1+a*r)*exp(-b*r)
A=1/sqrt((4*pi*integrate(psi**2*r**2,[r,0,+oo])))
A
```

Функція повинна задовільняти (радіальному) рівнянню Шредінгера:
$$H\psi = e\,\psi \Rightarrow \frac{1}{r}\,(r\,\psi)'' + \frac{2\,\psi}{r} = -e\,\psi$$

```python
ham=expand((diff((r*psi),r,r)/r +(2/r + e)*psi)*exp(b*r)*r)
ham
```

```python
Poly(ham,r).all_coeffs()
```

```python
nonlinsolve(Poly(ham,r).all_coeffs(),(a,b,e))
```

$$a=-1/2\,,\,\,b = 1/2\,,\,\,\,e=-1/4$$
