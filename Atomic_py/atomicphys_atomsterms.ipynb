## Багатоелектронні атоми
Враховуючи тільки електростатичну взаємодію, гамільтоніан багатоелектронного атома можна записати так:
<!--  -->
\begin{equation}
\label{eq:hamel} H = {\sum\limits_{k = 1}^{Z} {H_{k}}}   +
\frac{e^{\,2}}{2}\,\sum\limits_{i \ne k} \frac{1}{\vert r_{i} - r_{k} \vert}
\end{equation}
<!--  -->
Тут перша сума є сумою одноелектронних гамільтоніанів, друга - електростатична взаємодія електронів. Гамільтоніан містить тільки сферично-симетричні взаємодії, а отже орбітальний момент імпульсу ${\mathbf{L}}$ зберігається так само, як і у випадку атома водню. Зрозуміло, що спін також є інтегралом руху, оскільки \eqref{eq:hamel} не містить спін-залежних взаємодій. На основі такого гамільтоніана
можна ввести поняття про одноелектронні стани за допомогою методу самоузгодженого поля, яке є сумою поля ядра та поля, що створюється всіма електронами. Детально на обчисленні цього поля ми зупинятись не будемо. Головне те, що можна ввести поняття про одноелектронні стани у такій системі, застосувавши результати, отримані раніше для атома водню.  Дійсно, з фізичної точки зору введення одноелектронних станів можливе у полі ядра і інших електронів, яке можна вважати центральним. Оскільки таким електроном може бути обраний довільний, необхідно, щоб дане поле вводилося самоузгодженим чином, враховуючи той факт, що воно само залежить від одноелектронних станів. Енергія електронної конфігурації атома в цьому наближенні є адитивною величиною і складається з енергії заповнених одночастинкових станів в самоузгодженому полі. Заповнення одночастинкових станів визначається принципом Паулі (не більше двох електронів в одному орбітальному стані). Оскільки основний стан — це стан з мінімальною енергією, то при такому заповненні конкурують два основні ефекти: збільшення заряду ядра, який призводить до зниження енергії, і перехід у стан з більш високим квантовим числом, що веде до підвищення енергії. Стани із заданими $n,l$ називають електронною оболонкою. Таким чином, енергія зв'язку додаткового електрона повинна зростати до моменту повного заповнення оболонки, тоді вона різко падає. Така картина заповнення оболонок характерна для всієї послідовності елементів. Всі електронні стани можна розподілити за групами, які послідовно заповнюються. Виконується так зване правило Клечковського–Маделунга: заповнення одночастинкових станів на орбіталях відбувається у порядку зростання суми квантових чисел $n+l$, причому при однаковому значенні $n+l$ найменшу енергію має конфігурація з найменшим $n$ (найближча відстань до ядра).

Стан атома таким чином визначається зазначенням заповнених одноелектронних станів (електронна конфігурація) з використанням символів $s(l=0),p(l=1),d(l=2),f(l=3),\ldots$ та значень квантових чисел повних моментів $L,S,J$ (атомний терм) і літер $S(L=0),P(L=1),D(L=2),F(L=3)\,...$. 
Значення повного спіна $S$ вказується в форматі мультиплетності $2S+1$ ліворуч угорі, а значення повного моменту $J$ праворуч внизу. Найменшу можливу енергію має терм з максимально можливим значенням $S$ за даної конфігурації (незаповнена оболонка) і максимально можливим за даного $S$ значенням $L$ (емпіричне правило Хунда). Наприклад, основний стан атома водню $1s^1\,^{2}S_{1/2}$ атома гелію $1s^2\,^{1}S_{0}$.
Для конфігурації $p^2$ є $С_{6}^{2}=15$ способів зайняти два з шести одноелектронних спін-орбітальних станів. Принцип Паулі допускає терми $^{1}S$, $^{1}D$ і $^{3}P$. Після LS-зв'язування вони дають рівні $^{1}S_0$, $^{1}D_2$ та $^{3}P_{0,1,2}$, разом $1+5+(1+3+5)=15$ станів. Терми $^{1}P$, $^{3}S$ і $^{3}D$ для двох еквівалентних p-електронів заборонені.

Класифікація атомних рівнів енергії проводиться за такими правилами:

\begin{equation}
\label{ateq3} L = 
\left\{
\begin{array}{cccccc}
 0& 1&   2&   3&   4&   5 \\
 S& P&   D&   F&   G&   H \\
 \end{array} 
 \right.
\end{equation}

тощо. Зверху ліворуч  вказується мультиплетність $2S + 1$ (слід, однак, пам'ятати, що вона збігається з числом рівнів тонкої структури тільки при $L \ge S$). Справа внизу вказується значення

## Векторна модель LS-зв'язку без зовнішнього поля

Після [дублетів лужних металів](https://colab.research.google.com/github/kulinskij-tech/AtomicPhys/blob/main/Atomic_py/atomicphys_atom1el.ipynb) та [синглетів і триплетів гелію](https://colab.research.google.com/github/kulinskij-tech/AtomicPhys/blob/main/Atomic_py/atomicphys_atomhe.ipynb) узагальнимо внутрішній зв'язок моментів. У межах одного LS-терму просту модель задає

$$H_{LS}=A\,\frac{\mathbf L\cdot\mathbf S}{\hbar^2},$$

де $A$ має розмірність енергії. Векторна картина зображує $\mathbf L$ та $\mathbf S$ як моменти, що прецесують навколо сталого $\mathbf J=\mathbf L+\mathbf S$; їхні поперечні складові взаємно компенсуються. Без зовнішнього поля немає виділеної лабораторної осі. Це напівкласична ілюстрація квантового додавання моментів, а не буквальні траєкторії електрона чи його спіну.

Для геометрії використовуємо квантові довжини $|\mathbf L|=\hbar\sqrt{L(L+1)}$, $|\mathbf S|=\hbar\sqrt{S(S+1)}$ та $|\mathbf J|=\hbar\sqrt{J(J+1)}$. Із тотожності

$$2\mathbf L\cdot\mathbf S=\mathbf J^2-\mathbf L^2-\mathbf S^2$$

отримуємо в цій моделі

$$\Delta E_J=\frac A2\big[J(J+1)-L(L+1)-S(S+1)\big].$$

Енергія залежить від $J$, але не від $M_J$. Ця формула описує просту модель LS-розщеплення; для точного спектра He потрібно врахувати додаткові внески, зазначені в попередньому розділі.

## Магнітний момент і множник Ланде

Орбітальний і спіновий моменти дають різні магнітні внески. Для електронів

$$\boldsymbol\mu_L=-g_L\mu_B\frac{\mathbf L}{\hbar},\qquad
\boldsymbol\mu_S=-g_S\mu_B\frac{\mathbf S}{\hbar},\qquad g_L=1,\quad g_S\simeq2.$$

Тут $\mu_B>0$ — магнетон Бора; знак мінус відповідає від'ємному заряду електрона. В СІ $\mu_B=|e|\hbar/(2m_e)$, у гаусовій системі — $|e|\hbar/(2m_ec)$. Символи $\boldsymbol\mu_L,\boldsymbol\mu_S$ позначають магнітні моменти; їх не слід плутати з магнітними квантовими числами $M_L,M_S$.

Оскільки $g_L\ne g_S$, миттєва сума $\boldsymbol\mu_L+\boldsymbol\mu_S$ загалом не паралельна $\mathbf J$. При усередненні за внутрішньою прецесією поперечні магнітні складові зникають. Залишається ефективний момент уздовж $\mathbf J$.[^1]

<img src="./figs/fig_lande_vector_model.png" alt="Прецесія L і S навколо J та усереднення магнітного моменту з різними g-факторами" width="900">

*Ліворуч: $\mathbf L+\mathbf S=\mathbf J$. Праворуч: $\boldsymbol\mu$ і його середня за прецесією складова; показано приклад $^{2}P_{3/2}$.*

Проєктуємо обидва моменти на $\mathbf J$:

$$\overline{\boldsymbol\mu}=-\frac{\mu_B}{\hbar}
\left(g_L\frac{\mathbf L\cdot\mathbf J}{\mathbf J^2}
+g_S\frac{\mathbf S\cdot\mathbf J}{\mathbf J^2}\right)\mathbf J
\equiv-g_J\mu_B\frac{\mathbf J}{\hbar}.$$

Із $\mathbf J=\mathbf L+\mathbf S$ маємо

$$2\mathbf L\cdot\mathbf J=\mathbf J^2+\mathbf L^2-\mathbf S^2,\qquad
2\mathbf S\cdot\mathbf J=\mathbf J^2+\mathbf S^2-\mathbf L^2.$$

Підставляючи квантові довжини, для $J>0$ отримуємо

$$g_J=g_L\frac{J(J+1)+L(L+1)-S(S+1)}{2J(J+1)}
+g_S\frac{J(J+1)+S(S+1)-L(L+1)}{2J(J+1)}.$$

За $g_L=1$, $g_S=2$ це **формула Ланде**:

$$g_J=1+\frac{J(J+1)+S(S+1)-L(L+1)}{2J(J+1)}.$$

Отже, $g_J$ — вага орбітального та спінового внесків після проєктування на повний момент, а не нова спін-орбітальна енергія. У квантовій теорії таке проєктування дає той самий результат у межах чистого LS-мультиплету.[^2]

| Рівень | $L$ | $S$ | $J$ | $g_J$ |
|---|---:|---:|---:|---:|
| $^{2}P_{1/2}$ | 1 | $1/2$ | $1/2$ | $2/3$ |
| $^{2}P_{3/2}$ | 1 | $1/2$ | $3/2$ | $4/3$ |
| $^{1}P_1$ | 1 | 0 | 1 | 1 |
| $^{3}S_1$ | 0 | 1 | 1 | 2 |
| $^{3}P_1,{}^{3}P_2$ | 1 | 1 | 1, 2 | $3/2$ |

Для $J=0$ формула з діленням на $J(J+1)$ не застосовується. Ізотропний рівень $J=0$ має нульовий власний векторний магнітний момент; приписувати йому результат ділення $0/0$ не можна.

Отриманий момент використаємо в [розділі про зовнішні поля](https://colab.research.google.com/github/kulinskij-tech/AtomicPhys/blob/main/Atomic_py/atomicphys_atominemfield.ipynb), де внутрішня LS-структура стає основою опису ефекту Зеємана.

## Періодична система елементів з точки зору квантової механіки
<!--  -->
Пояснення природи періодичного закону зміни хімічних властивостей атомів, відкритого Д.І. Менделєєвим, демонструє силу квантової механіки як фізичної теорії, яка адекватно відображує певні закономірності Всесвіту. Задача про спектр атома водню лежить в основі нашого розуміння більш складних багатоелектронних систем. Оскільки електрон є фермі-частинкою, то суттєвим фактом є принцип заборони Паулі, який забороняє двом ферміонам знаходитися в одному стані (фермі-дираківський індивідуалізм). 
Із загальних властивостей стаціонарних станів частинки у центральному полі випливає, що у стані з певним головним квантовим числом (енергією) та орбітальним числом $l$ не може знаходитись більше ніж $2(2l+1)$ електронів. __Розв'язок__ Потужність випромінювання визначається кількістю енергії, випроміненої за одиницю часу. Відповідно це є кількість збуджених атомів у стані $2P$. Використовуючи розподіл Больцмана з урахуванням виродженості відповідних станів ($g_{2P} = 6,\, g_{2S} =2$):
$$\hbar\,\omega\,n\,\frac{g_{2P}}{g_{2S}}\,e^{-\frac{\hbar\,\omega}{k\,T}} = P\,\tau \Rightarrow \tau = \frac{\hbar\,\omega\,n}{P}\,\frac{g_{2P}}{g_{2S}}\,e^{-\frac{\hbar\,\omega}{k\,T}}$$
<!--  -->

```python
import numpy as np # базовий пакет для математичних обчислень та роботи з масивами  
from scipy import constants
type(constants.physical_constants)#type of an object (it's dictionary)
```

```python
pwr=0.3
n=3.6e16
lmd=671e-9
gp=6
gs=2
t=1500
kb=constants.value('Boltzmann constant')
enrg=constants.h*constants.c/lmd
tau=enrg/pwr*n*gp/gs*np.exp(-enrg/(kb*t))
print('tau = ',"{:.2e}".format(tau),'s')
```

__Розв'язок__ Згідно з розподілом Больцмана ймовірності поглинального та випромінювального переходів пов'язані співвідношенням:
<!--  -->
$$\frac{P_{-}}{P_{+}} = e^{-\frac{\hbar\,\omega}{k\,T}}$$
<!--  -->
причому $P_{+} = P^{(s)}_{+}+P^{(ind)}_{+}$ і $P^{(ind)}_{+} = P_{-}$
<!--  -->
$$\frac{P^{(ind)}_{+}}{P^{(s)}_{+}+P^{(ind)}_{+}} = e^{-\frac{\hbar\,\omega}{k\,T}} \Rightarrow \frac{P^{(s)}_{+}}{P^{(ind)}_{+}} = e^{\frac{\hbar\,\omega}{k\,T}}-1$$
<!--  -->


```python
t=3000
e0=constants.value('Rydberg constant times hc in J')
erad=e0*(1-1/4)
prat = 1/(np.exp(erad/(kb*t))-1)
print('pind/ps = ',"{:.2e}".format(prat))
```

Відповідь на останнє питання дає розв'язок рівняння $e^{\frac{\hbar\,\omega}{k\,T}} = 2$:
$$T = \frac{\hbar\,\omega}{k\,\ln{2}}$$
