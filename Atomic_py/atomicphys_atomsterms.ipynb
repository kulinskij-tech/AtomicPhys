Опис багатоелектронного атома є дуже складною задачею внаслідок великої кількості ступенів свободи, які взаємодіють між собою як звичайною електростатичною взаємодією, так і взаємодіями релятивистської природи: спін-орбітальною, спін-спіновою та інш.. ## Тонка структура спектральних ліній водню та лужних металів
<!--  -->
Лужні метали найбільш близьки до атома водню, оскільки мають один слабко звязаний ''зовнішній'' електрон. Це обумовлює їх високу хімічну активність (дуже легко окислюються віддаючи цей електрон). Стаціонарні стани такого воднеподібного електрона можуть бути описані модифікованою формулою Рідберга:
<!--  -->
$$
E_{n,l} = -\frac{E_0}{(n+\Delta_{l})^2}\,,\,\,\,E_0 = \frac{m\,e4}{2\hbar^2}
$$
<!--  -->
тут $\Delta_{l}$ - поправка Рідберга яка залежить від $l$ і типу атома (значень $L,S$), оскільки поле в якому знаходиться електрон не співпадає з чисто кулонівським і відповідно виродження енергії за $l$ як у атома водня не має. Із збільшенням $l$ (за данимиb $L,S$) зрозуміло що $\Delta_l$ зменшується. В атомі водню та лужних металах (та відповідних іонах) зміна стану слабко звязаного електрону дозволяє відслідкувати наявність спінового магнітного моменту електрона. Такий магнітний момент взаємодіє із магнітним полем зовнішнім (ефект Зеємана) або внутрішнім (спін-орбітальна взаємодія, взаємодія із магнітним моментом ядра). Розглянемо вплив спін-орбітальної взаємодії яка обумовлює, т.з. тонке розщеплення спектральних ліній (взаємодія магнітних моментів електрона та ядра в тисячі раз слабкіша і обумовлює надтонке розщеплення яким ми будемо нехтувати). Спін-орбітальна взаємодія має релятивістську природу (так само як і спін) і її можна пояснити наступним чином. З т.з. класиної електродинаміки магнітний диполь що рухається відносно (нерухомого) ядра має також електричний дипольний момент $\mathbf{d} = \mathbf{v}\times \mathbf{m}_{s}/c$. У електричному полі ядра $\mathbf{E}$ відповідно виникає енергія взаємодії:
$$U_{SL} = -\mathbf{d}\cdot\mathbf{E} = -\mathbf{m}_s\cdot (\mathbf{E}\times \mathbf{v}/c)$$
аналогічно, в системі відліку електрона  що рухається відносно ядра зі швидкістю $\mathbf{v}$ зявляється магнітне поле $\mathbf{H} = \mathbf{E}\times \mathbf{v}/c$ (ми нехтуємо поправками o(v/c)). Оскільки $\mathbf{E} \sim \mathbf{r}$ то фактично виникає проекція спіну на напрямок моменту імпульса (звідси і назва взаємодії спін-орбіта), що має тільки два значення $S_{z} = \pm \dfrac{\hbar}{2}$. Додаткова енергія розщеплення відповідно 
$$\Delta E_{SL} = \pm A\,\mu_e$$. 
Відмітимо, що для станів з $l=0$ спін-орбітальна взаємодія відсутня: $s$-стани не розщеплюються.
Тонка структура енергетичних рівнів повністю пояснює особливості спектру випромінювання лужних металів.
Aтоми з одним валентним електроном (слабкозвязаний 1-частиковий стан: водень, лужні метали) можна характеризувати квантовими числами $n,l,m,m_s=\pm 1/2$. Спін-орбітальна взаємодія $\sim \mathbf{l}\cdot\mathbf{s}$ знімає виродження станів з $l\ne 0$ за проекцією спіна (аналогічно магнітному полю - аномальний ефект Зеємана) - такі стани стають спіновими дублетами. Відповідно розщеплюються спектральні лініі однофотонних переходів ($l\to l \pm 1$). ## Багатоелектронні атоми
Враховуючи тільки електростатичну взаємодію гамільтоніан багатоелектронного атома можна записати наступним чином:
<!--  -->
\begin{equation}
\label{eq:hamel} H = {\sum\limits_{k = 1}^{Z} {H_{k}}}   +
\frac{e^{\,2}}{2}\,\sum\limits_{i \ne k} \frac{1}{\vert r_{i} - r_{k} \vert}
\end{equation}
<!--  -->
Тут перша сума є сумою одноелектронних гамільтоніанів, друга - електростатична взаємодія електронів. Гамільтоніан містить тільки сферично-симетричі взаємодії, а отже орбітальний момент імпульсу ${\mathbf{L}}$ зберігається так само і як і у випадку атома водню. Зрозуміло, що спін також є інтегралом руху оскільки \eqref{eq:hamel} є не містить спін-залежних взаємодій. На основі такого гамільтоніану
можна ввести поняття про одноелектронні стани за допомогою методу самоузгодженого поля яке є сумою поля ядра та поля що створюється всіма електронами. Детально на обчисленні цього поля ми зупинятись не будемо. Головне те, що можна ввести поняття про одноелектронні стани у такій системі, застосувавши результати отримані раніше для атома водню.  Дійсно, з фізичної точки зору введення одноелектронних станів можливо у полі ядра і інших електронів яке можна вважати центральним. Оскільки таким електроном може бути обраний довільний, необхідно, що би дане поле вводилося самоузгодженим чином, враховуючи той факт, що воно само залежить від одноелектронних станів. Енергія електронної конфігурації атома, в цьому наближенні, є адитивна величина і складається з енергії заповненних одночастинкових станів в самоузгодженному полі. Заповнення одночастинкових станів визначається принципом Паулі (не більш 2-х електронів в одному орбітальному стані). Оскільки основний стан це стан з мінімальною енергією, то при такому заповнені конкурують два основних ефекта: збільшення заряду ядра, який приводить до к пониження енергії, і перехід у стан з більш високим квантовим числом, що веде до підвищення енергії. Стани із заданими $n,l$ називають електронною оболонкою. Таким чином, енергія зв'язку додаткового електрона повинна зростати до моменту повного заповнення оболонки, тоді вона різко падає. Така картина заповнення оболонок характерна для всієї послідовності елементів. Всі електронні стани можна розподілити по групам які послідовно заповнюються. Виконується т.з. правило Клечковского-Маделунга: заповнення одночастинкових станів на орбіталях відбувається у порядку зростання суми квантових чисел $n+l$, при чому при однаковому значені $n+l$ найменшим за енергією є конфігурація з найменшим $n$ (найближча відстань до ядра).

Стан атома таким чином визачається вказанням заповнених одноелектронних станів (електронна конфігурація) з використанням символів $s(l=0),p(l=1),d(l=2),f(l=3),\ldots$ та значень квантових чисел повних моментів $L,S,J$ (атомний терм) і літер $S(L=0),P(L=1),D(L=2),F(L=3)\,...$. 
Значення повного спіна $S$ вказується в форматі мультиплетності $2S+1$  зліва на горі а значення повного моменту $J$ зправа внизу. Найменшу можливу енергією має терм з максимально можливим значенням $S$ за даної конфігурації (не заповнена оболонка) і максимально можливим при даному $S$ значенні $L$ (емпірічне правило Хунда). Наприклад основний стан атома водню $1s^1\,^{2}S_{1/2}$ атома гелія $1s^2\,^{1}S_{0}$.
Ще для приклада розглянемо можливі терми у випадку 2-х електронів в стані з $l = 1$ (p - орбіталь). Повний орбітальний момент може приймати значення $L = 0,1,2$, а спін $S = 0,1$. Число можливих розміщень електронів по станам дорівнює $C^{3}_{6} = 15$. Одразу можна сказати, що завдяки принципу Паулі терми ${}^{3}S$ і ${}^{3}D$ заборонені (в них хоча б 2 електрони вимушені зайняти однаковий стан). Залишаються терми
${}^{1}P,\,{}^{1}S,\,\,{}^{3}P,\,{}^{1}D$ з числом станів $3,1,9,5$ відповідно. Очевидно, $15 = {}^{1}S + {}^{3}P + {}^{1}D = 1 + 9 + 5$. Тому можливі тільки терми ${}^{1}S,\,\,{}^{3}P,\,
{}^{1}D$.

При врахуванні релятивистських ефектів типа спін-орбітального і спін-спінового взаємодій виродження по $J$ знімається (тонка структура) і залишається тільки (універсальний) закон збереження повного моменту імпульсу $\mathbf{J} = \mathbf{L} + \mathbf{S}$, який характеризує стаціонарні стани своїм квантовим числом $J = \vert L - S\vert \,\,\ldots\,\,L + S$. В більшості випадків такі взаємодії є малими, так що стаціонарні стани можна характеризувати даними $L,S$ ($L\,S$ зв'язок). Це означає, що величина розщеплення менша ніж відстань між різним термами. Таким чином, рівень з даними $L,S$
розщеплюється на:
<!--  -->
$$
 2S + 1\,,\quad L \ge S
$$
 і на
$$
 2L + 1\,,\quad S > L 
$$
<!--  -->
підрівней. Кожний з цих підрівнів залишається виродженим по проекції повного моменту $J_{z}$. Це виродження вже знімається зовнішнім магнітним полем. 
Класифікація атомних рівнів енергії проводиться за наступними правилам:
<!--  -->
\begin{equation}
\label{ateq3} L = 
\left\{
\begin{array}{cccccc}
 0& 1&   2&   3&   4&   5 \\
 S& P&   D&   F&   G&   H \\
 \end{array} 
 \right.
\end{equation}
<!--  -->
і інш. Зверху ліворуч  вказується мультиплетність $2S + 1$ (слід, однак, пам'ятати, що воно співпадає з числом рівнів тонкої структури тільки при $L \ge S)$. Справа внизу вказується значення
$\vert L - S\vert \, \le J \le \,\,L + S$, наприклад ${}^{\,2}S_{1/2} \,,\,{}^{\,2}P_{3 / 2} \,,\,{}^{\,2}P_{\,1 / 2} \,.$

## Періодична система елементів з точки зору квантової механіки
<!--  -->
Пояснення природи періодичного закону зміни хімічних властивостей атомів, відкритого Д.І. Менделєєвим, демонструє силу квантової механіки, як фізичної теорії, яка адекватно відображує певні закономірності Всесвіту. Задача про спектр атома водню лежить в основі нашого розуміння більш складних багатоелектронних систем. Оскільки електрон є фермі-частинкою, то суттєвим фактом є принцип заборони Паулі, який забороняє двом ферміонам знаходитися в одному стані (фермі-дираківський індивідуалізм). 
З загальних властивостей стаіонарних станів частинки у центральному полі витікає що у стані з певним головним квантовим числом (енергією) та орбітальним числом $l$ не може знаходитись більше ніж $2(2l+1)$ електронів. Ця обставина є основою теорії електроних оболонок та періодичного закону (див. нижче).

За цим правилом маємо $S=3/2$ і $L=3$ і $J = 3-3/2 = 3/2$ тобто терм $\vphantom{F}^{4}F_{3/2}$ . <br>

__Розв'язок__ Потужність випромінювання визначається кількістю енергії випромінювання  в одиницю часу. Відповідно це є кількість збуджених атомів у стані $2P$. Використовуючі розподіл Больцмана з врахуванням виродженності відповідних станів ($g_{2P} = 6,\, g_{2S} =2$):
$$\hbar\,\omega\,n\,\frac{g_{2P}}{g_{2S}}\,e^{-\frac{\hbar\,\omega}{k\,T}} = P\,\tau \Rightarrow \tau = \frac{\hbar\,\omega\,n}{P}\,\frac{g_{2P}}{g_{2S}}\,e^{-\frac{\hbar\,\omega}{k\,T}}$$
<!--  -->

```python
import numpy as np # базовий пакет для математичних обчислень та роботи з масивами  
from scipy import constants
type(constants.physical_constants)#type of an object (it's dictionary)
```

```python
pwr=0.3
n=3.6e16
lmd=671e-9
gp=6
gs=2
t=1500
kb=constants.value('Boltzmann constant')
enrg=constants.h*constants.c/lmd
tau=enrg/pwr*n*gp/gs*np.exp(-enrg/(kb*t))
print('tau = ',"{:.2e}".format(tau),'s')
```

__Розв'язок__ Згідно з розподілом Больцмана ймовірності поглинального та випромінювального переходів повязані співвідношенням:
<!--  -->
$$\frac{P_{-}}{P_{+}} = e^{-\frac{\hbar\,\omega}{k\,T}}$$
<!--  -->
при чому $P_{+} = P^{(s)}_{+}+P^{(ind)}_{+}$ і $P^{(ind)}_{+} = P_{-}$
<!--  -->
$$\frac{P^{(ind)}_{+}}{P^{(s)}_{+}+P^{(ind)}_{+}} = e^{-\frac{\hbar\,\omega}{k\,T}} \Rightarrow \frac{P^{(s)}_{+}}{P^{(ind)}_{+}} = e^{\frac{\hbar\,\omega}{k\,T}}-1$$
<!--  -->


```python
t=3000
e0=constants.value('Rydberg constant times hc in J')
erad=e0*(1-1/4)
prat = 1/(np.exp(erad/(kb*t))-1)
print('pind/ps = ',"{:.2e}".format(prat))
```
