__Задача__ (Irodov 5.260) Точкове ізотропне джерело випромінює світло із довжиною хвилі $\lambda = 589\, nm$ з потужністю $P = 10\, W$. Знайдіть
 -  середню густину потоку фотонів на відстані $r = 2\,m$ від джерела
 - відстань від джерела до точки де густина фотонів $n=100\, m^{-3}$.
 
 __Розв'язок__ Густина потоку $j$ це кількість частинок які перетинають одиничний переріз за одиницю часу 
<!--  -->
$$\mathbf{j} = n\,c\,\frac{\mathbf{k}}{|\mathbf{k}|}$$ 
<!--  -->
Зважаючі на ізотропність джерела середня густина потоку на відстані $r$ визначається наступним чином (кожна частинка-фотон має енергію $\varepsilon = h\,\nu$ та швидкість $c$):
 $$I =\varepsilon\,j\, =  \varepsilon\,c\,n$$
 $$ P = 4\pi\,r^2\,I\,\,\,\, \Rightarrow j = \frac{P}{4\pi\,r^2\,h\,\nu}$$
 $$r(n) = \sqrt{\frac{P}{4\pi\,n\,h\,\nu\,c}}$$

```python
# wavelength
lmbd = 589e-9
nu = constants.c/lmbd
print('частота фотону: '"{:.3e}".format(nu),'1/sec')#print in numeric sciformat
r0=2
n1=100
enrg=nu*constants.h #photon enrgy
print('енергія фотону: '"{:.3e}".format(enrg),'J')
power=10
j0=power/(r0**2*4*constants.pi*enrg)
print('густина потоку: '"{:.3e}".format(j0),'1/(m^2 sec)') 
#printing in scientific format with 3 float digits
```

```python
print('відстань = ',"{:.2f}".format(np.sqrt(power/(4*constants.pi*n1*constants.c*enrg))),'m')
#printing with 2 float digits
```

__Задача__ (Irodov 5.262) Лазер випромінив світловий імпульс впродовж $\tau = 0.13\, ms$ з енергією $E = 10 \,J$. Знайти середній тиск такого світлового імпульсу якщо його сфокусувати у пляму діаметром $d = 10^{-6}\,m$ на поверхню з коефіцієнтом відбиття $\rho = 0.5$.
<!--  -->

__Розв'язок__ Тиск визначається силою яка повязана із зміною імпульсу. Останній складається з двох частин: поглинання 
$$\Delta_{-} p = (1-\rho)\,E/c$$
та відбиття
$$\Delta_{+} p = 2\,\rho\,E/c$$
<!--  -->
$$P = F/S =\frac{4}{\pi\,d^2}\, \frac{\Delta_{+} p + \Delta_{-} p}{\tau} = \frac{4}{\pi\,d^2}\, \frac{2\,E\,\rho+E\,(1-\rho)}{c\,\tau} =  \frac{4}{\pi\,d^2}\, \frac{E\,(1+\rho)}{c\,\tau}$$

```python
enrg=10
d=10e-6
rho=0.5
tau = 0.13e-3
print('тиск = ',"{:.3e}".format(4*enrg*(1+rho)/(constants.pi*d**2*constants.c*tau)),'Pa')
```

<!--  -->
__Розв'язок__ Спочатку отримаємо відповідь з простих міркувань розмірності. Очевидно, що з точнітю до числового коефіцієнта сила визначається світловим тиском:
<!--  -->
$$F \simeq \frac{I}{c}\,S_{\perp}$$
<!--  -->
де $S_{\perp}$ - ''середня'' площа перерізу сфери у напрямку потоку, тобто можна очікувати, що за розмірністю $S_{\perp}\simeq \pi\,R^2$. Отже
<!--  -->
$$F \simeq \frac{I}{c}\,\pi\,R^2$$
<!--  -->

```python
i0=0.7/(0.01)**2
r0=0.05
print('сила = ',"{:.3e}".format(i0*constants.pi*r0**2/constants.c),' N')
```

Ця відповідь була би точною якщоб ми мали справу із плоским диском відповідного перерізу. Для отримання точної відповіді треба врахувати те що кожна частина площі знаходиться під певним кутом до потіку.

Позначимо $\mathbf{e}_{\parallel}$ за напрямок розповсюдження хвилі. Дзеркальність поверхні означає що потрібно враховувати тільки нормальну складову зміни імпульсу на елемент площини. Cила визначається як сумарна зміна імпульсу сфери по всій площі за одиницю часу. Звісно що зміна імпульсу сфери залежить від кута нормалі поверхні (див Рис) і при одиночному зіткненні з одиночним фотоном дорівнює
<!--  -->
$$\Delta \mathbf{p} = -2\,(\mathbf{p}\cdot \mathbf{n})\,\mathbf{n}\,,\,\,\,\mathbf{p} = p\,\mathbf{e}_{\parallel} $$
<!--  -->
Враховуючі, що за час $\Delta t$ на елемент площі падає $\frac{I}{p\,c}\,\Delta t\, \cdot |\mathbf{e}_{\parallel}\cdot\mathbf{n}|\,dS$ таких фотонів можна записати, що
$$d\mathbf{F} = -2\,(\mathbf{p}\cdot \mathbf{n})\,\mathbf{n}\frac{I}{p\,c}\, |\mathbf{e}_{\parallel}\cdot\mathbf{n}|\,dS = - 2\frac{I}{c}\,(\mathbf{e}_{\parallel}\cdot\mathbf{n})^2\,\mathbf{n}\,dS$$
<!--  -->
Завдяки симетрії сфери сумарна сила буде мати тільки складову вздовж напрямку $\mathbf{e}_{\parallel}$  
***
<img src="./figs/atphys_prob2_1.png" alt="Image" width="200">

***
$$F_{\parallel} = \mathbf{F}\cdot \mathbf{e}_{\parallel} = -2\,\frac{I}{c}\,\int (\mathbf{n}\cdot \mathbf{e}_{\parallel})^3\, dS = -4\pi\,R^2\,\frac{I}{c}\,\int\limits_{0}^{\pi/2} \cos^3{\theta}\,\sin{\theta}\,d\theta = 4\pi\,R^2\,\frac{I}{c}\,\int\limits_{0}^{1} x^3\,dx = \pi\,R^2\,\frac{I}{c}$$

Як ми бачимо, ми отримали такий ж самий вираз як і раніше. Це ''випадковий'' збіг обумовлений компенсацією числових множників. Якщо б потік був нерівномірний, то ми б отримали інший числовий коефіцієнт.

E = \sqrt{p^2\,c^2+E^2_0}\,,\,\,\, 
E_0 = m_0\,c^2$$

які взаємодіють із високоенергетичними фотонами

__Розв'язок__ Використовуючі закони збереження енергії та імпульсу:
<!--  -->
$$
E = \gamma\,m_0\,c^2\,,\,\,\,E/c = \gamma\,m_0\,v\,
$$
<!--  -->
і оскільки $v<c$ то потрібний результат доведено. Саме ця обставина лежить в основі Комптон-ефекту.

<a id='comptonprob'>__Задача(Комптон-ефект):__</a> Обчислити зсув довжини хвилі при розсіянні фотона, який розглядається як частинка із певним імпульсом та енергією, на вільному електроні (зсув Комптона).<br>

__Розв'язок__  Запишемо закони збереження імпульсу та енергії (з урахуванням релятивістського характеру руху електрона)

$$
\label{compton}
\hbar\,\omega_0 + E_0 = \sqrt{(p\,c)^2+E^2_{0}}+\hbar\,\omega \\
\hbar \, \mathbf{k}_0 = \mathbf{p} + \hbar \, \mathbf{k}
$$

$E_0 = m_0\,c^2$ - енергія спокою. Проводячи необхідні підстановки:

$$
\hbar^2 (\omega_0 - \omega)^2 + 2\,m_0\,c^2\,\hbar(\omega_0 - \omega) = p^2\,c^2\,,\quad
p^2c^2 = \hbar^2(\omega^2-2\omega\,\omega_0\cos \theta+\omega^2_0)
$$
отримуємо співвідношення частот:
$$
\frac{\hbar}{m_0\,c^2}\,\omega\,\omega_0\,(1-\cos \theta) = (\omega - \omega_0)
$$
в терминах довжини хвилі $\omega = c\frac{2\pi}{\lambda}$ це дає:
$$
\lambda - \lambda_0 = 2\,\Lambda_{C}\,\sin^2\frac{\theta}{2}\,\,.
$$
Характеристична довжина $\Lambda_{C} = \frac{2\pi\,\hbar}{m_0\,c}$ називається комптонівською довжиною хвилі, що для електрона складає $\Lambda_{C}\approx 2.43\cdot 10^{-12}$ м.

```python
# Compton length
compl=constants.h/(constants.m_e*constants.c)
print(compl)
print(constants.find('Compton'))
constants.physical_constants[constants.find('Compton')[0]]
```

Бачимо, що ця довжина відповідає енергії фотона яка дорівнює енергії спокою електрона $E_{0}=m_{0}\,c^2$. Вона менша за розмір атома (боровський радіус):

```python
constants.value('Bohr radius')/constants.value('Compton wavelength')
```

і обмежує точність визначення координати електрона з причини виникнення суто релятивістських квантових ефектів на малих масштабах (народження електрон-позитроних пар). Як витікає з формули, комптонівський зсув спостерігається при прямому куті розсіяння $\theta = \pi/2$.

### Задачі на чорне випромінювання
<!--  -->
__Задача__ Закон Віна $\omega_{max}\sim T$ дозволив йому сформулювати гіпотезу про те що середня енергія ___однієї___ моди ЕМВ з частотою $\omega$ 

$$U_{\omega} = u_0\,e^{-\frac{u_0}{T}}\,,\quad u_0 \sim \omega$$

що звичайно може розлядатись як твердження про те що енергія моди пропорційна до частоти і відповідним розподілом є розподіл Больцмана. Використвуючи термодинамічну рівність 

$$\left.\frac{dS}{dU}\right\vert_{V} = \frac{1}{T}$$ 

та просту інтерполяційну формулу Планка для ентропії

$$
\frac{d^2 S}{d U^2} =-\frac{a}{U(U+b)}
$$

виведіть формулу Планка для спектральної густини енергії.
<br>
__Рішення__ Нагадаємо, що кількість мод ЕМП в інтервалі частот $d\omega$ у кубічному резонаторі обєма $V = L^3$ дорівнює:
$$dN_{\omega} = 2\,\left(\frac{L}{2\pi}\right)^3\,4\pi\,k^2\,dk = \frac{V}{\pi^2\,c^3}\,\omega^2\,d\omega
$$
тому гіпотеза Віна призводила до спектральної густини енергії
$$
\rho_{\omega} = \frac{1}{V}\frac{d E_{\omega}}{d\omega} = \frac{1}{V}\frac{dN_{\omega}}{d\omega}\,U_{\omega} \sim \frac{\omega^3}{\pi^2\,c^3}\,e^{-\frac{\omega}{T}}
$$
яка добре узгоджувалась із даними для великих частот, але не збігалась із класичним законом Релея-Джинса для малих $\omega: \rho_{\omega}  = \frac{T}{\pi^2\,c^3}\,\omega^2$. Виходячи з гіпотези Віна отримаємо
$$U_{\omega} = u_0\,e^{-\frac{u_0}{T}}\Rightarrow \frac{1}{T} = -\frac{1}{u_0}\,\ln \frac{U_{\omega}}{u_0}$$
При великих значеннях $T$ значення $U$ також будуть великі (максимум Віна зсувається в область великих частот) а отже в інтервалі частот $\omega \ll T/\hbar$, очевидно, повинна діяти класична формула з $U\sim T$, тому між двома граничними випадками
$$ 
\frac{d}{d\,U}\left(\frac{1}{T}\right)\sim \begin{cases}
 -\frac{1}{U^2}\,,& T\gg u_0\,\,\,\, \text{Rayleigh–Jeans} \\
 -\frac{1}{u_0\,U}\,,& T\ll u_0\,\,\,\text{Wien}
\end{cases}
$$ 
зробимо найпростішу інтерполяцію:
$$
\frac{d^2 S}{d U^2} =-\frac{1}{U(U+u_0)}\Rightarrow \frac{1}{T} = \frac{1}{u_0}\,\ln \frac{U+u_0}{U}\Rightarrow U_{\omega} = \frac{u_0}{e^{\frac{u_0}{T}}-1}
$$
Остаточно якщо позначити $u_0 = \hbar\,\omega$ отримаємо:
$$
\rho_{\omega} = \frac{\omega^2}{\pi^2\,c^3} \frac{\hbar\,\omega}{e^{\frac{\hbar\,\omega}{T}}-1}
$$
це і є знаменитий закон Планка.
