# Тонка структура спектральних ліній водню та лужних металів
<!--  -->
Лужні метали найбільш близьки до атома водню, оскільки мають один слабко звязаний ''зовнішній'' електрон. Це обумовлює їх високу хімічну активність (дуже легко окислюються віддаючи цей електрон). Стаціонарні стани такого воднеподібного електрона можуть бути описані модифікованою формулою Рідберга:
<!--  -->
$$
E_{n,l} = -\frac{E_0}{(n+\Delta_{l})^2}\,,\,\,\,E_0 = \frac{m\,e4}{2\hbar^2}
$$
<!--  -->
тут $\Delta_{l}$ - поправка Рідберга яка залежить від $l$ і трему атома (значень $L,S$), оскільки поле в якому знаходиться електрон не співпадає з чисто кулонівським і відповідно виродження енергії за $l$ як у атома водня не має. Із збільшенням $l$ (за данимиb $L,S$) зрозуміло що $\Delta_l$ зменшується. В атомі водню та лужних металах (та відповідних іонах) зміна стану слабко звязаного електрону дозволяє відслідкувати наявність спінового магнітного моменту електрона. Такий магнітний момент взаємодіє із магнітним полем зовнішнім (ефект Зеємана) або внутрішнім (спін-орбітальна взаємодія, взаємодія із магнітним моментом ядра). Розглянемо вплив спін-орбітальної взаємодії яка обумовлює, т.з. __тонке розщеплення__ спектральних ліній (взаємодія магнітних моментів електрона та ядра в тисячі раз слабкіша і обумовлює __надтонке розщеплення__ яким ми будемо нехтувати). Спін-орбітальна взаємодія має релятивістську природу (так само як і спін) і її можна пояснити наступним чином. З т.з. класиної електродинаміки магнітний диполь що рухається відносно (нерухомого) ядра має також електричний дипольний момент $\mathbf{d} = \mathbf{v}\times \mathbf{m}_{s}/c$. У електричному полі ядра $\mathbf{E}$ відповідно виникає енергія взаємодії:
$$U_{SL} = -\mathbf{d}\cdot\mathbf{E} = -\mathbf{m}_s\cdot (\mathbf{E}\times \mathbf{v}/c)$$
аналогічно, в системі відліку електрона  що рухається відносно ядра зі швидкістю $\mathbf{v}$ зявляється магнітне поле $\mathbf{H} = \mathbf{E}\times \mathbf{v}/c$ (ми нехтуємо поправками o(v/c)). Оскільки $\mathbf{E} \sim \mathbf{r}$ то фактично виникає проекція спіну на напрямок моменту імпульса (звідси і назва взаємодії спін-орбіта), що має тільки два значення $S_{z} = \pm \dfrac{\hbar}{2}$. Додаткова енергія розщеплення відповідно 
$$\Delta E_{SL} = \pm A\,\mu_e$$. 
Відмітимо, що для станів з $l=0$ спін-орбітальна взаємодія відсутня: $s$-стани не розщеплюються.
Тонка структура енергетичних рівнів повністю пояснює особливості спектру випромінювання лужних металів.
Aтоми з одним валентним електроном (слабкозвязаний 1-частиковий стан: водень, лужні метали) можна характеризувати квантовими числами $n,l,m,m_s=\pm 1/2$. Спін-орбітальна взаємодія $\sim \mathbf{l}\cdot\mathbf{s}$ знімає виродження станів з $l\ne 0$ за проекцією спіна (аналогічно магнітному полю - аномальний ефект Зеємана) - такі стани стають спіновими дублетами. Відповідно розщеплюються спектральні лініі однофотонних переходів ($l\to l \pm 1$). У випадку атому водня, наприклад, перехід $2p\to 1s$ серії Лаймана розщеплюється на дві лінії: $2p_{1/2} \to 1s_{1/2}$, $2p_{3/2} \to 1s_{1/2}$. __Розв'язок__ Потужність випромінювання визначається кількістю енергії випромінювання  в одиницю часу. Відповідно це є кількість збуджених атомів у стані $2P$. Використовуючі розподіл Больцмана з врахуванням виродженності відповідних станів ($g_{2P} = 6,\, g_{2S} =2$):
$$\hbar\,\omega\,n\,\frac{g_{2P}}{g_{2S}}\,e^{-\frac{\hbar\,\omega}{k\,T}} = P\,\tau \Rightarrow \tau = \frac{\hbar\,\omega\,n}{P}\,\frac{g_{2P}}{g_{2S}}\,e^{-\frac{\hbar\,\omega}{k\,T}}$$
<!--  -->

```python
import numpy as np # базовий пакет для математичних обчислень та роботи з масивами  
from scipy import constants
type(constants.physical_constants)#type of an object (it's dictionary)
```

```python
pwr=0.3
n=3.6e16
lmd=671e-9
gp=6
gs=2
t=1500
kb=constants.value('Boltzmann constant')
enrg=constants.h*constants.c/lmd
tau=enrg/pwr*n*gp/gs*np.exp(-enrg/(kb*t))
print('tau = ',"{:.2e}".format(tau),'s')
```

__Розв'язок__ Згідно з розподілом Больцмана ймовірності поглинального та випромінювального переходів повязані співвідношенням:
<!--  -->
$$\frac{P_{-}}{P_{+}} = e^{-\frac{\hbar\,\omega}{k\,T}}$$
<!--  -->
при чому $P_{+} = P^{(s)}_{+}+P^{(ind)}_{+}$ і $P^{(ind)}_{+} = P_{-}$
<!--  -->
$$\frac{P^{(ind)}_{+}}{P^{(s)}_{+}+P^{(ind)}_{+}} = e^{-\frac{\hbar\,\omega}{k\,T}} \Rightarrow \frac{P^{(s)}_{+}}{P^{(ind)}_{+}} = e^{\frac{\hbar\,\omega}{k\,T}}-1$$
<!--  -->


```python
t=3000
e0=constants.value('Rydberg constant times hc in J')
erad=e0*(1-1/4)
prat = 1/(np.exp(erad/(kb*t))-1)
print('pind/ps = ',"{:.2e}".format(prat))
```

відповідь на останнє питання зрозуміло це розв'язок рівняння $e^{\frac{\hbar\,\omega}{k\,T}} = 2$:
$$T = \frac{\hbar\,\omega}{k\,\ln{2}}$$
