# Спін-орбітальна взаємодія: водень і лужні метали

## Один валентний електрон

У лужних металів заповнені внутрішні оболонки мають сумарні $L=S=0$, а спектр низьких збуджених станів визначається переважно одним зовнішнім електроном. Тому водень і лужні метали зручно розглядати разом, хоча поле атомного остова не є чисто кулонівським. Для лужного металу енергію наближено записують як

$$E_{n,l}\simeq-\frac{E_0}{(n-\delta_l)^2},\qquad E_0\simeq13.6\,\text{еВ}.$$

Тут $\delta_l$ — квантовий дефект, що враховує відмінність поля остова від кулонівського; для водню в нерелятивістському наближенні $\delta_l=0$. Це електростатична поправка, яку слід відрізняти від тонкої структури.

## Внутрішня взаємодія без зовнішнього поля

Електрон має орбітальний момент $\mathbf l$ і власний спін $\mathbf s$. У його миттєвій системі відліку електричне поле ядра й остова також має магнітну складову. Взаємодія з нею спінового магнітного моменту дає якісне пояснення спін-орбітального зв'язку. Для центрального поля відповідний внесок має структуру

$$H_{SO}=\xi(r)\,\frac{\mathbf l\cdot\mathbf s}{\hbar^2}.$$

На цьому етапі важлива залежність енергії від взаємної орієнтації моментів, а не обчислення $\xi(r)$. Релятивістський вираз із поправкою Томаса наведено в додатку наприкінці цього розділу.

Внутрішнє ефективне поле пов'язане з орбітальним рухом; воно не задає нерухомої лабораторної осі. У зв'язаному стані зручно користуватися повним моментом $\mathbf j=\mathbf l+\mathbf s$ і числами $j,m_j$, а не окремими проекціями $m_l,m_s$. Оскільки $s=1/2$, для $l\ge1$ можливі два значення

$$j=l-\frac12,\qquad j=l+\frac12.$$

Так виникають два рівні одного дублетного терму: наприклад, $^{2}P_{1/2}$ і $^{2}P_{3/2}$. Для $l=0$ є лише $^{2}S_{1/2}$: спін-орбітальний внесок не розщеплює цей рівень. Відсутність такого розщеплення не робить стан синглетним — його спінова мультиплетність залишається $2s+1=2$.[^1]

Без зовнішнього поля кожний рівень із заданим $j$ залишається $(2j+1)$-кратно виродженим за $m_j$. Отже, тонка структура розділяє **різні $j$**, а не проекції на обрану вісь.

__Розв'язок__ Потужність випромінювання визначається кількістю енергії, випроміненої за одиницю часу. Відповідно це є кількість збуджених атомів у стані $2P$. Використовуючи розподіл Больцмана з урахуванням виродженості відповідних станів ($g_{2P} = 6,\, g_{2S} =2$):
$$\hbar\,\omega\,n\,\frac{g_{2P}}{g_{2S}}\,e^{-\frac{\hbar\,\omega}{k\,T}} = P\,\tau \Rightarrow \tau = \frac{\hbar\,\omega\,n}{P}\,\frac{g_{2P}}{g_{2S}}\,e^{-\frac{\hbar\,\omega}{k\,T}}$$
<!--  -->

```python
import numpy as np # базовий пакет для математичних обчислень та роботи з масивами  
from scipy import constants
type(constants.physical_constants)#type of an object (it's dictionary)
```

```python
pwr=0.3
n=3.6e16
lmd=671e-9
gp=6
gs=2
t=1500
kb=constants.value('Boltzmann constant')
enrg=constants.h*constants.c/lmd
tau=enrg/pwr*n*gp/gs*np.exp(-enrg/(kb*t))
print('tau = ',"{:.2e}".format(tau),'s')
```

<img src="./figs/atphys_atomradiat_probs2.png" alt="Image" width="600">

__Розв'язок__ Згідно з розподілом Больцмана ймовірності поглинального та випромінювального переходів пов'язані співвідношенням:
<!--  -->
$$\frac{P_{-}}{P_{+}} = e^{-\frac{\hbar\,\omega}{k\,T}}$$
<!--  -->
причому $P_{+} = P^{(s)}_{+}+P^{(ind)}_{+}$ і $P^{(ind)}_{+} = P_{-}$
<!--  -->
$$\frac{P^{(ind)}_{+}}{P^{(s)}_{+}+P^{(ind)}_{+}} = e^{-\frac{\hbar\,\omega}{k\,T}} \Rightarrow \frac{P^{(s)}_{+}}{P^{(ind)}_{+}} = e^{\frac{\hbar\,\omega}{k\,T}}-1$$
<!--  -->


```python
t=3000
e0=constants.value('Rydberg constant times hc in J')
erad=e0*(1-1/4)
prat = 1/(np.exp(erad/(kb*t))-1)
print('pind/ps = ',"{:.2e}".format(prat))
```

Відповідь на останнє питання дає розв'язок рівняння $e^{\frac{\hbar\,\omega}{k\,T}} = 2$:
$$T = \frac{\hbar\,\omega}{k\,\ln{2}}$$

## Додаток: звідки береться внутрішній спін-орбітальний внесок

Для електрона в центральній потенціальній енергії $U(r)$ провідний нерелятивістський спін-орбітальний внесок із урахуванням прецесії Томаса має вигляд[^2]

$$H_{SO}=\frac{1}{2m_e^2c^2r}\frac{dU}{dr}\,\mathbf L\cdot\mathbf S.$$

Він виникає вже за відсутності зовнішнього магнітного поля; множник $1/2$ враховує поправку Томаса. Тут $U$ — потенціальна **енергія**, а не електричний потенціал, що усуває неоднозначність знака заряду. Для водню інші релятивістські внески також потрібні для повної тонкої структури; для He одноелектронний вираз не охоплює всіх двоелектронних поправок.


[^2]: University of Virginia. [Thomas precession](https://galileoandeinstein.phys.virginia.edu/Elec_Mag/2022_Lectures/EM_68_Thomas_Precession.html).
