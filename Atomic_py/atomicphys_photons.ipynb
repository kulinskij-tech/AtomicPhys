## Основні експериментальні факти
### Теплове випромінювання
(Нерелятивістська) Квантова теорія це інструмент аналізу явищ мікросвіту - атомів, молекул, властивостей елементарних частинок там, де характерна енергія руху набагато менша, ніж енергія спокою $E_0 = m_0\,c^2$, тобто характерні швидкості менші за швидкість світла $c$. Вона також дає нову основу для розуміння загальних закономірностей Природи, зокрема пояснює важливу роль поняття симетрії та яким чином вона пояснює структуру матерії (див [кваркова структура адронів](https://en.wikipedia.org/wiki/Quark)).

Історично нерелятивістська квантова теорія - квантова механіка -  виникла при намаганні розв'язати фундаментальну проблему встановлення рівноваги при випромінюванні та поглинанні світла (М. Планк) речовиною. Ключом до вирішення цієї проблеми було припущення про дискретний характер енергетичного обміну між електромагнитним полем (ЕМП) та атомами речовини. Таким чином, ми бачимо, як розуміння природи світла (ЕМП)
призвело до фундаментальних зрушень основ фізики на початку XX століття. Гіпотеза про дискретний характер обміну енергією між матерією та рівноважним випромінюванням порціями чи квантами величини $\varepsilon =\hbar\, \omega$ спочатку розглядалась як суто математичний трюк (очікувалось, що при переході $\varepsilon = \hbar\, \omega \to 0$ розбіжність зникне), який дозволяв би коректно розглянути проблему розбіжності повної енергії рівноважного випромінювання, до якої веде класична формула Релєя-Джинса для енергетичної густини на одиничний інтервал частот:
<!--  -->
\begin{equation}
\label{eq:rj}
\rho(\omega)=\frac{1}{V}\frac{d\,E}{d\,\omega}= U_{\omega}\, \frac{d\,N_{\omega}}{d\,\omega} =
4\frac{k_\,T}{2}\frac{4\,\pi\,k^2 \,d\,k}{(2\,\pi)^3\,d\,\omega}=
\frac{\omega^{\,2}}{\pi^2\,c^{3}}\,T\,.
\end{equation}
<!--  -->

Суттєво, що ця формула заснована на припущенні про можливість як завгодно малого обміну енергією між середовищем та випромінюванням, а також рівного розподілу енергії за степенями свободи (на кожну степінь свободи у стані теплової рівноваги приходиться енергія $T/2$).
Закон розподілу густини енергії за частотами, отриманий М. Планком, приводив до певної залежності енергетичної густини випромінювання даної частоти
<!--  -->
<img src="./figs/fig_plank1.png" alt="Image" width="600">
<!--  -->
\begin{equation}
\label{eq:plank1}
\rho (\omega ,T) = \frac{\omega^{2}}{\pi^{\,2}\,c^{3}}
\frac{\hbar\,\omega}{\exp \left(\frac{\hbar\,\omega}{T} \right) - 1}\,.
\end{equation}
<!--  -->
Цей вираз досить точно описував експериментальні дані, і при $\hbar~\to~0$ давав класичний закон \eqref{eq:plank1}. <br>

### Фотоефект
Ідея дискретного обміну енергією в індивідуальних (а не тільки статистичних) явищах мікросвіту повною мірою "запрацювала" у роботах А. Айнштайна з
фотоефекту - виникнення струму електронів при опромінюванні металів світлом ультрафіолетового спектру. Це явище є квантовим феноменом який обумовлений квантовою структурою світла яке складається з окремих фотонів. Відповдно, низка експериментальних закономрностей не може бути пояснена з класичної точки зору (світло як неперервне електромагнітне поле у класичній електродинаміці Максвелла). Нагадаємо основні закономірності фотоефекту:
<!--  -->
1. Існування граничної частоти ("червоної" границі)
1. Кінетична енергія електронів розподілена від 0 до граничної $E_{max}$ яка не залежить від інтенсивності падаючого світла і визначається тільки його частотою.
1. При фіксованій частоті кількість фотоелектронів у одиницю часу пропорційна до інтенсивності світла
1. Фотострум починається одразу при опромінюванні без якоїсь затримки у часі (безінерційність фотоефекту)
<!--  -->
Фотоелектричний ефект фактично вводить поріг частоти та енергію фотона $E=h\nu$;

<!--  -->

\end{equation}

тут $W_0$ - т.з. робота виходу що визначає "червону" границю і залежить від матеріалу. Формула Айнштайна \eqref{eq:photoeff} дає логічне пояснення всім зазначеним закономірностям.

### <a href='https://en.wikipedia.org/wiki/Compton_scattering'>Комптон ефект</a>
Квантова природа світла проявляється у розсіянні на вільних чи слабкозв'язаних електронах і є свідченням наявності у світла корпускулярних властивостей. Сутність цього ефекту що характерний для дуже коротких рентгенівських ''хвиль'' ($\lambda \sim 1 \overset{\circ}{A}$). Рентгенівськи хвилі можуть взаємодіяти з речовиною на атомному рівні (наприклад, фотоелектричний ефект та розсіювання Релея), на рівні ядра або лише з електроном. Утворення пар та ефект Комптона відбуваються на рівні електрона. Коли високочастотний фотон розсіюється внаслідок взаємодії із зарядженою частинкою, енергія фотона зменшується, і таким чином його довжина хвилі збільшується. Цей компроміс між довжиною хвилі та енергією у відповідь на зіткнення називається ефектом Комптона. Завдяки закону збереження енергії енергія, яку втрачає фотон, передається частинці, що відбивається (такий електрон називатиметься «комптонівським електроном віддачі»). Як свідчать експериментальні рентгенівські проміні розсіюються на вільних електронах подібно до потоку частинок з імпульсом $p=E/c$ з відповідним зменшенням імпульсу (збільшенням довжини хвилі). Такий характер розсіяння явно протирічить хвильової природі світла яка базується на його представленні як коливаннях ЕМП. За такими класичними уявленнями не повинно відбуватись зміни довжини хвилі, оскільки коливання електрону це вимушені коливання які відбуваються з тією ж самою ча'стотою що й частота падаючого світла (зовнішня сила). При зсуві частоти, наприклад до оптичного спектру, звісно розсіяння відбувається без зміни довжини хвилі (частоти). Це говорить про перехід від квантового до класичного електродинамічного (хвильового) характеру взаємодії ЕМП із речовиною коли довжина хвилі значно більша за характерні розміри розсіювачів (атомів, молекул). Точніше, класична теорія електромагнітної хвилі, розсіяної зарядженими частинками, не може пояснити зсув довжини хвилі при низькій інтенсивності: класичне світло достатньої інтенсивності, щоб електричне поле прискорило заряджену частинку до релятивістської швидкості, викличе віддачу радіаційного тиску та пов'язаний з цим доплерівський зсув розсіяного світла але ефект стане як завгодно малим при достатньо низьких інтенсивностях світла незалежно від довжини хвилі. Таким чином, якщо ми хочемо пояснити комптонівське розсіювання низької інтенсивності, світло повинно поводитися так, ніби воно складається з частинок. Або припущення, що електрон можна розглядати як вільний, є недійсним, що призводить до фактично нескінченної маси електрона, що дорівнює масі ядра. Експеримент Комптона переконливо доводить, що світло можна розглядати як потік частиноподібних об'єктів (квантів, що називаються фотонами), енергія яких пропорційна частоті. 
Якщо енергія звязку електрона з ядром велика (що характерно для електронів внутрішніх оболонок) то розсіяння на таких електронах відбувається майже пружньо, практично без зсуву частоти (довжини хвилі).
Зазначимо, що у оригінальних дослідах Комптона спостерігався тільки інтегральний результат - сумарний світлопотік. Індивідуальні ж акти розсіяння окремих фотонів на окремих електронах було спостережено у дослідах Боте-Вільсона у 1923р.

Обчислимо зсув довжини хвилі при розсіянні фотона, який розглядається як частинка із певним імпульсом та енергією, на вільному електроні (зсув Комптона).<br>

Запишемо закони збереження імпульсу та енергії (з урахуванням релятивістського характеру руху електрона)

$$
\label{compton}
\hbar\,\omega_0 + E_0 = \sqrt{(p\,c)^2+E^2_{0}}+\hbar\,\omega \\
\hbar \, \mathbf{k}_0 = \mathbf{p} + \hbar \, \mathbf{k}
$$

$E_0 = m_0\,c^2$ - енергія спокою. Проводячи необхідні підстановки:

$$
\hbar^2 (\omega_0 - \omega)^2 + 2\,m_0\,c^2\,\hbar(\omega_0 - \omega) = p^2\,c^2\,,\quad
p^2c^2 = \hbar^2(\omega^2-2\omega\,\omega_0\cos \theta+\omega^2_0)
$$
отримуємо співвідношення частот:
$$
\frac{\hbar}{m_0\,c^2}\,\omega\,\omega_0\,(1-\cos \theta) = (\omega - \omega_0)
$$
в терминах довжини хвилі $\omega = c\frac{2\pi}{\lambda}$ це дає:
$$
\lambda - \lambda_0 = 2\,\Lambda_{C}\,\sin^2\frac{\theta}{2}\,\,.
$$
Характеристична довжина $\Lambda_{C} = \frac{2\pi\,\hbar}{m_0\,c}$ називається комптонівською довжиною хвилі, що для електрона складає $\Lambda_{C}\approx 2.43\cdot 10^{-12}$ м.
З Комптон-ефекту витікає що імпульс фотона $p=h/\lambda$ через вимірюваний зсув довжини хвилі.
