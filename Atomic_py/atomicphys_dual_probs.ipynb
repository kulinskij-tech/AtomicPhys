__Задача__ На яку кінетичну енергію слід розраховувати електрон-електроний (і протоний) прискорювач для дослідження структур з лінійнимі розмірами 1 фермі ($l \simeq 10^{-15}$ м)?

__Розв'язок__ У зв'язку з дуже малими відстанями енергії будуть великими у порівнянні з енергію спокою. тому
$$
E = m\,c^2\,\sqrt{1+\left(\frac{p}{m\,c}\right)^2} - m\,c^2\,,\,\,p=h/l \\
E = m\,c^2\,\left(\sqrt{1+\left(\frac{\lambda_{C}}{l}\right)^2} - 1\right)
$$

```python
l=1e-15
le=(constants.value('Compton wavelength')/l)
ene=constants.value('electron mass energy equivalent in MeV')*(np.sqrt(le**2+1)-1)
# 
print("Energy electron = ","{:.2}".format(ene/1e3),'GeV')
# 
lp=(constants.value('proton Compton wavelength')/l)
enp=constants.value('proton mass energy equivalent in MeV')*(np.sqrt(lp**2+1)-1)
print("Energy proton = ","{:.2}".format(enp),'MeV')
# 
```

__Задача__ У дослідах з дифракції електронів на полікристалічній фользі знайдено, що діаметр дифракційного кільця, що відповідає відображенню першого порядку від площин міжплощинною відстанню $d$, дорівнює $D = 3\cdot 10^{-2}$ м. Відстань від фольги до екрану $l = 0.15$ м. Знайти $d$. Енергія електронів 200 еВ.

```python
r=0.03/2
l=0.15
e0=200*constants.value('elementary charge')
p0=np.sqrt(2*constants.value('electron mass')*e0)
si=np.sin(np.arctan(r/l)/2)
d=constants.value('Planck constant')/(p0*si)
print('d = ',"{:.2e}".format(d),'m')
```
