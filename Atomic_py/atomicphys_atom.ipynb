## Задачі на модель Томсона
<!--  -->
__Задача__ Вичначити характерну частоту коливань електрона у атомі водню в рамках моделі Томсона і відповдну довжину хвилі ЕМВ.
<!--  -->
__Розв'язок__ У моделі Томсона електрон знаходиться у рівномірно зарядженій кулі заряду $+e$. Рівняння руху електрона враховує, що на нього діє поле  відповідної частини заряду (використовуємо систему СГС):
<!--  -->
$$\mathbf{F} =  -e\,\frac{e\,r^3}{R^3}\frac{\mathbf{r}}{r^3}\,\\
m\,\ddot{\mathbf{r}} = -\frac{e^2}{R^3}\,\mathbf{r}\Rightarrow  \ddot{\mathbf{r}} = -\omega^2\,\mathbf{r}\,,\,\,\,\omega = \sqrt{\frac{e^2}{m\,R^3}}$$

```python
import numpy as np # базовий пакет для математичних обчислень та роботи з масивами  
from scipy import constants
type(constants.physical_constants)#type of an object (it's dictionary)
massing=constants.value('electron mass')*1e3 # mass in CGS
elchcgs=constants.e*3e9 # electric charge in CGS
radiusa=1e-8
freq = elchcgs/radiusa*1/np.sqrt(massing*radiusa)/(2*constants.pi)
lmbd=constants.c/freq
print('freq = ',"{:.2e}".format(freq),'Hz')
print('wavelength = ',"{:.1f}".format(lmbd*1e9),'nm')
```

відповідне значення енергії $E = h\,\nu$ також доволі близько до значення $E_0 = 13.6\,eV$

```python
print('energy = ',"{:.1f}".format(constants.value('Planck constant in eV/Hz')*freq),'eV')
```

__Розв'язок__ Потенціал іонізації це енергія по виходу електрона з атома Томсона на нескінченність. Вона складається з енергії у пружньому "внутрішньому" полі (див. попередню задачу) $0<r<R$ та енерії виходу на нескінченність $R<r<+\infty$ у полі сферичного розподілу заряда $+e$:
<!--  -->
$$E_0 = \frac{m\,\omega^2\,R^2}{2}+\frac{e^2}{R} =\frac{3}{2}\,\frac{e^2}{R} \Rightarrow R = \frac{3}{2}\,\frac{e^2}{E_0}$$

```python
e0=constants.value('Rydberg constant times hc in J')*1e7 #in cgs unit erg
e0
```

```python
rt=3/2*(constants.e*3e9)**2/e0
freq = elchcgs/rt*1/np.sqrt(massing*rt)/(2*constants.pi)
lmbd=constants.c/freq
# 
print('R = ',"{:.1e}".format(rt),'cm')
print('wavelength = ',"{:.1e}".format(lmbd),'cm')
```

## Розсіяння Резерфорда
<!--  -->
Формула Резерфорда яка описує [розсіяння $\alpha$-частинок на ядрах атомів](https://en.wikipedia.org/wiki/Rutherford_scattering) походить з задачі класичної механіки двох тіл, які взаємодіють законом Кулона.
<img src="./figs/fig_rutherfordwiki.png" alt="Image" width="400">
<!--  -->

Якщо вважати массу атома значно більшою за маси $\alpha$-частинки (заряд $+2e$), то відповідно можна записати закони збереження
<!--  -->
$$
m\,v_{\infty}\,\rho = m\,r^2\,\dot{\theta},\,\,\,\\
m\,\frac{\dot{r}^2}{2}+m\,v^2_{\infty}\,\frac{\rho^2}{2\,r^2}+\frac{\gamma}{r} = E_{\infty} = m\,\frac{v^2_{\infty}}{2}
$$
звідки витікає рівняння траєкторії
$$
d\theta = \frac{\frac{\rho}{r^2}}{\sqrt{1-\frac{\rho^2}{r^2} - 2\,a\,\frac{\rho}{r}}}\,dr\,,\,\,\,\,a = \frac{\gamma}{m\,v^2_{\infty}\,\rho}
$$
мінімальна відстань взначається рівнянням $\dot{r} = 0$:
<!--  -->
$$ \rho/r_0 =x_0 = \sqrt{1+a^2} - a$$
Звідси для кута розсіяння (по відношенню до початкового напрямку):
<!--  -->
$$\theta = \pi - 2\int\limits^{x_0}_{0} \frac{dx}{\sqrt{1-x^2-2\,a\,x}}$$
<!--  -->
<img src="./figs/fig_cross_section.svg" alt="Image" width="600"/>
<center>Геметрія розсіяння та диференціальний переріз $d\sigma = \rho(\theta)\,d\rho\,d\varphi$</center>

```python
import sympy
from sympy import *
sympy.__version__
```

```python
x = symbols('x',real=True)
a = symbols('a',positive=True)
# 
pi-2*simplify(integrate(1/sqrt(1-x**2-2*a*x),[x,0,sqrt(1+a**2)-a]))
```

або
<!--  -->
\begin{equation}\label{eq:rezerfordtan}
\tan{\theta/2} = a = \frac{Z_{\alpha}\,Z_{n}\,e^2}{m\,v^2_{\infty}\,\rho}
\end{equation}

<!--  -->
Звідси визначається диференціальний переріз розсіяння $\alpha$-частинки:
<!--  -->
\begin{equation}\label{eq:rezerfordcross}
d\sigma = \rho\,d\rho\,d\varphi = \frac{\rho\,d\rho}{\sin\theta}\,d\Omega = \left(\frac{Z_{n}\,e^2}{m\,v^2_{\infty}}\right)^2\frac{d\Omega}{\sin^4\frac{\theta}{2}}
\end{equation}


__Задача__  Що потрібно змінити у формулі \eqref{eq:rezerfordtan} , якщо  ядро-мішень не вважати "закріпленим", тобто яким чином врахувати імпульс віддачи?

__Розв'язок__ Безпосередньо з формули \eqref{eq:rezerfordtan} отримуємо:

```python
kin=0.27e6*1.6e-12 # enrgy in erg
ang = np.pi/3
zau=79 #atomic number Au
rho = zau*elchcgs**2/(kin*np.tan(ang/2))
print("rho = ","{:.2f}".format(rho*0.01*1e12),'pm')
```

__Розв'язок__ У випадку Pb $m_{Pb}\gg m_{He}$ так що імпульсом віддачі можна знехтувати і отже для лобового зіткнення мінімльна відстань обчислюється на основі ЗЗЕ:
$$ \frac{2\,Z_{Pb}\,e^2}{r_{min}} =  T \Rightarrow  r_{min} = \frac{2\,Z_{Pb}\,e^2}{T}$$

```python
kin=0.4e6*1.6e-12 # enrgy in erg
zpb = 82
rmin = 2*zpb*elchcgs**2/(kin)
print("rmin = ","{:.2f}".format(rmin*0.01*1e12),'pm')
```

У випадку з ядрами літію потрібно враховувати імпульс віддачи. Тому доцільно розглянути систему центра мас де імпульси ядер рівні за модулем і протилежні:
$$p_0 = \sqrt{2\,m_{He}\,T} = m_{He}\,v_0$$
$$p_{Li} = m_{Li}\,\frac{m_{He}}{m_{He}+m_{Li}}\,\frac{p_0}{m_{He}}\,\,\,\,p_{He} = m_{He}\,\frac{m_{Li}}{m_{He}+m_{Li}}\,\frac{p_0}{m_{He}}$$
$$\frac{p^2_{Li}}{2\,m_{Li}}+\frac{p^2_{He}}{2\,m_{He}} =\frac{m_{Li}}{m_{He}(m_{Li}+m_{He})}\,\frac{p^2_0}{2} = \frac{Z_{Li}\,Z_{He}\,e^2}{r_{min}}$$
$$r_{min} = \frac{Z_{Li}\,Z_{He}\,e^2}{T}\left(1+\frac{m_{He}}{m_{Li}}\right)$$

## Модель атома Резерфорда-Бора-Зоммерфельда
<!--  -->
__Задача__ Оцінити енергію орбіти мінімального радіусу електрона в атомі водню. 
<!--  -->
__Розв'язок__ На мінімальній орбіті повина вкладатись одна (взагалі ціла кількість - правило квантування орбіт Бора-Зомерфельда) довжина хвилі де Бройля:

$$\lambda = 2\pi \,r_1 = h/p_1 \Rightarrow p_1\,r_1 = \hbar = m\,\omega_1\,r_1^2  \\
m\,\omega_1^2\,r_1 = e^2/r^2_1  \Rightarrow r_1 = \frac{\hbar^2}{m\,e^2}\\
E_1 = \frac{p^2_1}{2\,m} - \frac{e^2}{r_1} = -\frac{e^2}{2\,r_1} = -\frac{m\,e^4}{2\hbar^2}$$

```python
enrg0=constants.value('electron mass')*elchcgs**4/(2*(constants.value("reduced Planck constant")*1e7)**2)/constants.value('elementary charge')*1e-4
print('enrg0 = ',"{:.2f}".format(enrg0),'eV')
```

__Розвязок__ Прискорення при круговому русі:
$$w = \omega^2\,r = \frac{e^2}{m\,r^2}$$
на круговій орбіті
$$E = -\frac{e^2}{2\,r} \Rightarrow \frac{d\,E}{d\,t} = \frac{e^2}{2\,r^2}\,
\frac{d\,r}{d\,t} = - \frac{2\,e^6}{3\,m^2\,c^3\,r^4}  \Rightarrow 
\frac{d\,r}{d\,t} = - \frac{4\,e^4}{3\,m^2\,c^3\,r^2}
$$
$$t_0 = r^3_0 \,\frac{m^2\,c^3}{4\,e^4}$$

```python
t0=(constants.value('Bohr radius')*1e2)**3*(constants.value('electron mass')*1e3)**2*(constants.c*1e2)**3/(4*elchcgs**4)*1e12
print('t падіння = ',"{:.2f}".format(t0),'ps')
```

 ## Правила квантування Бора-Зоммерфельда
<!--  -->
Правмла квантування орбіт Бора полягають в тому що, стаціонарні стани частинки відповдають стоячим хвилям. Тому на класичній орбіті яка відповідає квантовому руху повинна вкладатись ціла кількість хвиль де Бройля. Узагальнюючи цей принцип із застосуванням інтеграла дії вздовж траєкторії правила квантування Бора-Зоммерфельда ставлять у відповідність квантовим станам квантовані значення класичної дії для періодичних траєкторій:
<!--  -->
$$
S_{cl} = \oint \mathbf{p}\cdot\,d\mathbf{r} = 2\pi\,\hbar\,n
$$
<!--  -->
<img src="./figs/atomphys_atom_probsomm.png" alt="Image" width="600">

Знайдемо спектр атома водню використовуючі правила квантування Бора-Зоммерфельда. Оскільки стан електрона в атомі є стоячою хвилею, то скористуємость умовою квантування радіального руху:
<!--  -->
\begin{equation}\label{eq_hatomqclass}
	\int\limits^{r_{+}}_{r_{-}} p_{r}\,d\,r =
	\int\limits^{r_{+}}_{r_{-}}
	\sqrt{2m\left(E-\frac{\hbar^2\,(l+1/2)^2}{2\,m\,r^2}+\frac{\gamma}{r}\right)}\,d\,r=\,\hbar\,
	 \pi\,\left(n_{r}+\frac{1}{2}\right)
\end{equation}
<!--  -->
ми скористаємося спрощеним виразом для відцентрового потенціала, для того щоб отримати точну формулу наприкінці. Великої помилки ми не робимо, оcкільки квазикласичний випадок відповідає $l\gg 1$, $L^2 = \hbar^2\,l\,(l+1)\approx \hbar^2\,(l+1/2)^2$. Далі обчислимо інтеграл:
<!--  -->
$$
\int\limits^{r_{+}}_{r_{-}} p_{r}\,d\,r =\sqrt{2 m |E|}\, \int\limits^{r_{+}}_{r_{-}}
\frac{\sqrt{(r-r_{-})(r_{+}-r)}}{r}\,d\,r= \sqrt{2 m |E|}\,\pi\,
\frac{\left(\sqrt{r_{+}}-\sqrt{r_{-}}\right)^2}{2}
$$
<!--  -->
$$r_{\pm} = \frac{\gamma}{2\,|E|}
\left(1 \pm\sqrt{1-\frac{\hbar^2\,(2l+1)^2\,|E|}{2\,m\,\gamma^2}}\right)
$$
<!--  -->
звідки отримаємо:
$$
\hbar\left(2\,n_{r}+1\right) = \gamma\sqrt\frac{2m}{|E|}\left(1-\frac{\hbar(2l+1)}{\gamma}\sqrt\frac{|E|}{2m}\right)\quad\Rightarrow \quad
E = -\frac{m\,\gamma^2}{2\,\hbar^2\left(n_r+l+1\right)^2}
$$
<!--  -->
що співпадає із точним виразом якщо трактувати $n=n_r+l+1$ як головне квантове число. інші приклади можна розглянути анлогічно.<br>

Інші задачі розвязуються аналогічним чином.
