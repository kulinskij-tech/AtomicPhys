```python
import numpy as np
import matplotlib.pyplot as plt
import sympy
from sympy import *
```

Гамільтоніан гармонічного осцилятора може бути побудовано на основі класичного
виразу із відповідним операторним переозначенням символів
координати та імпульсу:
<!--  -->
\begin{equation}\label{eq_oscxp}
\hat{H} = \frac{\hat{p}^{\,2}}{2m} + \frac{m\,\omega^{\,2}\,\hat{x}^{\,2}}{2}\,.
\end{equation}
<!--  -->
Квадратичний характер \eqref{eq_oscxp} і просте комутаційне співвідношння між операторами координати та імпульсу, дозволяє представити \eqref{eq_oscxp} у вигляді добутку операторів лінійних за символами $\hat{x},\,\hat{p}$:
<!--  -->
\begin{equation}\label{eq_osc_creatanih}
\hat a = 
\frac{1}{\sqrt {2\,\hbar m\omega}}
\left(\hat{p} - i\,m\omega \,\hat{x}\right)\,,\,\,\,\, \hat
a^{\dagger} = \frac{1}{\sqrt{2\,\hbar m\omega}}\left(\hat{p} +
i\,m\omega \,\hat{x}\right)\,,\\
\hat{x} = -i\,\sqrt{\frac{\hbar}{2\,m\,\omega}}\,\left(\hat{a}^{\dagger} - \hat{a}\right)\,,\,\,\,\,
\hat{p} = \sqrt{\frac{\hbar\,m\,\omega}{2}}\,\left(\hat{a}^{\dagger} + \hat{a}\right)
\end{equation}
<!--  -->


Отже \eqref{eq_oscxp} дістає вигляд:
***
\begin{equation}\label{eq_hamosc}
H=\hbar\,\omega\, \left(a^{\dagger}\,a +\frac{1}{2}\right)
\end{equation}
***
Мають місце наступні комутаційні співвідношення:
$$ 
[\hat{a},\hat{a}^{\dagger}]= \hat{a}\,\hat{a}^{\dagger}- \hat{a}^{\dagger}\,\hat{a} = 1 \\
[\hat{a},\hat{H}]= \hat{a}\,\hat{H} - \hat{H}\,\hat{a} = \hbar\, \omega\,[\hat{a},\hat{a}^{\dagger}\,\hat{a}] = \hbar\, \omega
\,\hat{a}
$$
Оператори $\hat a$ и $\hat a^{\dagger}$ є
аналогічними операторам $\hat{L}_{\,\pm}$, оскільки здійснюють
перехід між стаціонарними станами по енергетичній
сходинці:
<!--  -->
$$
\hat{H}\,\hat{a} \ket{E} = (E-\hbar\,\omega)\,\hat{a}\,\ket{E}  \simeq \left|\,E\,-\hbar\,\omega\right\rangle\\
0 \le \langle E\,| \hat{a}^{\dagger} \hat{a} \ket{E} = 
\frac{E}{\hbar\,\omega} - 1/2
$$
тобто значення енергії можна вважати цілими кратними $\hbar\,\omega$ з урахуванням "нульового рівня" $E_{0} = \hbar\,\omega/2$ - енергії основного стана ([наявність такого стана вже було продемостровано раніше](./qm1_wavepack_zad.ipynb#enrgosc0)):
<!--  -->
$$
E_{n} = \hbar\,\omega \left(n+\frac{1}{2}\right)\\
\hat a \left|\,n\,\right\rangle
 =\, \sqrt {n}
\left|\,n-1\,\right\rangle\,,\,\,\,
\hat a^{\dagger} \left|\,n\,\right\rangle = \sqrt {n + 1}
\left|\,n+1\,\right\rangle \,. \notag
$$
<!--  -->
Зокрема:
\begin{equation}\label{eq_osc_creatanih2} 
|\, n \rangle \, = 
\,\frac{\left(\hat a^{\dagger}\right)^{n}}{\sqrt {n\,!}}\,
|\,  0\,\rangle\,,\,\,\,
\langle n\,| \, = \, \langle 0\,|\,
\frac{\hat{a}^{n}}{\sqrt {n\,!}}\,.
\end{equation}

Такі оператори можна інтерпретувати як оператори народження $\hat{a}^{\dagger}$ та знищення $\hat{a}$ кванту енергії $\hbar\,\omega$. Рівні енергії одномерного гармонічного осциллятора мають вигляд:
<!--  -->
\begin{equation}
\label{eq105} E = \hbar\, \omega \left( {n + {\frac{1}{2}}}
\right)\,,\,\,n = 0,1,2\,...
\end{equation}
<!--  -->
і тому власні стани оператора \eqref{eq_oscxp} індексуються невід'ємним цілим індексом - головним квантовим числом $n$. У власному для оператора \eqref{eq_oscxp} представленні вони можуть бути ототожнені із вектор стовбцями (ket-векторами в термінології Дірака) $| \, n \rangle $ на $n$-м місці у яких стоїть одиниця, а всі інші дорівнюють 0. Завдяки самоспряженості оператора \eqref{eq_oscxp} вони є визначають базис гільбертова простору $L_{2} $. Інваріантність гамільтоніана відносно перетворення $x\to -x$ інверсії веде до того що власні стани мають певну парність, яка співпадає з парністю квантового числа $n$. Дію операторів $\hat a^{\dagger} ,\,\hat a$ можна інтерпретувати як народження та знешкодження кванта енергії $\hbar\, \omega $. Тому вони називаються операторами народження і знешкодження відповідно. Оператор $N = \hat a^{\dagger} \,\hat a$ є оператором числа квантов енергії $\hbar\, \omega $  чи
оператором числа заповнення рівня $E=\hbar\,\omega$. Він відіграє важливу роль при описі стану багаточастинкових систем. [Хвильові функції стаціонарних станів осилятора у координатному представленні було отримано раніше](./qm1_stac_zad.ipynb#wfosc). 

```python
psi,f,f0 = symbols('psi f f0',cls=Function)
x = symbols('x',real=True)
k = symbols('k')
# 
def hamosc(f):
    a=-diff(f,x,2)/2+x**2*f/2
    return a
print('Hamiltonian:')
hamosc(psi(x))
```

```python
# 
def acreat(f):
    a=1/sqrt(2)*(-diff(f,x)+x*f)
    return a
# 
def annih(f):
    a=1/sqrt(2)*(diff(f,x)+x*f)
    return a
# a*a+1/2 ham
print('check: rerpresentation of the hamiltonian: H = a*a+1/2')
simplify(acreat(annih(psi(x)))+1/2*psi(x))
```

```python
#[a,a*] commut
print('[a,a*]=',simplify((annih(acreat(psi(x)))-acreat(annih(psi(x))))/psi(x)))
```

### Когерентні стани осцилятора
Знайдемо власні стани оператора $\hat{a}$:
$$
\hat{a}\,\psi_{\lambda}(x) = \lambda\, \psi_{\lambda}(x)
$$

```python
diffeq0=Eq(annih(f(x)),k*f(x))
# diffeq0
sol0=dsolve(diffeq0,f(x))
sol0
```

```python
rho=(abs(sol0.args[1])**2)
rho
```

Як бачимо ці функції належать простору $L_2$ при довільному $\lambda \in \mathbb{C}$ (у формулах вище цей параметр позначено як $k$). Тому спектр оператора $\hat{a}$ складається з множини комплексних чисел. Відмітимо, що основний стан відповідає власному значенню $\lambda=0$. Такі стани у координатному (та імпульсному) представленні виявляються гаусівськими пакетами зі зміщенням визначаємим ${\rm Re} \lambda$. Порахуємо дисперсію координати:

```python
cnorm=refine(integrate(rho,(x,-oo,+oo)),Q.positive(k))
x1av=simplify(refine(integrate(x*rho,(x,-oo,+oo)),Q.positive(k)))/cnorm
x2av=simplify(refine(integrate(x**2*rho,(x,-oo,+oo)),Q.positive(k)))/cnorm
```

```python
sqrt(expand(x2av-x1av**2))
```

```python
diffeq1=Eq(acreat(f(x)),k*f(x))
# diffeq0
sol1=dsolve(diffeq1,f(x)).subs(Symbol('C1'),1)
sol1
```

```python
from IPython.display import display, Markdown
display(Markdown(r'густина $|\psi|^2: $'))
abs(sol1.args[1])**2
```

Оскільки стани $\lambda\ne 0$ не є стаціонарними, то вони демонструють нетривіальну еволюцію:
$$
i\,\hbar\,\frac{d\,\hat{a}}{d\,t} = [\hat{a},\hat{H}] = \hbar\,\omega\,\hat{a}
$$
відповідно:
$$
i\,\hbar\,\frac{\partial\,\psi_{\lambda}}{\partial\,t} = \hbar\,\omega\,\lambda(t)\,\psi_{\lambda}
\Rightarrow \lambda(t) = \lambda(0)\,e^{-i\,\omega\,t} 
$$
тобто параметр $\lambda$ змінює тільки фазу. Зміст назви "когерентний" обумовлений єдиною частотою коливань для всіх станів. Побудуємо наприклад густину ймовірності координати $|\psi_{\lambda}(x,t)|^2$.

```python
from ipywidgets import interactive
# 
def rho(x,t):
    r= np.exp(-x**2-2*np.sqrt(2)*x*np.sin(2*np.pi*t))
    return r
# 
def f1(t):
#     plt.figure()
    x = np.linspace(-4, 4, num=100)
    plt.plot(x,rho(x,t))
    plt.fill_between(x,rho(x,t)) #filled region between plot and x-axis
#    
    plt.xlim(-5,5)
    plt.title('$\\rho_{0}=|\psi_{0}(x;t)|^2$') #not simply \rho but with escape!: \\rho
    plt.show()
# 
interactive_plot = interactive(f1, t=(0, 2, 0.1))
interactive_plot
```

Власні функції оператора народження $\hat{a}$:
$$
\hat{a}^{\dagger}\,\psi_{\lambda}(x) = \lambda\, \psi_{\lambda}(x)
$$

не належать простору $L_2$. Cпектр цього оператору теж складається з множини комплексних чисел оскільки він спряжений до $\hat{a}$.<br>

Когерентні стани цікаві певними характеристичними властивостями. Перша їх
характерна властивість це мінімізація співвідношення невизначеностей,
тобто в таких станах:
$$
\langle \Delta x^{\,2}\rangle \,\langle \Delta p^{\,2}\rangle  = \frac{\hbar^{\,2}}{4}
$$
і дисперсії як координати так і імпульса одночасно мінімальні:
$$
\langle\Delta x^{\,2}\rangle  = \frac{\hbar}{2\,m\omega }\,,\quad   \left\langle\Delta p^{\,2} \right\rangle = \frac{\hbar \,m\omega }{2}
$$
<!--  -->
Друга властивість це їх еволюція. Вона призводить тільки до зміни фази параметра $\lambda $:
<!--  -->
\begin{equation}
\label{eq_coherevolution}
S(t,0)\ket{\lambda} =
e^{\frac{H}{i\,\hbar}\,t}\,\ket{\lambda} = \ket{\lambda(t)}
\,\,,\,\,\,\lambda (t) = \lambda (0)\,e^{- i\,\omega\,t}
\end{equation}
<!--  -->
Це пояснює зміст назви "когерентний". Кванти енергії, які формують такі стани еволюціонують зберігая різницю фаз, так що когерентний характер станів не змінюється. Якщо отожнити комплексне число 
<!--  -->
$$\lambda ={\rm Re}\lambda + i\,{\rm Im}\lambda $$ 
<!--  -->
з вектором на класичній фазовій площині
<!--  -->
$$z = x\omega \sqrt{\frac{m}{2}}+ i\,\frac{p}{\sqrt{2m}}$$
<!--  -->
то відповідність з рухом класичного осцилятора є практично повна. Середні значення координати та імпульса змінюються за гармонічним законом (див. Рис.).
Таким чином ці стани є аналогами станів класичного осциллятора. Слід однак пам'ятати, що когерентні стани не є власними для гамільтониана, тобто для них енергія (число квантів) не має певного значення, тому повної аналогії немає (і в принципі бути не може). Тим не менш виконується більш слабка вимога. Оскільки:
<!--  -->
\begin{equation}
\label{coherosc_lambda}
 \av{\lambda}{\hat{H}}{\lambda} = \hbar\, \omega \,\left(|\lambda|^{\,2}+ 
 \frac{1}{2}\right),
\end{equation}
<!--  -->
то, враховуючи \eqref{eq_coherevolution}, це означає, що в результаті еволюції когерентного стана середнє значення енергії не змінюється. З \eqref{coherosc_lambda} видно, що величина $|\lambda |^{\,2}$ може бути пройтерпретована як середнє число квантів у когерентному стані. Розподіл числа квантів в когерентномустані визначається як:
<!--  -->
\begin{equation}
\label{eq9913}
p_{\lambda}(n)\, =\vert\braket{n}{\lambda}\vert^{\,2}
\end{equation}
<!--  -->
Використовуючи \eqref{eq_osc_creatanih2}:
<!--  -->
\begin{equation}\label{eq9914}
\braket{n}{\lambda} =
 \av{0}{ \frac{\hat{a}^{n}}{\sqrt{n!}} }{ \lambda } = \,\frac{\lambda^{n}}{\sqrt{n!}}
 \braket{0}{\lambda}
\end{equation}
<!--  -->
і умову нормування:
<!--  -->
$$
\braket{\lambda}{\lambda}= \sum\limits_{n} \vert\braket{n}{\lambda}\vert^{\,2}\, = 1
$$
<!--  -->
отримуєм
<!--  -->
$$
| \braket{0}{\lambda}|^{\,2}\, = \left(\sum\limits_{n}
\frac{|\lambda |^{2n}}{n!}\right)^{-1}= \exp{\left(- | \lambda |^{\,2}\right)}
$$


Як результат, для \eqref{eq9913} остаточно отримаєм розподіл Пуассона:

$$\label{eq_poisspn}
p_{\lambda} (n) = \,\exp \left(- |\lambda|^{\,2}\right)\,\frac{|\lambda |^{2n}}{n!}
$$

Таким чином у когерентному стані кванти енергії реєструються некорельованим чином подібно до дробового шуму. Множина когерентних станів утворює переповнену систему векторів (це витікає з того, що вони параметризуються множиною комплексних чисел). Повнота витікає з того, що довільний базисний вектор $\ket{n}$ може бутий розкладений за когерентними станами. Дійсно, використовуючи \eqref{eq_poisspn}, отримаємо:

$$
\ket{n}= \,\frac{1}{\pi}
\int\limits_{C}\exp\left(-\frac{| \lambda |^{\,2}}{2}\right)
\frac{\bar{\lambda}^{n}}{\sqrt{n!}}\,\ket{\lambda} \,d\lambda \,d\bar{\lambda}
$$

Тут інтегрування ведется по всій комплексній площині. Вектори когерентніх станів не є ортогональними, однак:

$$
 \braket{\lambda _{1}}{\lambda _{2}}= 
 \exp \left(-\frac{|\lambda_{1}|^{\,2}+ |\lambda _{2}|^{\,2}}{2}\right)
 \sum\limits_{m,n}\frac{\bar{\lambda}_{1}^{n}\lambda_{2}^{m}}{\sqrt{n!m!}}\, 
 \braket{n}{m} = 
 \exp\left(- \frac{|\lambda _{1}|^{\,2}}{2}- \frac{| \lambda_{2}|^{\,2}}{2}+
 \bar{\lambda}_{1}\lambda_{2} \right)
$$

і 

$$
|\braket{\lambda_{1}}{\lambda_{2}}|^{\,2}= \exp\left(- |\lambda_{1}- \lambda_{2}|^{\,2}\right)
$$

тобто когерентні стани із суттєво різними середнім числом квантів "майже ортогональні". Квантовий осцилятор незважаючи на модельність еквивалентний до моді ЕМП в ідеальному резонаторі. Теорію когерентних станів було розвинуто саме для застосування у теорії квантових генераторів ЕМВ - лазерів.

Тепер можна описати стан осцилятора, у тепловії рівновазі із термостатом. Звичайно осцилятор не може знаходитись у стаціонарному стані тому що термостат діє як зовнішє джерело обміну енергією. Але у стані теплової рівноваги, як ми знаємо повинен бути певний розподіл по енергії по нсамблю. Цей розподіл як відомо повинен мати больцманівський вигляд: 

$$
p_{n}= C\exp \left(- \beta \,E_{n} \right)
$$

Тому стан теплової рівноваги для осцилятора
представляється стат. оператором:

$$\label{densmatrix_osc}
 \hat \rho = \sum\limits_{n} p_{n}\ket{n}\bra{n}\,,\\
 \sum\limits_{n}p_n =1 \,,\,\,\Rightarrow\,\,\,
 C = 1 - \exp \left(- \beta \,\hbar \,\omega  \right)
$$

Середнє значення числа заповнення довівнює:

$$
\label{avnum_osc}
 \langle N \rangle \, ={\rm Tr}\left(\hat{\rho} \,\hat{N}\right) =\sum\limits_{n} p_{n}\,\av{n}{\hat N}{n} = \sum\limits_{n} n\,p_{n} =
\frac{1}{\exp \left(\beta \,\hbar \,\omega \right) - 1}
$$

Цей результат свідчить про те що розподіл числа квантів є розподілом Бозе-Айнштайна.
