## Експериментальні витоки: розщеплення спектральних ліній у полях

Ефекти Штарка і Зеємана вимірюють, як зовнішнє електричне або магнітне поле знімає виродження атомних рівнів. Спостерігається не просто зміщення однієї лінії, а поява кількох компонент, кількість і поляризація яких кодують квантові числа та правила відбору.

**Ідея експерименту.** Газ або атомний пучок розміщують між електродами чи полюсами магніту і спостерігають спектр високої роздільної здатності. Залежність положень ліній від поля прямо перевіряє матричні елементи дипольного або магнітного моменту.


## Атом у електричному полі: ефект Штарка
<!--  -->
Гамільтоніан атома у однорідному електростатичному полі має вигляд:
<!--  -->
\begin{equation}
\label{hamstark} H = H_{0} - \mathbf{D}\,\cdot\,\mathbf{\mathcal{E}}
\end{equation}
<!--  -->
де $\mathbf{D}$ - оператор дипольного моменту атома:
$$
\mathbf{D} = -\sum\limits^{Z}_{k}e\,\mathbf{r}_i
$$
Звісно у атома з'являється додаткова енергія 
<!--  -->
$$ - \langle D_z\,\cdot\,\mathcal{E}_z \rangle$$
<!--  -->
Далі ситуація залежить від того чи є у атома (молекули) власний дипольний момент чи ні.
Якщо у атома є власний дипольний момент то має місце лінійний по полю ефект. У випадку відсутності власного момента зявляється індукований дипольний момент $\mathbf{D}^{(ind)} = \alpha\,\mathbf{\mathcal{E}}$ (взгалалі $\alpha$ тензор поляризуємості). Відповідна додаткова енергія
$$
\Delta U = -\frac{\alpha}{2}\,\mathcal{E}_z^2
$$
і ефект буде квадратичним. Звісно квадратичний ефект додається до лінійного у випадку наявності власного дипольного момента. В атомах лужних металів ефект Штарка є квадратичним по полю. Лінійний ефект відсутній завдяки тому що густина розподілу електронної хмари дається парною функцією, а проекція дипольного момента це непарна функція при дзеркальному перетворенні $\mathbf{r}\to -\mathbf{r}$:
$$
\langle D_z\,\cdot\,\mathcal{E}_z\rangle_{n,l} \sim \int\limits^{\pi}_{0} \cos \theta\,\vert P^{|m|}_{l}(\theta)\vert^2\, d(\cos{\theta})
$$
В атомі водню ситуація інша завдяки тому що має місце виродження рівнів енергії по орбітальному числу $l$ (специфічне для кулонівського потенціала).
Очевидно, що в стаціонарних станах атома водню $\langle\mathbf{D}\rangle_n = 0$ і
$$
\av{n,l,m}{D_z\,\cdot\,\mathcal{E}_z}{n,l',m'} \sim \int\limits^{\pi}_{0} \cos \theta\,Y^{*}_{l,m}\,Y_{l',m'}\, d(\cos{\theta})\sim \delta_{l,l'\pm 1}\,\delta_{m,m'}
$$
У звязку зз цим рівень $n$ розщеплюється на $2n-1$ рівней. При чому стани $l,\pm m$ мають одинакову енергію. Таким чином, ефект Штарка в атомі водню це
ефект першого порядка (лінійний по полю). Наприклад, при $n=2$, є 4-х кратне виродження рівня енергії $E_{n=2,l=0} = E_{n=2,l=1}$ (ми не враховуємо виродженя за проекцією спіна, т.к. електричне поле не впливає на спін у нерелятивистському наближені) і у однорідному електричному полі утвориться триплет так як з 4 станів 2 будуть мати однакову енегргію.

## Атом у магнітному полі: ефект Зеємана
Оскільки спін взаємодіє з магнітним полем завдяки спіновому магнітному моменту, то гамільтоніан частинки зі спіном у зовнішньому ЕМП має вигляд
<!--  -->
\begin{equation}\label{hampauli}
H = \frac{1}{2m}
\left(- i\,\hbar \nabla -
\frac{q}{c} \mathbf{A} \right)^{\,2} + q\,\varphi -
\mathbf{M}_s \cdot \mathbf{B}\\
\mathbf{M}_s = \frac{q}{mc}\mathbf{S}
\end{equation}
<!--  -->
Відповідно, гамільтоніан атома в однорідному зовнішньому магнітному поле має вигляд:
<!--  -->
\begin{equation}
\label{tfeq14}
\begin{array}{l}
 H = \frac{1}{2\,m}\sum\limits_{k}^{N} \left[ \hat{\mathbf{p}}_{k} -
\frac{q}{c}\hat{\mathbf{A}}(\mathbf{r}_{k} ) \right]^{2} + U -
\frac{q\,\hbar}{m\,c}\mathbf{H}\cdot\mathbf{S} \\
 {\mathbf{S} = \sum\limits_{k}^{N} \mathbf{s}_{k}\,,\quad
\hat{\mathbf{A}}(\mathbf{r}_{k} ) = \frac{1}{2}}
\mathbf{H}\times \hat{\mathbf{r}}_{k} \\
 \end{array}
\end{equation}
<!--  -->
сума проводиться за всіма електронами, $q = - \,e$. Тут ми виділили сталу Планка $\hbar$ з оператора спина, так що оператор спина є безрозмірним. Виділяючи в (\ref{tfeq14})
незбурений гамільтоніан, отримуємо:
<!--  -->
\begin{multline}\label{tfeq15}
 H = H_{0} + \frac{e}{m\,c}
 \sum\limits_{k}^{N} \mathbf{A}(\mathbf{r}_{k})\cdot\mathbf{p}_{k}+\\ 
 \frac{e^2}{8\,m\,c^2}\,\mathcal{H}^2\,\sum\limits_{k}^{N} \vert\mathbf{e}_H\times\mathbf{r}_{k}\vert^2 -
\frac{q\,\hbar }{m\,c}
\mathbf{H}\cdot\mathbf{S}= \text{лінійне наближення по полю}\to \\ 
H_{0} +
\mu _{B} \mathbf{H} \cdot \left(\mathbf{L}+
2\,\mathbf{S} \right) =
    H_{0} +
    \mu _{B} \mathbf{H} \cdot \,\left(\mathbf{J} + \mathbf{S}
    \right)\\
\end{multline}
<!--  -->
$$ \mu _{B} = {\frac{{e\,\hbar}} {{2\,m\,c}}} $$.

Оскільки у подальшому магнітне поле за припущенням є малим, ми опустили квадратичний доданок в (\ref{tfeq15}). Зовнішнє магнітне поле приводить до того, що стани з різними проекціями повного моменту імпульсу $M_J$ мають різні енергії. За відсутності спіна $\mathbf{J} = \mathbf{L}$ і ми отримаємо т.з. нормальний ефект Зеємана  - розщеплення рівня на $2L+1$ підрівнів які відповідають різним проекціям $M_{L} = -L,\ldots L$. Присутність спіна призодить до складного ефекту Зеємана який розглянемо нижче.

### Спін-орбітальна взаємодія. Тонка структура
Спін облонькі взаємодіє не тільки із зовнішнім магнітним полем але і з внутрішнім, яке створюється спіном ядра (надтонка структура) та орбітальним рухом електронів (спін-орбітальна взаємодія, тонка структура). Останній ефект значно більший взаємодії спін-спін електрнної взаємодії та взаємодії спін-спін(ядро) (магнетон Бора $\gg$ ядерний магнетон, оскільки $m_e \ll m_{p}$). З класичної т.з. спін-орбітальна взаємодія виникає завдяки релятивістському ефекту взаємодії спіна з магнітним полем струма орбітального руху електронів. У миттєвій системі відліку електрона виникає магнітне поле завдякі електростатичному полю ядра (лабсистема відліку). Тому  виникає відповідна магнітна взаємодія:
<!--  -->
\begin{equation}\label{ham_spm}
	H_{sp} =-\mathbf{M}_{s}\cdot\mathcal{H}'
\end{equation}
<!--  -->
де
<!--  -->
$$\mathbf{M}_{s} = -\frac{e}{mc}\,\mathbf{S}\,,\quad
\mathcal{H}' = -\frac{1}{c}\,\mathbf{v}\times \mathbf{E}$$
<!--  -->
і ми скористались формулами перетворення ЕМП: $\mathbf{E}'_{\parallel}=\mathbf{E}_{\parallel}\,,
\,\,\,\mathcal{H}'_{\parallel}=\mathcal{H}_{\parallel}$
<!--  -->
$$\mathbf{E}'_{\perp} =\gamma\,\left( \mathbf{E}_{\perp}+\frac{1}{c}\,\mathbf{v}\times\mathcal{H}\right)\,,\quad
\mathcal{H}'_{\perp} =\gamma\,\left( \mathcal{H}_{\perp}-\frac{1}{c}\,\mathbf{v}\times\mathbf{E}\right)$$
<!--  -->
Додатково, однак слід врахувати, що електрон рухається не вздовж прямої, а по круговій орбіті і має місце додаткова прецесія Томаса. Це призводить до зменшення швидкості прецесії у 2 рази:
<!--  -->
\begin{equation}\label{ham_sp}
H_{sp} =\frac{e}{2\,m^2c^2}\,\mathbf{S}\cdot\left(\mathbf{E}\times \mathbf{p}\right)
\end{equation}
<!--  -->
Це т.з. спін-орбітальна взаємодія, оскільки для центрального поля:
<!--  -->
\begin{equation}\label{eq:spinorb}
\mathbf{S}\cdot\left(\mathbf{E}\times \mathbf{p}\right) = -\frac{1}{r}\frac{d\,\varphi}{d\,r}\mathbf{S}\cdot\left(\mathbf{r}\times \mathbf{p}\right) = -\frac{1}{r}\frac{d\,\varphi}{d\,r}\,
\mathbf{S}\cdot\mathbf{L}
\end{equation}
Це можна інтерпретувати як взаємодію спінового магнітного момену з магнітним моментом струма породженного орбітальним рухом.
<!--  -->
<!-- Цей внесок має порядок величини:
$$H_{sp}\sim \frac{e^2\,\hbar^2}{m^2c^2\,r^3_{0}} =
\frac{e^2/r_0}{m\,c^2}\,\frac{\hbar^2}{m\,r^2_{0}}\sim \alpha^2\,E_{0}
$$ -->
<!--  -->

Величина розщеплення LS-термів з різними значеннями квантового числа $J$ за рахунок спін-орбітальної
взаємодії може бути обчислена наступним чином:
<!--  -->
$$\mathbf{L}\cdot\mathbf{S} = \frac{1}{2}\left((\mathbf{L}+\mathbf{S})^2 - \mathbf{L}^2 - \mathbf{S}^2\right)$$
<!--  -->
\begin{multline}\label{ateq4}
\av{\psi}{A\,\mathbf{L}\cdot\mathbf{S}}{\psi}_{J} -
\av{\psi}{A\,\mathbf{L}\cdot\mathbf{S}}{\psi}_{J - 1} = 
\frac{\bar{A}}{2}
\left( J(J + 1) - L(L + 1) - S(S + 1)\right)-\\
\frac{\bar{A}}{2}\left( J(J - 1) - L(L + 1) - S(S + 1) \right) = \bar{A}\,J\,.
\end{multline}
<!--  -->
Якщо $\bar{A}>0$,  то такий мультиплет називають нормальним. У протилежному випадку -- оберненим. Якщо магнітне поле мало настільки, що не виникає перекриття мультиплетів при розщепленні, то збурені рівні можна нумеровати квантовими числами $J,L,S$. Середнє значення оператора $J_{z} $ дорівнює $\hbar \,M_{J} $. Середнє значення оператора проекції спина в станах з певними $J,L,S$ (хвильові функції мультиплету) можна розглядати як оператор, який діє в цьому просторі станів. Його дія визначається тільки квантовим числом $M_{J} $. Оскільки він комутує з оператором $J_{z} $, то:

\begin{equation}\label{tfeq16} 
\av{\psi _{J,L,S,M_{J}}}{S_{z}}{\psi_{J,L,S,M_{J}}} = C\,M_{J} ,
\end{equation}
<!--  -->
Тому:
<!--  -->
\begin{equation}\label{tfeq18} 
\av{\psi _{J,L,S,M_{J}}}{S_{z}}{\psi_{J,L,S,M_{J}}} = M_{J}
\frac{\mathbf{J}\cdot\mathbf{S}}{\mathbf{J}^{\,2}} = M_{J}
\frac{(\mathbf{L}+\mathbf{S})\cdot\mathbf{S}}{\mathbf{J}^{\,2}}
\end{equation}
<!--  -->
і для енергії розщеплення остаточно отримаємо:
<!--  -->
\begin{equation}
\label{tfeq19}
\begin{array}{l}
 \Delta \,E = \mu _{B} g\,M_{J} \mathcal{H} \\
 g = 1 + {\frac{{J(J + 1) - L(L + 1) + S(S + 1)}}{{2\,J(J + 1)}}} \\
 \end{array}
\end{equation}
<!--  -->
$g$ називається __множником Ланде__ або гіромагнітним множником. Фізично зрозуміло, що атоми з повністю заповненими оболонками $L = S = 0$ є діамагнітними.

З (\ref{tfeq19}) видно, що магнітне поле повністю знімає виродження по проекції $\mathbf{J}$. Ця обставина відрізняє ефект Зеємана від ефекту Штарка, де анізотропія обумовлена електричним полем. При ефекті Штарка залишається виродження рівнів з проекціями $\pm M_{J} $, Оскільки гамільтоніан є інваріантним відносно віддзеркалень в площині, яка містить напрямок поля (проекція моменту імпульсу на ось змінює знак). У випадку ефекту Зеємана гамільтоніан (\ref{tfeq14}) не є інваріантним при такому
перетворенні, тому таке виродження відсутнє. Якщо магнітне поле таке, що енергія спін-орбітальної взаємодії мала у порівнянні з енергією атома у магнітному поле, то виникає перекриття термів і збереження $L,S$ не має місця (LS звязок розривається). Однак можна говорити про збереження проекцій операторів на напрямок поля. В цьому випадку енергія розщеплення, очевидно, дорівнює:

\begin{equation}
\label{tfeq20} \Delta \,E = \mu _{B} \,\mathcal{H}{\rm (}M_{L} +
2M_{S} {\rm )}
\end{equation}

Розщеплення у випадку сильних (у порівнянні з інтервалами тонкої структури) полів називається ефектом Пашена -Бака.
