## Експериментальні витоки: розщеплення спектральних ліній у полях

Ефекти Штарка і Зеємана вимірюють, як зовнішнє електричне або магнітне поле знімає виродження атомних рівнів. Спостерігається не просто зміщення однієї лінії, а поява кількох компонент, кількість і поляризація яких кодують квантові числа та правила відбору.

**Ідея експерименту.** Газ або атомний пучок розміщують між електродами чи полюсами магніту і спостерігають спектр високої роздільної здатності. Залежність положень ліній від поля прямо перевіряє матричні елементи дипольного або магнітного моменту.


## Атом у електричному полі: ефект Штарка
<!--  -->
Гамільтоніан атома у однорідному електростатичному полі має вигляд:
<!--  -->
\begin{equation}
\label{hamstark} H = H_{0} - \mathbf{D}\,\cdot\,\mathbf{\mathcal{E}}
\end{equation}
<!--  -->
де $\mathbf{D}$ - оператор дипольного моменту атома:
$$
\mathbf{D} = -\sum\limits^{Z}_{k}e\,\mathbf{r}_i
$$
Звісно у атома з'являється додаткова енергія 
<!--  -->
$$ - \langle D_z\,\cdot\,\mathcal{E}_z \rangle$$
<!--  -->
Далі ситуація залежить від того чи є у атома (молекули) власний дипольний момент чи ні.
Якщо у атома є власний дипольний момент то має місце лінійний по полю ефект. У випадку відсутності власного моменту з'являється індукований дипольний момент $\mathbf{D}^{(ind)} = \alpha\,\mathbf{\mathcal{E}}$ (взгалалі $\alpha$ тензор поляризуємості). Відповідна додаткова енергія
$$
\Delta U = -\frac{\alpha}{2}\,\mathcal{E}_z^2
$$
і ефект буде квадратичним. Звісно квадратичний ефект додається до лінійного у випадку наявності власного дипольного моменту. В атомах лужних металів ефект Штарка є квадратичним по полю. Лінійний ефект відсутній завдяки тому що густина розподілу електронної хмари дається парною функцією, а проекція дипольного моменту це непарна функція при дзеркальному перетворенні $\mathbf{r}\to -\mathbf{r}$:
$$
\langle D_z\,\cdot\,\mathcal{E}_z\rangle_{n,l} \sim \int\limits^{\pi}_{0} \cos \theta\,\vert P^{|m|}_{l}(\theta)\vert^2\, d(\cos{\theta})
$$
В атомі водню ситуація інша завдяки тому що має місце виродження рівнів енергії по орбітальному числу $l$ (специфічне для кулонівського потенціала).
Очевидно, що в стаціонарних станах атома водню $\langle\mathbf{D}\rangle_n = 0$ і
$$
\av{n,l,m}{D_z\,\cdot\,\mathcal{E}_z}{n,l',m'} \sim \int\limits^{\pi}_{0} \cos \theta\,Y^{*}_{l,m}\,Y_{l',m'}\, d(\cos{\theta})\sim \delta_{l,l'\pm 1}\,\delta_{m,m'}
$$
У зв'язку зз цим рівень $n$ розщеплюється на $2n-1$ рівней. При чому стани $l,\pm m$ мають одинакову енергію. Таким чином, ефект Штарка в атомі водню це
ефект першого порядка (лінійний по полю). Наприклад, при $n=2$, є 4-х кратне виродження рівня енергії $E_{n=2,l=0} = E_{n=2,l=1}$ (ми не враховуємо виродженя за проекцією спіна, т.к. електричне поле не впливає на спін у нерелятивістському наближені) і у однорідному електричному полі утвориться триплет так як з 4 станів 2 будуть мати однакову енегргію.

## Атом у зовнішньому магнітному полі: застосування LS-результату

У [розділі про атомні терми](https://colab.research.google.com/github/kulinskij-tech/AtomicPhys/blob/main/Atomic_py/atomicphys_atomsterms.ipynb) ми вже визначили внутрішню LS-структуру та ефективний магнітний момент рівня:

$$\overline{\boldsymbol\mu}=-g_J\mu_B\frac{\mathbf J}{\hbar}.$$

Тепер додаємо однорідне зовнішнє поле $\mathbf B=B\mathbf e_z$. Воно виділяє лабораторну вісь і взаємодіє з цим моментом. Нижче $\mathbf L,\mathbf S,\mathbf J$ мають розмірність моменту імпульсу, а $\mu_B$ та $B$ використовуються в СІ.

### Слабке поле: ефект Зеємана

Якщо характерна магнітна енергія $\mu_B B$ значно менша від відповідних інтервалів тонкої структури, зовнішнє поле майже не змішує рівні з різними $J$. Спочатку враховуємо внутрішній LS-зв'язок, а поле розглядаємо як збурення:

$$H_Z=-\overline{\boldsymbol\mu}\cdot\mathbf B
=g_J\mu_B B\frac{J_z}{\hbar},\qquad
\Delta E_{J,M_J}=g_J\mu_B B M_J.$$

Тепер різні $M_J=-J,-J+1,\ldots,J$ можуть мати різні енергії. Це відмінність від тонкого розщеплення без зовнішнього поля, яке залежить від $J$, але зберігає виродження за $M_J$.[^1]

Для $S=0$, $J=L>0$ маємо $g_J=1$ — нормальний ефект Зеємана. Для $S\ne0$ орбітальний і спіновий внески мають різні ваги, що дає складний, або аномальний, ефект Зеємана. Наприклад, у лужного металу рівні $^{2}P_{1/2}$ і $^{2}P_{3/2}$ мають $g_J=2/3$ і $4/3$, а $^{2}S_{1/2}$ — $g_J\simeq2$.

Кількість компонент спектральної лінії не дорівнює автоматично $2J+1$: треба враховувати підрівні обох станів переходу та правила відбору. В електричному дипольному наближенні $\Delta M_J=0,\pm1$.[^2]

Для рівня $J=0$ лінійний зеєманівський зсув відсутній. Так само за $g_J=0$ лінійний внесок не розділяє $M_J$-стани. Змішування з іншими рівнями може давати ефекти вищого порядку.

### Зв'язок із гамільтоніаном електронів у полі

Той самий магнітний внесок отримуємо, розкладаючи гамільтоніан із мінімальним зв'язком. Для заряду електрона $-|e|$ та калібрування $\mathbf A=\tfrac12\mathbf B\times\mathbf r$ в СІ

$$H=\sum_i\frac{(\mathbf p_i+|e|\mathbf A_i)^2}{2m_e}+U
+\frac{g_S\mu_B}{\hbar}\mathbf S\cdot\mathbf B
=H_0+\frac{\mu_B}{\hbar}\mathbf B\cdot(\mathbf L+g_S\mathbf S)
+\frac{|e|^2}{8m_e}\sum_i|\mathbf B\times\mathbf r_i|^2.$$

Останній доданок є квадратичним за полем. У лінійному наближенні залишається сума орбітального та спінового магнітних внесків; проєктування на вже сформований LS-рівень дає наведений вище результат із $g_J$.

### Сильніше поле: ефект Пашена–Бака

Якщо магнітна енергія переважає над спін-орбітальними інтервалами, але ще мала порівняно з електростатичними відстанями між термами, спочатку враховуємо взаємодію окремих моментів із полем:

$$H_Z=\frac{\mu_B B}{\hbar}(g_L L_z+g_S S_z),\qquad
\Delta E\simeq\mu_B B(g_L M_L+g_S M_S).$$

За $g_L=1$, $g_S\simeq2$ це $\Delta E\simeq\mu_B B(M_L+2M_S)$. У цьому наближенні добрими квантовими числами є $L,S,M_L,M_S$, а $J$ уже не є добрим квантовим числом: прецесія навколо зовнішнього поля переважає над внутрішнім зв'язком $\mathbf L$ і $\mathbf S$. Між слабким і сильним полем потрібен розгляд обох взаємодій разом.

Ми не враховуємо тут ядерний спін і надтонку структуру. За їх наявності найслабші поля можуть вимагати опису через повний атомний момент $\mathbf F=\mathbf J+\mathbf I$.


[^1]: NIST. [Zeeman effect and magnetic splitting factors](https://www.nist.gov/pml/atomic-spectroscopy-compendium-basic-ideas-notation-data-and-formulas/atomic-spectroscopy-zeeman).
[^2]: NIST. [Spectral lines: selection rules](https://physics.nist.gov/Pubs/AtSpec/node17.html).
