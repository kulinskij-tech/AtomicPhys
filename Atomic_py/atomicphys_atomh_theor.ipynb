Задача про спектр енергії зарядженої частинки у полі кулонівського потенціала
є однією з основних точно розв'язуваних задач квантової механіки (разом із задачею про гармонічний осцилятор). Її можна вважати "нульовим" наближенням для більш складних атомних систем.

Гамільтониан моделі має вигляд:
<!---->
\begin{equation}\label{ham_coulomb}
\hat{H} = -\frac{\hbar^{\,2}}{2m}\,\Delta - \frac{\gamma}{r}
\end{equation}
<!---->
де $m = \frac{m_e M}{m_e+M}$ -- приведена маса системи ядро + електрон,
$\gamma = Z e^{\,2}$. 

Енергію основного стану, як відомо, можна отримати із квазикласинчих міркувань квантування орбіт (більш детально дивись [Квазикласичне наближення](./qm2_quasiclass.ipynb)). Дійсно, якщо вважати основний стан відповідаючий руху по колу із мнімальним значенням моменту імпульса, то:
<!--  -->
$$
m\,v^2/r_0 = \frac{Z\,e^2}{r^2_0}\,,\,\,\,\, m\,v\,r_0 = L_0 = \hbar\\
r_0 = \frac{\hbar^2}{Z\,m\,e^2}
$$
<!--  -->
звідки отримаємо "радіус орбіти" Бора та енергію (згідно з теоремою віріала). Це дійсно збігається із точним квантовомеханічним розражунком (див. нижче), але слід пам'ятати що ніякої "кругової орбіти" у основному стані немає, оскільки в цьому стані взагалі момент імпульсу відсутній $l=0$ тобто розподіл координати електрона подібний до стоячої сферичносиметричної хвилі.

## Експериментальний ключ: лінійчастий спектр водню

Лінії Бальмера та узагальнена формула Рідберга були відомі до повної квантової теорії атома. Саме регулярність спектральних частот підказує, що випромінювання відповідає переходам між дискретними рівнями енергії, а не довільним орбітальним частотам класичного електрона.

**Ідея експерименту.** Розряд у водні створює збуджені атоми, а спектрометр розкладає випромінювання за довжинами хвиль. Емпірична регулярність $1/\lambda=R(1/n_1^2-1/n_2^2)$ стає прямим свідченням дискретного спектра гамільтоніана атома.


```python
from scipy import constants
# 
ab=constants.hbar**2*(4*np.pi*constants.epsilon_0)/(constants.m_e*constants.e**2)
print('Bohr radius = ',ab, 'm')
# 
constants.physical_constants['Bohr radius']
```

З теореми віріала отримуємо повну енергію:

$$
E_0 = -\frac{Z\,e^2}{2\,r_0} = -Z^2\,\frac{m\,e^4}{2\,\hbar^2}
$$
В перерахунку на електрон-вольти це складає ($Z=1$)

```python
print('Energy of ground state = ',"{:.2f}".format(constants.e/(8*np.pi*constants.epsilon_0*ab)), 'eV')
```

Оскільки кулонівський потенціал є
сферично симетричним, то інтегралами руху (крім гамільтоніана) є
компоненти оператора момента імпульса $L_{i}$. Таким чином стаціонарні стани атома водню можна характеризувати трьома квантовими числами, які відповідають трьом комутуючими операторам:
<!---->
$$
\hat{H}, \,\,\hat{\mathbf{L}}^2\,,\,\,\,\hat{L}_z
$$
<!---->
Як і у класичному випадку можна скористатись сферичною симетрією для розділення змінних:
<!---->
$$\psi_{n,l,m} (r,\theta,\varphi) = R_{n,l}(r)\,Y_{l,m}(\theta,\varphi)$$
ця х.ф. є власною функцією операторів $\mathbf{L}^2, L_z$ та гамільтоніану:
<!--  -->
\begin{equation}\label{ham_sph}
\hat{H} \psi_{n,l,m} = E\, \psi_{n,l,m} \Rightarrow \left( -\frac{\hbar^{\,2}}{2m}\,\frac{1}{r^2}\frac{\partial}{\partial\,r}
\left(r^2\,\frac{\partial}{\partial\,r}\right)+ \underbrace{\frac{\hbar^2\,l(l+1)}{2\,m\,r^2} - \frac{Ze^2}{r}}_{U_{eff}(r)}\right)\,R_{n,l} = E\,R_{n,l}\,,
\end{equation}

Після введення безрозмірної змінної $\tilde{r}\to  \varkappa\,r$, де

$$\varkappa^2 = \frac{2\,m\,|E|}{\hbar^2}\,,\,\,E<0$$

радіальне рівняння має вигляд:
<!--  -->
$$
-\frac{1}{\tilde{r}}
\frac{d^2}{d\tilde{r}^2}\left(\tilde{r}\,R \right) + \left(\frac{\,l(l+1)}{\tilde{r}^2} - \frac{b}{\tilde{r}}\right)\,R_{n,l} + R_{n,l} = 0\\
b = \frac{2\,m}{\hbar^2}\,\frac{Z\,e^2}{\varkappa}
$$
<!--  -->
Враховуючи асимптотики при $\tilde{r}\to \infty, R\sim e^{-\tilde{r}}$ та $\tilde{r}\to 0, R\sim \tilde{r}^{l}$ можна спростити рівняння, проводячи заміну: 
<!--  -->
$$
R = \tilde{r}^{\,l}\,e^{-\tilde{r}}\,L(\tilde{r})
$$
<!--  -->
В результаті отримаєму рівняння для визначення $L(r)$:
<!--  -->


```python
L = Function('L')
R,l,r,k=symbols('R l r k')
R = exp(-r)*r**(l)*L(r)
expand(simplify(r**2*expand(diff(R*r,(r,2))/r-(l*(l+1)/r**2-b/r)*R+R)*exp(r)/r**(l+1)))
```

$$(b -2(l+1)) L + 2(1 + l - \tilde{r}) L' + 
 \tilde{r} L'' = 0$$
 Розвязок якого будемо шукати у вигляді ряду
 $$L = \sum\limits_{k=0}\, a_k\,\tilde{r}^k$$
 підстановка якого до рівняння дає наступні рекурентні співвідношення для коефіцієнтів $a_k$:
 $$a_{k+1} = a_k\frac{2(1+l+k)-b}{(1+k)(2(l+1)+k)}$$
 Зрозуміло, що при $k\gg 1$:
 $$a_{k+1}\simeq \frac{2}{k+1}\,a_{k} \Rightarrow a_k = \frac{2^k}{k!}$$
 що відповідає ряду для $\exp(2\tilde{r})$ s це призвведе до ненормованої хвильової функції $\psi\sim e^{2\tilde{r}}$. Якщо ж для певного індексу $k_0$ виконується співвідношення
<!--   -->
\begin{equation}
\label{eq:enrg00}
1+l+k_0-b/2 = 0
\end{equation}
<!--   -->
то $L$ буде поліномом і відповідна х.ф. буде нормована. Таке число, що визначає порядок полінома $L(r)$ називається радіальним квантовим числом $n_r$. Т.ч. співвідношеня \eqref{eq:enrg00} визначає рівні енергії:
\begin{equation}
\label{eq:enrg0}
\frac{2\,m}{\hbar}\,\frac{Z\,e^2}{\sqrt{2\,m\,|E|}} = 2(\underbrace{1+l+n_r}_{n})\Rightarrow E  = -\frac{m\,e^4}{2\hbar^2}\frac{Z^2}{n^2}
\end{equation}
$n =1+l+n_r$ - головне квантове число. Звідси витікає що рівні енергії не залежать а ні від азимутального квантового числа $m$ (сферична симетрія) а ні від орбітального квантового числа $l$ так що енергетичному рівню $E_n$ відповідає 
$$\sum\limits_{l=0}^{n-1}(2l+1) = n^2$$
різних хвильових функцій які визначаються набором квантових чисел $n,l,m$. Така ситуація називається виродженням енергетичного рівня. Наприклад, енергетичні стани вільної частинки (кінетична енергія) мають нескінченне виродження за напрямками імпульсу. За наявності зовнішнього поля, або інших взаємодій що порушують сферичну симетрію, таке виродження частково або повністю зникає. Наприклад у воднеподібних атомах (лужні метали) енергетичні стани слабко звязаного електрону залежать як від головного так і від орбітального квантового числа.

## Класична задача (коротка довідка)

З класичної механіки відомо, що при русі у кулонівському(гравітаційному)
полі існує одатковий інтеграл руху -- вектор Лапласа-- Рунге -- Ленца:
<!---->
$$\label{laprunglenz}
\mathbf{R} = \mathbf{p}\times \mathbf{L} - \gamma \,\mu\,\frac{\mathbf{r}}{r}
$$
<!---->
Дійсно:
<!---->
$$\label{laprunglenz_vyvod}
\frac{d}{d\,t}\left({\mathbf{p}}\times {\mathbf{L}}\right) = \dot{\mathbf{p}}\times \mathbf{L} = 
-\frac{\gamma}{r^3}\,\mathbf{r}\times (\mathbf{r}\times \mathbf{p}) =
-\frac{\gamma}{r^3}\,\left(\mathbf{r} (\mathbf{r}\cdot \mathbf{p}) - \mathbf{p}\,r^2\right) =\\
\gamma\,\mu\,\left(\frac{\dot{\mathbf{r}}}{r} - \frac{\mathbf{r} (\mathbf{r}\cdot \dot{\mathbf{r}})}{r^3}\right) = \gamma\,\mu\,\frac{d}{d\,t}\left(\frac{\mathbf{r}}{r}\right)\,.
$$
<!---->
Знайдемо модуль цього вектора:
<!---->
\begin{multline}\label{r2}
  \mathbf{R}^2= p^2\,L^2-2\,\mu\,\frac{\gamma}{r}\,\mathbf{r}\cdot (\mathbf{p}\times \mathbf{L})+m^2\,\gamma^2 = 
  2\,\mu\,E\,L^2 + m^2\,\gamma^2 =\\ m^2\,\gamma^2\left(1+\frac{2\,E\,L^2}{\mu\,\gamma^2}\right)=
  m^2\,\gamma^2\,e^2\,.
\end{multline}
<!---->
\[R = \gamma\,\mu\,e\]
<!---->
де $e$ - ексцентриситет:
$$
e = \sqrt{1+\frac{2\,E\,L^2}{\mu\,\gamma^2}}
$$
Рівняння траекторії безпосередньо витікає з факту інваріантності вектора $\mathbf{R}$:
<!---->
$$\label{R_traj}
  \mathbf{R}\cdot \mathbf{r} = R\,r\,\cos\varphi = L^2 - \gamma\,\mu\,r
$$
<!---->
Звідки:
<!---->
$$\label{eq_classtraj}
r = \frac{r_0}{1+e\,\cos{\varphi}}\,,\,\,\,\,\, r_0 = \frac{L^2}{\mu\,\gamma}\,.
$$

```python
x = np.linspace(0.1,10,200)
# 
def ueff(x): 
    return 4*x**2 - 10*x
# 
x1=2
y1=-2
coeff=[-y1,-10,4]
roots=np.roots(coeff)
plt.plot(x,ueff(1/x))
plt.plot(x,y1*x/x,'r',dashes=[2,3])
plt.plot(roots[0],y1,'ro',ms=12)
plt.plot(roots[1],y1,'ro',ms=12)
plt.axhline(y=0,xmin=-0.1,xmax=10,color='black',dashes=[3,3])
plt.text(8,2.5,'E>0',fontsize=16)
plt.text(8,-3.5,'E<0',fontsize=16)
plt.title('Effective potential energy $U_{eff}(r)$\n',fontsize=14)
plt.ylim(-8, 4)
plt.show
```
