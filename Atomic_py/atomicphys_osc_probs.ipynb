Розв'язки задач цього розділу базуються на алгебрі операторів $a$ і $a^\dagger$. Алгебраічний метод побудови спектра гармонічного осциллятора може бути узагальнено наступним чином. Нехай є оператор Шредингера $\hat{H}$ з дискретним невиродженим спектром $E_n\,, n=0,1\ldots$. Визначимо сходинкові оператори:
$$
\hat{a}
\left|\,E_n\,\right\rangle = \sqrt{E_n - E_0}
\left|\,E_{n-1}\,\right\rangle\,,\quad \hat{a}^{\dagger}
\left|\,E_n\,\right\rangle = \sqrt{E_{n+1}- E_0}
\left|\,E_{n+1}\,\right\rangle\,,
$$
вони мають комутаційні властивості:
$$
 \left\langle\, m
\,\right|\left[\hat{a},\hat{a}^{\dagger}\right] \left|\,n
\,\right\rangle = (E_{n+1} - E_n)\delta_{m n}\,.
$$
В такому представленні, аналогічно до \eqref{eq_osc_creatanih2},  всі стани можуть бути отримані в результаті дії оператора народження на основний стан $\vert E_0 \rangle$.  
Очевидно має місце представлення:
$$
\hat H_{0} = a^{\dagger}\,a +
E_0\,.
$$
Крім того, кожен гамільтоніан зі спектром $E_{0}<E_{1}<\ldots$ природньо породжує послідовність гамільтоніанів зі спектрами:
$$
{\rm Spec} H_{n} = E_{n}<E_{n+1}< \ldots
$$
Спробуємо побудувати таку послідовність. Якщо $\hat{a}_{0}$ обрати у вигляді:
$$\hat{a}_{0} = p + i\,\frac{\psi_0'}{\psi_0}\,,\quad  
\hat{a}_{0}\,\psi_0 = 0 $$ де $\psi_0$ - хвильова функція основного стана, тобто $\hat{a}_{0}$ знешкоджує основний стан, тоді:
$$
\hat{a}^{\dagger}_{0} \hat{a}_{0} = \left(\hat p - i\,\frac{\psi_0'}{\psi_0}\right)
\left(\hat p + i\,\frac{\psi_0'}{\psi_0}\right) = p^2 +
\frac{\psi_0''}{\psi_0} = \hat{p}^2+V(x)-E_0 = H_{0}-E_0\,,\quad m=1/2
$$
Послідовно виключаючи зі спектру значення $E_n$, знайдемо оператори:
$$
\hat{H}_{n} = \hat{a}_{n-1}\hat{a}_{n-1}^{\dagger} + E_{n-1} = 
\hat{a}_{n}^{\dagger}\,\hat{a}_{n} + E_n\,,\\
\hat{a}_n
\left|\,E_k\,\right\rangle = \sqrt{E_k - E_n}
\left|\,E_{k-1}\,\right\rangle\,,\quad \hat{a}^{\dagger}_{n}
\left|\,E_k\,\right\rangle = \sqrt{E_{k+1}- E_n}
\left|\,E_{k+1}\,\right\rangle\,.
$$
Зрозуміло, що мають місце включення $Spec \hat{H}_{n+1} \subset Spec \hat{H}_{n}$ і виконується співвідношення:
\begin{eqnarray}
  \hat{a}_{n} \, \hat{H}_{n} =& \hat{H}_{n+1}\,\hat{a}_{n} \notag\\
  a^{\dagger}_n\, \hat{H}_{n+1} =& \hat{H}_{n}\, a^{\dagger}_n \label{han}
\end{eqnarray}
з якого витікає, що:
\begin{equation}\label{anh}
\hat{H}_{n+1} = \hat{a}_{n} a^{\dagger}_n + E_n\,.
\end{equation}
Тобто, задача о знахождені спектра оператора $\hat{H}_{0}$
вирішується побудовою послідовності операторів $\hat{H}_{n}$.

__Задача__ Використовуючи метод операторів народження-знищення покажіть,
що має місце наступне рекурентне співвідношення для поліномів Эрміта:
$$2\,x\,H_n = H_{n+1} +2\,n\,H_{n-1}$$
<br>
__Розв'язок__ Визначемо оператори $a$ і $\hat{a}^{\dagger}$ наступним чином:

$$
\hat x = \frac{\hat{a}^{\dagger} + \hat{a} }{\sqrt{2}}
$$
Тоді:
$$
\hat{x}\left|\,n \right\rangle = \frac{\hat{a}^{\dagger} + \hat{a}
}{\sqrt{2}}\frac{\left( \hat a^{\dagger} \right)^{n}}{\sqrt {n\,!} }\,
\left|\,0 \,\right\rangle
$$
Знайдемо комутатор:
<!--  -->
$$
[\hat{a},\hat{a}^{\dagger\,n}] = \hat{a}\,\hat{a}^{\dagger\,n} -
\hat{a}^{\dagger\,n}\,\hat{a} = \hat{a}\,\hat{a}^{\dagger\,n} -
\hat{a}^{\dagger\,(n-1)}\,\hat{a}\hat{a}^{\dagger} +
\hat{a}^{\dagger\,(n-1)}\,\hat{a}\hat{a}^{\dagger} -
\hat{a}^{\dagger\,n}\,\hat{a} =\\
[\hat{a},\hat{a}^{\dagger\,(n-1)}]\,\hat{a}^{\dagger} +
\hat{a}^{\dagger\,(n-1)} =
\left(\,[\hat{a},\hat{a}^{\dagger\,(n-2)}]\,\hat{a}^{\dagger}+
\hat{a}^{\dagger\,(n-2)}\,\right)\,\hat{a}^{\dagger} +
\hat{a}^{\dagger\,(n-1)} = \ldots\\ = n\,\hat{a}^{\dagger\,(n-1)}
$$
<!--  -->
звідки:
<!--  -->
$$
2\,\hat{x}\left|\,n \right\rangle = \sqrt{2(n+1)}
\left|\,n+1\,\right\rangle + \sqrt{2n}\,
\left|\,n-1\,\right\rangle\,.
$$
Позначимо:
$$ 
\left\langle\,x\,\vert\,n\,\right\rangle = C_n\,
H_n(x)\, \left\langle\,x\,\vert\,0\,\right\rangle
\quad 
\Leftrightarrow \quad  \psi_{n}(x) = C_{n}\,H_{n}(x)\,\psi_{0}(x)
$$
тоді:
\begin{equation}\label{hnc}
  2\,x\,C_n\,H_n = \sqrt{2(n+1)}\,C_{n+1}\,H_{n+1} +
\sqrt{2n}\,C_{n-1}\,H_{n-1}\,.
\end{equation}
<!--  -->
Визначимо константи нормування таким чином:
$$
\frac{C_{n+1}}{C_n} = \frac{1}{\sqrt{2(n+1)}} \Rightarrow C_n = \frac{1}{\sqrt{2^{n}\,n!}}\,.
$$
В результаті з \eqref{hnc} отримуємо необхідне співвідношення. Якщо ввести генеруючу функцію таких полиномів:
$$f(x,t) = \sum\limits_{n=0}\,\frac{H_{n}(x)}{n!}\,t^n$$
то це співвідношення дає рівняння для $f(x,t)$:
$$2(x-t)\,f(x,t)=\frac{\partial\,f}{\partial t}$$
остаточно:
$$f(x,t) = e^{2x\,t-t^2}$$

\ket{\psi(t=0)} = \ket{0}+\ket{1}

Очевидно еволюція вектора стана описується наступним чином:

$$\ket{\psi(t)}=\frac{1}{\sqrt{2}}\left(\ket{0}+e^{-i\,\omega\,t}\ket{1}\right)$$

далі за означенням

$$\left\langle x \right \rangle_{\psi} = -i\,\sqrt{\frac{\hbar}{2\,m\,\omega}}\,\av{\psi(t)}{\hat{a}^{\dagger} - \hat{a}}{\psi(t)}$$

$$\langle E\rangle = \av{\psi(t)}{\hat{a}^{\dagger}\,\hat{a}+\frac{1}{2}}{\psi(t)}$$

__Задача__ Методом сходинкових операторів визначіть спектр частинки у
$\infty$-глибокій ямі.
<br>
__Рішення__  В данному випадку оператор Шредінгера має формальний вигляд ($m=1/2$):
$$\hat{H} = \hat{p}^2\,.$$
Будемо шукати оператор $\hat{a}_0$ початкового крока у вигляді:
$$\hat{a}_0 = \hat{p} -i\,f_0(x)\,,
$$
$f_0(x)$ - певна (гладка) функція координати $x$. Тоді ($\hbar = 1$):
$$
\hat{a}_0^{\dagger}\hat{a}_0 = \hat{p}^2+f^2_0(x) - f^{'}(x) = p^2 - E_0\,,
$$
тобто
\begin{equation}\label{e0}
  f^2_0(x) - f^{'}(x) + E_0 =0\,.
\end{equation}
Для зручності обмежимо область руху інтервалом $-\pi/2 \le x\le \pi/2$. Виключаючи значення $E_0 = 0$ з \eqref{e0} видно, що функцію $f_0$ доцільно обрати наступним чином:
$$f_0(x) = \alpha_0\,\tan x\,.$$
Тоді \eqref{e0} приводить до:
\begin{equation}\label{e1}
  \alpha_0 = 1\,,\quad E_0 = \alpha_0^2 = 1\,.
\end{equation}

Використовуючи вже знайдені значеня параметрів знаходимо:

$$
\hat{a}_0 \hat{a}_0^{\dagger} = \hat{p}^2+1+2\,\tan^2{x}\,.
$$

З іншого боку  зкористаємось співвідношенням:
\begin{equation}\label{e2}
H_{1} = \hat{a}_1^{\dagger} \hat{a}_1 +E_1 = \hat{a}_0 \hat{a}_0^{\dagger} +
E_0 = p^2+\frac{2}{\cos^2{x}}\,,
\end{equation}

з якого витекає, що $\hat{a}_1$ слід знову шукати як:

$$\hat{a}_1 = \hat{p}-i\,\alpha_1\,\tan x\,.$$

Стала $\alpha_1$ визначається з \eqref{e2}: $\alpha_1 = 2$.
Діючи за індукцією, визначим оператори:

\begin{equation}\label{na}
  \hat{a}_{n} = \hat{p}-i\,\alpha_n\,\tan \hat{x}
\end{equation}

і отримуємо рекурентні співвідношення:

$$\alpha_{n+1}(\alpha_{n+1}-1) = \alpha_{n}(\alpha_{n}+1)\,,$$

з відповідним розв'язком:

$$\alpha_{n} = (n+1)\,, \quad E_n = (n+1)^2\,.$$
