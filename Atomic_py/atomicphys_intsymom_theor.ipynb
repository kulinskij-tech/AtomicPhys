У попередніх лекціях ми вже бачили, що у квантовій теорії фізичні величини (координата й імпульс) представлюються операторами які діють на хвильову функцію (як функцію положення частинки - т.з. координатне представлення):
<!--  -->
$$\hat{x} \psi(\mathbf{r}) = x\, \psi(\mathbf{r})\\
\hat{p}_x \psi(\mathbf{r}) = -i\,\hbar\,\frac{\partial \psi(\mathbf{r})}{\partial\,x}
$$
<!--  -->
Аналогічно для інших координат. Для імпульсного представлення х.ф. як амплітуд хвиль де Бройля $c(\mathbf{p})$:
<!--  -->
$$\hat{p}_x\, c(\mathbf{p}) = p_x\, c(\mathbf{p})\\
\hat{x}\, c(\mathbf{p}) = i\,\hbar\,\frac{\partial\, c(\mathbf{p})}{\partial\,p_x}
$$
При чому завдяки ймовірносному змісту х.ф. середні значення від функції координат чи імпульсу можуть бути знайдені за формулами:
<!--  -->
$$		\begin{array}{l}
		\langle f(x) \rangle \, = \,{\int\limits_{V}
			{\psi^* (\mathbf{r},t)}} \,f(x)\,\psi
		(\mathbf{r},t)\,dV = \,{
        \int\limits_{\Lambda}{c^*(\mathbf{p},t)\,}}\,\hat{f}\left( {i\hbar \,{\frac{{\partial}}
				{{\partial \,p_{x}}}}} \right)\,
                c(\mathbf{p},t)\,d\Lambda \\
		\langle g(p_{x}) \rangle \, = \,{\int\limits_{V}
			{c^* (\mathbf{p},t)\,p_{x}^{n}}
		}\,c(\mathbf{p},t)\,d\Lambda = {\int\limits_{V} {\psi^*
				(\mathbf{r},t)} }\,\,\hat{g}\left( {- i\hbar \,{\frac{{\partial}}
				{{\partial \,x}}}}\right) \,\psi (\mathbf{r},t)\,dV\,, \\
	\end{array}
$$
<!--  -->
Оператор енергії - гамільтоніан будується на основі цих операторів, особливо для тих систем для яких можна навести класичні аналоги (частинка у потенціальному полі). Важливою вимогою до оператору ф.в. є його ермітовість. Вона означає виконання наступної рівності  для середнього значення оператора (фізичної величини) у стані з х.ф. $\psi$:
<!--  -->
$$
\langle A\rangle_{\psi} = \int \psi^{*}(x)\,\left(\hat{A}\,\psi(x)\right)\,dx = \int \left(\hat{A}\,\psi(x)\right)^*\,\psi(x)\,dx = \langle A\rangle^{*}_{\psi} 
$$
тобто середнє значення - дійсне число. Наприклад шамільтоніан частинки у потенціальному полі:
$$
\langle H\rangle_{\psi} - \langle H\rangle^{*}_{\psi}  = \int \psi^{*}(x)\,\left(-\frac{\hbar^2}{2\,m}\frac{d^2}{d\,x^2}+V(x)\,\psi(x)\right)\,dx - \int \left(-\frac{\hbar^2}{2\,m}\frac{d^2}{d\,x^2}+V(x)\,\psi(x)\right)^*\,\psi(x)\,dx =\\
\frac{\hbar^2}{2\,m}\,\int\left( \psi\,\frac{d^2\,\psi^{*}}{d\,x^2}-\psi^{*}\,\frac{d^2\,\psi}{d\,x^2}\right)\,dx = \frac{\hbar^2}{2\,m}\,\int\,\frac{d}{d\,x}\left( \psi\,\frac{d\,\psi^{*}}{d\,x}-\psi^{*}\,\frac{d\,\psi}{d\,x}\right)\,dx = 0
$$
Остання рівність виконується завдяки накладанню граничних умов, що дозволяє визначити відповідний ермітовий оператор енергії. Умова ермітовості таким чином надає математичну форму представлення фізичної величини у вигляді оператора. Спектр такого оператора (власні значення) - це можливі значення фізичної величини. Власні х.ф.:
$$ \hat{A}\,\psi_{a} = a\,\psi_{a} $$
це такі стани в яких фізична величина має відповідне значення. 
Як правило, для опису стана системи однієї фізичної величини не
достатньо. Величина:
<!--  -->
$$
		[\hat A,\hat B]=\hat A\,\hat B-\hat B\,\hat A
$$
<!--      -->
є лінійним оператором і називається комутатором операторів ${A}$ і ${B}$. Дві фізичні величини ${A}$ і $\hat B$ називаються __комутуючими__, якщо:
\begin{equation}\label{comm}
[\hat A,\hat B]=0\,.
\end{equation}

Очевидно, що коли оператори мають спільні власні стани, тобто можуть бути одночасно виміряні, то вони комутують. За певних математичних умов для операторів справедлива і зворотна теорема. З цієї т.з. оператори  $\hat{x}\,, \hat{p}_x$ не комутують:
<!--  -->
$$
\left(\hat{x}\,\hat{p}_{x} -\hat{p}_{x}\,\hat{x}\right) \psi = i\,\hbar\,\psi \Rightarrow  [\hat{x},\hat{p}_{x}] = i\,\hbar\\
[\hat{x},\hat{p}_y]\psi = \left(\hat{x}\,\hat{p}_{y} -\hat{p}_{y}\,\hat{x}\right) \psi = 0$$

Більш детально питання зображення ф.в. ермітовими операторами розглядається в курсі квантової механіки.
Основна задача квантової механіки - знаходження спектра енергії квантової системи (гамільтоніан) тобто спектра оператора Шредінгера, власні стани це стаціонарні стани в яких енергія задана точно.

### Оператор момента імпульсу.
Розглянемо з цієї т.з. момент імпульса. Як відомо з курсу класичної механіки момент імпульсу $\mathbf{L} = \mathbf{r}\times\mathbf{p}$ є інтегралом руху у полі центральних сил (згадаємо 2-й закон Кеплера!)
<!--  -->
\begin{equation}\label{ham_newt2}
m\,\frac{d^2\mathbf{r}}{d\,t^2} = \frac{d\mathbf{p}}{d\,t} = -\frac{\partial V(r)}{\partial\,\mathbf{r}} \Rightarrow \mathbf{r}\times\frac{d\mathbf{p}}{d\,t}=0 \Rightarrow \frac{d}{dt} \mathbf{L} =0
\end{equation}
<!--  -->
так що енергія руху у такому полі може бути представлена як
<!---->
\begin{equation}\label{ham_central}
E = \frac{\mathbf{p}^2}{2\,m} + V(r) = \frac{p_r^2}{2\,m} + V(r)+ \frac{\mathbf{L}^2}{2\,m\,r^2}
\end{equation}
<!---->
Останній доданок є т.з. відцентровою енергією, оскільки повязана з обертальним рухом при $\mathbf{L}\ne 0$.

Кутовий (орбітальний) момент імпульсу у квантовій механіці можна ввести по аналогії з класичною механікою як наступний оператор:
<!--  -->
\begin{equation}
\label{eq10/7} \mathbf{L} = -i\,\hbar \,\mathbf{r}\times \nabla =
\mathbf{r}\times \mathbf{p}
\end{equation}
<!--  -->
$$
L_z = x\,p_y - y\,p_x = x\,\left(-i\,\hbar\,\frac{\partial}{\partial\,y}\right) - y\,\left(-i\,\hbar\,\frac{\partial}{\partial\,x}\right)
$$
<!--  -->
Інші компонени як завжди у випадку векторного добутку можуть бути отримані циклічною перестановкою $x\to y\to z\to x$. Можна переконатися, що компоненти цього оператора задовольняють наступним комутаційним співвідношенням:
<!--  -->
\begin{equation}\label{eq:qll}
[\hat{L}_{i},\hat{L}_{j}]=i\,\hbar\,\epsilon_{ijk}\,\hat{L}_{k}
\end{equation}
<!--  -->
Бачимо, що на відміну від компонент операторів координати чи імпульсу, компоненти $\mathbf{L}$ не комутує. Оператор квадрату моменту імпульса
<!--  -->
$$
\hat{\mathbf{L}}^2 = \hat{\mathbf{L}}\cdot \hat{\mathbf{L}} = L^2_{x}+L^2_{y}+L^2_{z}\\
[\hat{L}_{i},\hat{\mathbf{L}}^2]=0\,.
$$
<!--  -->
комутуеє з всіма компонентами. Таким чином, стан квантової системи може бути заданий власними значеннями операторів $L^{\,2}$ і однією з проекцій в якості якої стандартно обирають умовну вісь $z$. Розглянемо функції:
<!--  -->
$$
f(z),(x\pm i\,y)^m
$$
<!--  -->
Очевидно:
<!--  -->
$$
\hat{L}_z f(z) = 0\,,\,\,\,\hat{L}_z (x\pm i\,y)^m = \hbar\,m\,(x\pm i\,y)^m
$$
<!--  -->
це означає що вони є власними функціями оператора $\hat{L}_{z}$ з власними значеннями $0,\hbar\,m$ відповдіно. Взагалі можна вважати, що так само як дія оператора проекції імпульса дає швидкість зміни х.ф. при малому зсуві вздовж відповідного напрямку, дія оператора проекцї момента імпульсу дає швидкість зміни х.ф. при обертанні на малий кут навколо відповідної вісі.

```python
x,y,z,l=symbols('x y z l',positive=True)
# 
def lz(f):
    return I*(y*diff(f,x)-x*diff(f,y))
# 
def lx(f):
    return I*(z*diff(f,y)-y*diff(f,z))
# 
def ly(f):
    return I*(x*diff(f,z)-z*diff(f,x))
# 
def l2(f):
    return lx(lx(f))+ly(ly(f))+lz(lz(f))
# 
```

```python
m=3
simplify(lz((x+I*y)**m)-m*(x+I*y)**m)
```

Аналогічне має місце і для оператора квадрату імпульса чи кінетичної енергії (довести самостійно).<br> ___Вказівка:___ скористайтесь імпульсним представленням.

Зрозуміло, що:
<!--  -->
$$(x\pm i\,y)^l = r^l \sin^{l}{\theta}\,e^{\pm\,i\,l\,\varphi}$$
<!--  -->
Функції $\sin^{l}{\theta}\,e^{\pm\,i\,l\,\varphi}$ дають приклади т.з. сферичних гармонік:
<!--  -->
$$Y_{l,\pm l}(\theta, \varphi) = C_{l}\,\sin^{l}{\theta}\,e^{\pm\,i\,l\,\varphi}$$
<!--  -->
Які є власними функціями як для $\hat{\mathbf{L}}^2$ так і для $\hat{L}_{z}$. Використовуючи явний вираз для оператора проекції моменту імпульсу:
<!--  -->
\begin{equation}\label{eq:lz} 
L_{\,z} = - \,i\hbar \,\frac{\partial}{\partial \,\varphi}
\end{equation}
<!--  -->
і умову періодичності хвильової функції $\psi (\varphi ) = \psi (\varphi + 2\pi )$, отримаємо, що $l$ також повинно бути цілим, оскільки власна функція для \eqref{eq:lz} має вигляд:
$$
\psi_{m} = C\exp (i\,m\,\varphi )\,.
$$
<!--  -->
Таким чином, виходячи з комутаційних співвідношень між компонентами оператора момента імпульса, отримано:
<!--  -->
- оператор квадрата моменту імпульсу і проекції на певну вісь є одночасно спостережувальними величинами;
- власні значення оператора $\mathbf{L}^{\,2}$ дорівнюють $\hbar^{\,2}l\,(l + 1)$
- власні значення оператора проекції моменту імпульсу на певну вісь
<!--  -->
$\hbar \,m\,\,,\,\, - l \le m \le l\,\,,\,\,m \in Z$, 
тобто проекція моменту імпульсу на цю вісь є квантованою.

<!--  -->
Безпосереднім розрахунком можна переконатися, що у сферичних координатах:
<!--  -->
\begin{equation}\label{eq:l2} 
\mathbf{L}^{\,2} = - \hbar^{\,2}
\left(\frac{1}{\sin\theta}\frac{\partial}{\partial \,\theta}\left(\sin\theta \frac{\partial}{\partial \,\theta} \right) + 
\frac{1}{\sin^{\,2}\theta}\frac{\partial^{\,2}}{\partial \,\varphi^{\,2}}\right)
\end{equation}
<!--  -->
У дужках стоїть власне кутова частина оператора Лапласа. Дійсно,
як відомо сферичні координати допускають розділення змінних на радіальну і кутову у оператора Лаласа (оператор кінетичної енергії):
<!--  -->
\begin{equation}
\label{eq17/7} {H} =- \frac{\hbar ^{\,2}}{2\,m}\Delta =  - \frac{\hbar ^{\,2}}{2\,m}\,\left(\frac{1}{r^{\,2}}\,\frac{\partial}{\partial\,r}\,\left(r^{\,2}\,\frac{\partial}{\partial \,r} \right) + 
\frac{1}{r^{\,2}}\,\Delta_{\theta,\varphi}\right)
\end{equation}
<!--  -->
Отримаємо цей вираз. Для цього потрібно переписати похідні у відповідних координатах, наприклад:
<!--  -->
$$
\frac{\partial }{\partial\,x } = 
\frac{\partial r}{\partial\,x }\,\frac{\partial }{\partial\,r }+
\frac{\partial \theta}{\partial\,x }\,\frac{\partial }{\partial\,\theta }+
\frac{\partial \varphi}{\partial\,x }\,\frac{\partial }{\partial\,\varphi }
$$
і т.д. використовуючі відповідні формули перходу:
<!--  -->
$$
x = r\,\sin{\theta}\,\cos{\varphi}\,,\,\,\,\,\,\, r = \sqrt{x^2+y^2+z^2}\\
y = r\,\sin{\theta}\,\sin{\varphi}\,,\,\,\,\,\, \theta = {\rm atan}\frac{\sqrt{x^2+y^2}}{z} \\
z = r\,\cos{\theta}\,,\,\,\,\,\,\, \varphi = {\rm atan}\frac{y}{x}
$$
<!--  -->
Реалізацію за допомогою SymPy цієї нудної аналітики наведено нижче.([Sympy може двого "думати", тому значно швидкіше це можна зробити у WolframMathematica. Взагалі, imho WM найкраща з систем компалгебри і я раджу вам її використовувати!](https://colab.research.google.com/github/kulinskij-tech/AtomicPhys/blob/main/Atomic_py/sphcoordlaplace_we.ipynb))

```python
r,th,ph=symbols('r th ph',positive=True)
f = Function('f')
# r1 = Function('r1')
# th1= Function('th1') 
# ph1 = Function('ph1')
# 
# s0=[(x,r*sin(th)*cos(ph)),(y,*sin(th)*sin(ph)),(z,r*cos(th))]
# 
r1=sqrt(x**2+y**2+z**2)
th1=atan(sqrt(x**2+y**2)/z)
ph1 = atan(y/x) 
# 
def df(f,coord): #nabla operator in sph coord
    return simplify((diff(f,r)*diff(r1,coord)+diff(f,th)*diff(th1,coord)+diff(f,ph)*diff(ph1,coord)).subs([(x,r*sin(th)*cos(ph)),(y,r*sin(th)*sin(ph)),(z,r*cos(th)),(abs(sin(th)),sin(th))]))
```

Оператор $\Delta$ у сферичних координатах:

```python
expand(simplify(expand(df(df(f(r,th,ph),x).subs(abs(sin(th)),sin(th)),x)+df(df(f(r,th,ph),y).subs(abs(sin(th)),sin(th)),y)+df(df(f(r,th,ph),z).subs(abs(sin(th)),sin(th)),z))).subs(abs(sin(th)),sin(th)))
```

```python
# 
def lz(f): #angular momentum operator in sph coord
    return -I*(x*df(f,y)-y*df(f,x)).subs([(x,r*sin(th)*cos(ph)),(y,r*sin(th)*sin(ph)),(z,r*cos(th)),(abs(sin(th)),sin(th))])
# 
def lx(f):
    return -I*(y*df(f,z)-z*df(f,y)).subs([(x,r*sin(th)*cos(ph)),(y,r*sin(th)*sin(ph)),(z,r*cos(th)),(abs(sin(th)),sin(th))])
# 
def ly(f):
    return -I*(z*df(f,x)-x*df(f,z)).subs([(x,r*sin(th)*cos(ph)),(y,r*sin(th)*sin(ph)),(z,r*cos(th)),(abs(sin(th)),sin(th))])
```

Оператор $L_{z}$ у сферичних координатах:

```python
simplify(lz(f(r,th,ph)))
```

Оператор $L_{x}$ у сферичних координатах:

```python
simplify(lx(f(r,th,ph)))
```

Оператор $L_{y}$  у сферичних координатах:

```python
simplify(ly(f(r,th,ph)))
```

Оператор $\mathbf{L}^2$ у сферичних координатах

```python
l2x=simplify(lx(lx(f(r,th,ph))))
l2y=simplify(ly(ly(f(r,th,ph))))
l2z=simplify(lz(lz(f(r,th,ph))))
simplify(l2x+l2y+l2z)
```

кутових змінних.<br>

Власні функції оператора \eqref{eq:l2} у сферичних координатах є сферичними функціями, які виражаються через т.з. <a href="https://dlmf.nist.gov/14.2">поліноми Лежандра</a>.<br>
<!--  -->
$$Y_{l,m}(\theta,\varphi) \sim P^{m}_{l}(\cos{\theta})\,e^{i\,m\,\varphi}$$
<!--  -->
які, як витікає з попереднього задовільняють рівнянню (сингулярні при $x=\pm 1$ розвязки т.з. <a href="https://dlmf.nist.gov/14.2"> функції Лежандра 2-го роду</a> $Q^{m}_{n}(x)$ ми відкидаємо):
<!--  -->
$$
z=\cos{\theta}\,\,\to\,\,\,\,\,\frac{d}{d\,z}\left((1-z^2)\,\frac{d\,P^{m}_{l}(z)}{d\,z}\right) +\left(l\,(l+1)-\frac{m^2}{1-z^2}\right)\,P^{m}_{l}(z) = 0
$$
<!--  -->
Поліноми $P^{m}_{l}(z)\,, m\ne 0$ називаються приєднаними поліномами Лежандра оскільки вони виникають з <a href="https://dlmf.nist.gov/14.2">"базових" поліномів Лежандра $P_{l}(z) = P^{0}_{l}(z)$</a>. 
<!--  -->
$$
P^{m}_{l}(z) = (-1)^{m}\,(1-z^2)^{m/2}\,\frac{d^m}{dz^m}\,P_{l}(z)
$$
<!--  -->
Поліноми $P_{l}(z)$ повязани з т.з. мультипольним розкладом кулонівського потенціалу (функції Гріна оператора Лапласа):
<!--  -->
$$\frac{1}{\vert \mathbf{R} - \mathbf{r} \vert} = \frac{1}{R\,\sqrt{1-2\,z\,u+u^2}} = \frac{1}{R}\,\sum\limits_{l=0}^{\infty} P_{l}(z)\,u^l\\
z= \cos{\theta}\,,\,\,\,u = r/R \ll 1
$$
<!--  -->
Використовуючи цю генеруючу функцію можна отримати рекурентні співвідошення:
<!--  -->
$$
(1-2\,z\,u+u^2)\frac{d}{du}
\sum\limits_{l=0}^{\infty}\,P_{l}(z)\,u^{l} =  (1-2\,z\,u+u^2)
\sum\limits_{l=0}^{\infty}\,l\,P_{l}(z)\,u^{l-1} =
\frac{z-u}{\sqrt{1-2\,z\,u+u^2}}=\\
\sum\limits_{l=0} z\,P_{l}(z)\,u^{l} - \sum\limits_{l=0} P_{l-1}(z)\,u^{l} \Rightarrow \\
(2\,l+1)\,z\,P_{l}=(l+1)\,P_{l+1}+l\,P_{l-1}
$$

```python
import matplotlib.pyplot as plt
from matplotlib import cm, colors
from mpl_toolkits.mplot3d import Axes3D
import numpy as np
from scipy.special import sph_harm_y
# 
def sphplot(l,m):
    phi = np.linspace(0, 2*np.pi, 100)
    theta = np.linspace(0, np.pi, 100)
    phi, theta = np.meshgrid(phi, theta)

    # The Cartesian coordinates of the unit sphere
    x = np.sin(theta) * np.cos(phi)
    y = np.sin(theta) * np.sin(phi)
    z = np.cos(theta)
# 
#     m, l = 1, 3
# 
    # Calculate the spherical harmonic Y(l,m) and normalize to [0,1]
    fcolors = np.sqrt(sph_harm_y(l, m, phi,theta).real**2+sph_harm_y(l, m, phi,theta).imag**2)
    fmax, fmin = fcolors.max(), fcolors.min()
    if fmin==fmax: 
        fmin=fmax/1.2
    fcolors = (fcolors - fmin)/(fmax - fmin)
    fig, ax = plt.subplots(subplot_kw={"projection": "3d"})
    ax.plot_surface(x, y, z, rstride=1,cstride=1,facecolors=cm.plasma(fcolors))
    ax.view_init(elev=60)
    ax.set_title(r'$|Y_{l,m}(\theta,\phi)|^2, m=$'+str(m)+', l='+str(l))
    return plt.show()
# 
for n in range(5):
    print("n=",n,": P(n,x) = ",legendre(n,x))
```
