Подальший розвиток ідеї квантування руху мікроскопічних систем дозволив Бору сформулювати
"компромісні" принципи, прийняття яких забезпечує стабільність атомних систем і приводить до емпіричного комбінаційного принципу Рідберга-Ритца. 
Пряме експериментальне підтвердження дискретного обміну енергією між електронами і атомами було продемонстровано у дослідах Франка і Герца по вимірюванню залежності струма електронів у парах ртуті.

В цих дослідах спостерігалась не тільки резонансна природа збудження атомів при зіткненні з електронами певних енергій але і випромінювання, що виникало при переходах збуджених електронів на нижчі рівні. Це узгоджувалось з картиною запропонованою Н. Бором. Застосування ідей теорії коливань для пояснення існування ``квантування класичних орбіт`` руху електронів в атомах в рамках теорії Бора вказував на те, що електрон проявляє хвильові властивості (стояча хвиля - стаціонарний стан). Тому відкриття ефектів інтерференції та дифракції для електронів (та інших частинок мікросвіту) які раніше [асоціювались із хвильовими процесами (свтло)](https://www.youtube.com/watch?v=71Rp-jG6Eek) стало важливим чинником для побудови квантової теорії. 

Історично перші експерименти по прояву хвильових властивостей це спостреження ефектів підсилення та ослаблення інтенсивності реєстрації частинок. При цьому характерні патерни реєстрації дуже нагадували такі як і при спостережені
інтерференційних ефектів фізичної оптики. Канонічним прикладом є [експеримент Юнга по дифракції на двох щілинах](https://www.youtube.com/watch?v=PVyJFzx7zig). Дуже показовий у цьому сенсі [ефект Рамзауэра-Таунсенда (1921 г.)](https://en.wikipedia.org/wiki/Ramsauer–Townsend_effect) - зменшення перерізу розсіяння повільних (1eV) електронів при пружніх зіткненнях з атомами газу з високим потенціалом іонізації (аргон) так що електрон не взмозі передати атомам суттєву енергію (як в дослідах Франка-Герца). Останній ефект аналогичний відомій плямі Пуасона (дифракційний максимум інтенсивності в центрі геометричної тіні). [Експерименти Девіссона-Джермера](https://en.wikipedia.org/wiki/Davisson–Germer_experiment) продемонстрували дифракційні ефекти при відбитті електронів від кристалів. Одна із старих назв квантової механіки - хвильова механіка (картина Шредінгера) -
підкреслює наявність аналогій в методах опису реєстрації квантових частинок і хвиль.

Факт спостереження явища дифракції електронів на щілинах} який дає по суті тий ж самий візерунок як і при реєстрації фотонів дозволяє, за відповідною статистичною інтерпретацією про локалізацію електрона в даній області фотопластинки (М. Борн), висунути припущення про співвідношення між квантовим станом та деяким хвильовим процесом, який описується комплексною функцією -- хвильової функцією протилежно стану класичної системи, яке зображується точкою фазового простору. Власно кажучи, функцію також можна розглядати як точку деякого, взагалі, нескінченновимірного функціонального простору. У подальшому ми з'ясуємо структуру множини станів квантової системи. У квантовій механіці, як постулат, приймається, що хвильова функція саме і є найбільш повним способом опису стану системи.
Зокрема, квадрат модуля хвильової функції (інтенсивність хвилі) визначає густину ймовірності знаходження квантової системи в даному стані. Якщо стан системи можна характеризувати тільки координатою, то:
<!--  -->
\begin{equation}
	\label{eq3/1} p\,(\mathbf{r},t) =
	C\,|\,\psi(\mathbf{r},t)|^{\,2}\,.
\end{equation}
<!--  -->
Це означає, що модуль хвильової функції є експериментально спостережувальною величиною. Матеріальне існування частинки у просторі (в певний момент часу) визначається умовою нормування:
<!--  -->
\begin{equation}
\label{eq4/1} {\int\limits_{V} {p\,(\mathbf{r},t)}} dV =1
\,\,\,,\,\,\,\,C = \left( {{\int\limits_{V} {|\,\psi(\mathbf{r},t)|^{\,2}}} dV} \right)^{ - 1}\,,
\end{equation}
<!--  -->
при виконанні умови:
<!--  -->
\begin{equation}\label{norm}
\int\limits_{V}
{|\,\psi (\mathbf{r},t)\,|^{\,2}} d\,V  < \infty\,.
\end{equation}
<!--  -->
Умова \eqref{norm} називається __умовою квадратичної інтегрованості__. 
Множина всіх квадратично інтегрованих функцій, тобто тих, які задовольняють (\ref{norm}) і можуть бути нормовані, позначається $L_2$.
<!--  -->

Наведена вище ймовірносна інтерпретація стану квантової системи відображує принцип корпускулярно-хвильового дуалізму і приводить до відказу від поняття траєкторії квантової системи у фазовому просторі координат та імпульсів. Відзначимо, що функція $\lambda\psi$ з точки зору ймовірносного розподілу \eqref{eq3/1} описує такий ж самий стан що і $\psi $, і тому співвідношення ''стан - хвильова функція'' не є взаємно однозначним.

Хвильова функція не обов'язково повинна бути функцією координат. Якщо замість фотоплівки, яка фіксує (з певною точністю) координати електронів, використовувати дифракційну гратку (сортування станів
за різними значенням хвильового вектора), то, діючи електростатичним полем у певному напрямку, можна визначити імпульс електрона. Л.~де~Бройль прийняв, як постулат, що стану вільної частинки з імпульсом $$\mathbf{p} = \hbar \,\mathbf{k}\,,\,\,\,|\mathbf{k}| = \frac{2\pi}{\lambda}$$ 
відповідає хвильовий процес, який описується плоскою хвилею виду:
<!--  -->
\begin{equation}\label{eq:plainwave} 
\psi_{\mathbf{p}} ({\rm r},t) ={\frac{1}{{(2\pi \hbar)^{n / 2}}}}\exp \left( \frac{i\mathbf{p} \cdot
\mathbf{r}-i\,E\,t}{\hbar}  \right) = \frac{1}{{(2\pi \hbar)^{n / 2}}}\,
\exp \left(i\,\mathbf{k}\cdot\mathbf{r} - i\,\omega\,t \right)\,,
\end{equation}
<!--  -->
де $n$ - кількість просторових ступенів вільності частинки ($n=1,2,3$). Функції \eqref{eq:plainwave} відповідає постійна густина ймовірності, що узгоджується з уявленням про просторову однорідність (трансляційну інваріантість) вільного руху. Довжина хвилі яка відповідає імпульсу $p$ називається де бройлівською довжиною і визначає дифракційні властивості потоку таких частинок за аналогією зі світловими хвилями.

Тож, стан квантової системи може бути описаний не з точки зору ймовірності просторового положення, а з точки зору ймовірності знаходження у стані з певним імпульсом (за допомогою
дифракційної гратки). Такий дуальний характер опису стану у термінах пари ''імпульс-координата'' знаходить своє математичне відображення у співвідношеннях:
<!--  -->
\begin{equation}\label{eq6/1} 
\psi (\mathbf{r},t) = {\int\limits_{\Lambda}{c(\mathbf{p},t)}}\, \psi_{\mathbf{p}} (\mathbf{r},t)d\,\mathbf{p}
\end{equation}
<!--  -->
та
<!--  -->
\begin{equation}\label{eq7/1} 
c(\mathbf{p},t) = \int\limits_{V} \psi(\mathbf{r},t)\,\psi^*_{\mathbf{p}}(\mathbf{r},t)d\,V\,.
\end{equation}
<!--  -->
Вони відображують дуальний характер опису квантової системи у термінах або координат чи імпульсу.
\begin{equation}
\int\limits_{V}
{|\,\psi (\mathbf{r},t)\,|^{\,2}} d\,V  = \int\limits_{V}
{|\,c(\mathbf{p},t)\,|^{\,2}} d\,\Lambda
\end{equation}

Відзначимо, що у необмеженному просторі \eqref{eq:plainwave} умова \eqref{eq4/1} не виконується. Це також проявляється у тому що для хвиль де Бройля інтеграл
$$\int |\psi_{\mathbf{p}}(\mathbf{r},t)|^2\, dV$$
розбігається. Накладання умови обмеженості об'єму знімає цю проблему, однак при цьому рух частинки вже не може розглядатися як вільний, оскільки залежить від умов на межі об'єму.
