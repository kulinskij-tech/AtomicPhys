## Гамільтоніан та основний стан He 
<!--  -->
Продемоструємо основні принципи класифікації станів багатоелектронних атомів, на прикладі найпростішого з багатоелектронних атомів - атома гелію (Z=2). Його гамільтоніан в електростатичному наближенні:
<!--  -->
\begin{equation}\label{eq:hamhe}
 H = H_{1} + H_{2} + \frac{e^{\,2}}{\vert \mathbf{r}_{1} - \mathbf{r}_{2} \vert}  \\
 H_{i} = -\frac{\hbar^2}{2\,m}\Delta_{i} - 2\frac{e^{\,2}}{\vert \mathbf{r}_{\,i}
 \vert}
\end{equation}
<!--  -->
проте точного розв'язку для спектра такого гамільтоніану не існує.
Доводиться користуватись наближенними методами та якісними міркуваннями для пояснення спектроскопічних даних. Як приклад, оцінимо енергію основного стану атома гелія використовуючи відомі результати для атома водня. З вигляду гамільтоніану очевидно, що енергія складається з енергії притягання електронів до ядра (<0) та їх енергії відштовхування (>0):
<!--  -->
\begin{equation}\label{he_e0}
  E^{(0)}_{He} = \langle\hat{H}\rangle = -8\,E_{0} + e^2
\left\langle\, \frac{1}{r} \,\right\rangle = -8\,E_{0}+2\,E_{0}= -6\,E_{0} = - 81.6 \, \text{eV}
\end{equation}
<!--  -->
Для оцінки енергії кулонівського відштовхування ми скористались теоремою віріала.
Експериментальне значення $E^{(0)}_{He} = -79.1\,\text{eV}$.
Завищення оцінки пов'язане з тим, що електрон
рухається у полі заряда ядра та поля іншого електрона що частково екранує заряд ядра $Z=+2$ і отже трохи зменшує енергію притягання. Енергія однократної іонизації може бути розрахована на основі наступного. При іонизації електрон покидає область де він рухався у полі ядра $Z=2$ на нескінченність, де з врахуванням екраніровки іншим електроном $Z=1$. Це можно описати за допомогою певного ефективного значення $1<Z_{eff}<2$ (див. нижче). Оскільки $E\propto Z_{eff}^2$, то для оцінки скористаємось середнім геометричним між двома асимптотиками що враховує те що внесок області сильного екранування ($Z\to 1$) більше ніж ближньої області ($Z\to 2$) завдяки шивдкому зменшенню електронної хвильової функції зв'язаного стану. Тоді отримаємо:
<!--  -->
$$E_{1}\approx -\sqrt{E_{Z=2}\,E_{Z=1}}  = -2E_0= -27.2\,\text{eV}\,.$$
<!--  -->
Експериментальний результат $E^{(exp)}_{1} = -24.6\,\text{eV}$.

## Два електрони: спочатку синглет і триплет

Для атома гелію спочатку додаємо орбітальні та спінові моменти електронів окремо:

$$\mathbf L=\mathbf l_1+\mathbf l_2,\qquad \mathbf S=\mathbf s_1+\mathbf s_2.$$

Два спіни $s_1=s_2=1/2$ дають синглет $S=0$ або триплет $S=1$. Для різних ортонормованих орбіталей $a,b$ просторові функції можна записати як

$$\Phi_\pm(1,2)=\frac{\phi_a(1)\phi_b(2)\pm\phi_a(2)\phi_b(1)}{\sqrt2}.$$

Повна хвильова функція електронів має бути антисиметричною. Тому симетрична просторова частина поєднується з антисиметричним спіновим синглетом, а антисиметрична — із симетричним спіновим триплетом:

$$\Psi_{S=0}=\Phi_+\chi_{00},\qquad \Psi_{S=1,M_S}=\Phi_-\chi_{1M_S},$$

$$\chi_{00}=\frac{\uparrow_1\downarrow_2-\downarrow_1\uparrow_2}{\sqrt2},\quad
\chi_{11}=\uparrow_1\uparrow_2,\quad
\chi_{10}=\frac{\uparrow_1\downarrow_2+\downarrow_1\uparrow_2}{\sqrt2},\quad
\chi_{1,-1}=\downarrow_1\downarrow_2.$$

Це, зокрема, пояснює основний стан $1s^2\,{}^{1}S_0$: для двох однакових просторових орбіталей антисиметрична просторова комбінація дорівнює нулю, тож триплет заборонений принципом Паулі. Один визначник із різними орбіталями та протилежними спінами сам по собі не є чистим синглетом.

У конфігурації $1s2s$ можливі обидва спіни. У наближенні фіксованих орбіталей їхні енергії мають вигляд

$$E_{S=0}=E_a+E_b+D+K,\qquad E_{S=1}=E_a+E_b+D-K,$$

де $D$ — прямий кулонівський інтеграл, а $K$ — обмінний. Різниця $2K$ виникає вже з електростатичної взаємодії та вимоги антисиметрії, без спін-орбітального внеску. Для розглядуваної конфігурації гелію триплет лежить нижче за синглет.[^1]

## Потім — LS-зв'язок і тонка структура

У легкому атомі електростатична взаємодія переважає над релятивістськими внесками. Спочатку вона формує терми з певними $L,S$, а слабша внутрішня взаємодія зв'язує ці моменти у повний момент:
<!--  -->
$$\mathbf J=\mathbf L+\mathbf S,\qquad J=|L-S|,|L-S|+1,\ldots,L+S.$$
<!--  -->
| Конфігурація | $L$ | $S$ | Рівні терму |
|---|---:|---:|---|
| $1s2s$ | 0 | 0 | $^{1}S_0$ |
| $1s2s$ | 0 | 1 | $^{3}S_1$ |
| $1s2p$ | 1 | 0 | $^{1}P_1$ |
| $1s2p$ | 1 | 1 | $^{3}P_0,{}^{3}P_1,{}^{3}P_2$ |

У $1s2p$ спочатку відокремлюються терми $^{1}P$ і $^{3}P$. Усередині триплетного терму спін-орбітальний зв'язок якісно пояснює появу трьох рівнів із різними $J$. Синглет має лише $J=L=1$. Для $1s2s$ орбітальний момент дорівнює нулю, і цей механізм не дає додаткових рівнів усередині кожного терму.

<img src="./figs/fig_he_ls_hierarchy.png" alt="Конфігурація 1s2p: електростатичне розділення синглету і триплету, потім рівні з різними J" width="850">

*Рисунок показує послідовність класифікації, а не масштаб або порядок реальних енергій He. Після тонкого розщеплення зберігається виродження $2J+1$ за $M_J$.*

Проста модель $\mathbf L\cdot\mathbf S$ корисна для розуміння LS-класифікації, але не визначає точні інтервали тонкої структури гелію: суттєвими є також спін-спінові та спін-чужоорбітальні внески. Не слід ототожнювати синглет-триплетне розділення з тонким розщепленням.[^2]

Синглетні стани називають парагелієм, триплетні — ортогелієм. В LS-наближенні електричні дипольні переходи зберігають $S$, тому переходи між цими групами сильно пригнічені; слабке змішування станів може робити їх можливими.[^3]
