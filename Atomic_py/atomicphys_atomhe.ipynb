## Гамільтоніан та основний стан He 
<!--  -->
Продемоструємо основні принципи класифікації станів багатоелектронних атомів, на прикладі найпростішого з багатоелектронних атомів - атома гелію (Z=2). Його гамільтоніан в електростатичному наближенні:
<!--  -->
\begin{equation}\label{eq:hamhe}
 H = H_{1} + H_{2} + \frac{e^{\,2}}{\vert \mathbf{r}_{1} - \mathbf{r}_{2} \vert}  \\
 H_{i} = -\frac{\hbar^2}{2\,m}\Delta_{i} - 2\frac{e^{\,2}}{\vert \mathbf{r}_{\,i}
 \vert}
\end{equation}
<!--  -->
проте точного розвязку для спектра такого гамільтоніану не існує.
Доводиться користуватись наближенними методами та якісними міркуваннями для пояснення спектроскопічних даних. Як приклад, оцінимо енергію основного стану атома гелія використовуючи відомі результати для атома водня. З вигляду гамільтоніану очевидно, що енергія складається з енергії притягання електронів до ядра (<0) та їх енергії відштовхування (>0):
<!--  -->
\begin{equation}\label{he_e0}
  E^{(0)}_{He} = \langle\hat{H}\rangle = -8\,E_{0} + e^2
\left\langle\, \frac{1}{r} \,\right\rangle = -8\,E_{0}+2\,E_{0}= -6\,E_{0} = - 81.6 \, \text{eV}
\end{equation}
<!--  -->
Для оцінки енергії кулонівського відштовхування ми скористались теоремою віріала.
Експериментальне значення $E^{(0)}_{He} = -79.1\,\text{eV}$.
Завищення оцінки пов'язане з тим, що електрон
рухається у полі заряда ядра та поля іншого електрона що частково екранує заряд ядра $Z=+2$ і отже трохи зменшує енергію притягання. Енергія однократної іонизації може бути розрахована на основі наступного. При іонизації електрон покидає область де він рухався у полі ядра $Z=2$ на нескінченність, де з врахуванням екраніровки іншим електроном $Z=1$. Це можно описати за допомогою певного ефективного значення $1<Z_{eff}<2$ (див. нижче). Оскільки $E\propto Z_{eff}^2$, то для оцінки скористаємось середнім геометричним між двома асимптотиками що враховує те що внесок області сильного екранування ($Z\to 1$) більше ніж ближньої області ($Z\to 2$) завдяки шивдкому зменшенню електронної хвильової функції зв'язаного стану. Тоді отримаємо:
<!--  -->
$$E_{1}\approx -\sqrt{E_{Z=2}\,E_{Z=1}}  = -2E_0= -27.2\,\text{eV}\,.$$
<!--  -->
Експериментальний результат $E^{(exp)}_{1} = -24.6\,\text{eV}$.

## Класифікація рівнів атома гелію. Орто- і парагелій.
Стаціонарні стани атому гелію як двохелектроної системи згідно з правилом додавання моменту можуть бути двох типів: синглетні з $S = 0$ або триплетні з $S = 1$.
Основний стан атома гелію є синглетним $S = 0$, триплетний стан $S = 1$ заборонений завдяки принципу Паулі (оскільки він містить спінові стани $\uparrow\,\uparrow$, $\downarrow\,\downarrow$). 
Хвильова функція системи з двох електронів у конфігурації $1s^2$ (в сенсі самоузгодженного поля): 
<!--  -->
\begin{equation}\label{eq:psihe0}
\Psi = \frac{1}{\sqrt{2}}\left(\psi_{1,0,0,\uparrow}(r_1)\,\psi_{1,0,0,\downarrow}(r_2) -\psi_{1,0,0,\uparrow}(r_2)\,\psi_{1,0,0,\downarrow}(r_1)\right) = \frac{1}{\sqrt{2}}
\begin{vmatrix}\psi_{1,0,0,\uparrow}(r_1) &  \psi_{1,0,0,\uparrow}(r_2)
\\
\psi_{1,0,0,\downarrow}(r_1) &  \psi_{1,0,0,\downarrow}(r_2)
\end{vmatrix}
\end{equation}
<!--  -->
де $\psi_{1,0,0}(r)$ це хвильова функція (х.ф. одночастинкового стана у самоузгодженому полі) яку наближенно може взяти як х.ф. воднеподібного стана з $n=1, l=0, m=0$:
<!--  -->
$$ \psi_{1,0,0}(r)  = \frac{1}{\sqrt{\pi\,r^3_0}}\,e^{-r/r_0}$$
<!--  -->
де $r_0$ - підгоночний (під експериментальні дані) параметр - аналог борівської орбіти. Видно, що узгодженність із принципом Паулі здійснена за рахунок різниці в проекціях спінів. 
Наступний збуджений стан з електронною конфігурацією $1s^1\,2s^1$ може бути або синглетним або триплетним, оскільки тепер вже обмежень на конфігурацію спінів немає (зайняті різні одноастинкові стани). Зрозуміло, що з врахуванням проекції спіна електронів є 4 варіанти які відповідають одному синглетному та трьом станам що утворюють триплет (4=1+3). Х.ф. синглетного стана:
<!--  -->
\begin{equation}\label{eq:psihe1}
\Psi_{S=0} = \frac{1}{\sqrt{2}}\left(\psi_{1,0,0,\uparrow}(r_1)\,\psi_{2,0,0,\downarrow}(r_2) -\psi_{1,0,0,\uparrow}(r_2)\,\psi_{2,0,0,\downarrow}(r_1)\right) = \frac{1}{\sqrt{2}}
\begin{vmatrix}\psi_{1,0,0,\uparrow}(r_1) &  \psi_{1,0,0,\uparrow}(r_2)
\\
\psi_{2,0,0,\downarrow}(r_1) &  \psi_{2,0,0,\downarrow}(r_2)
\end{vmatrix}
\end{equation}
Всі х.ф. триплених станів що відповідають $S = 1$ ми тут будувати не будемо (це буде зроблено у курсі квантової механіки), наприклад х.ф. стану із однаковими проекціями спіну:
<!--  -->
\begin{equation}\label{eq:psihe2}
\Psi_{S=1} = \frac{1}{\sqrt{2}}\left(\psi_{1,0,0,\uparrow}(r_1)\,\psi_{2,0,0,\uparrow}(r_2) -\psi_{1,0,0,\uparrow}(r_2)\,\psi_{2,0,0,\uparrow}(r_1)\right) = \frac{1}{\sqrt{2}}
\begin{vmatrix}\psi_{1,0,0,\uparrow}(r_1) &  \psi_{1,0,0,\uparrow}(r_2)
\\
\psi_{2,0,0,\uparrow}(r_1) &  \psi_{2,0,0,\uparrow}(r_2)
\end{vmatrix}
\end{equation}

Найменшу енергію має стан з максимальним значенням $S$, тобто триплетний стан $1s^1\,2s^1\, ^{3}S_{1/2}$. Синглетний стан (терм $^{1}S_{1/2}$) має більшу енергію. Це повязано із різним розподілом заряду в такому стані так що середня енергія відштовхування у такому стані більша за таку у триплетному. Триплетний стан вироджений за проекцією спіна і відповідно розщеплюється зовнішнім магнітним полем на три енергетичних рівня.

Для наступної ел. конфігурації $1s^1\,2p^1$ ($L=1$) можливі $2\times 2\times 3 = 12$ станів які утворюють терми з $S=1=L \Rightarrow J = 0,1,2$: $^{3}P_{0,1,2},$ та $S=0,L=1 \Rightarrow J=1$: $^{1}P_{1},$ (разом $\underbrace{1+3+5}_{S=1=L}+\underbrace{3}_{S=0} = 12$). Знов триплетні стани лежать нижче за синглетних (правило Хунда: для даної електронної конфігурації найнижчу енергією має терм з максимальним значенням $S$ і найбільшим (при даному $S$) значенням $L$). Таким чином, енергетичні рівні багатоелектронного атому характеризуються значеннями спінового $L$, орбітального $L$ та повного $J$ моментів.

Якщо нехтувати спіновими взаємодіями (як наприклад спін-орбітальною взаємодію), то переходи між станами з різними значеннями спінового квантового числа не можливі (спін - інтеграл руху). Враховуючи слабкість спін-залежних взаємодій (релятивистських за своєю природою) у випадку атома гелію, можна представити цю ситуацію як наявність двох квазистаціонарних станів гелію: __парагелій__ - синглетні стани, __ортогелій__ - триплетні стани. Вони різняться своїми властивостями: __парагелій є діамагнетиком__ (спін 0), __ортогелій - парамагнетик__ (спін 1) має власний магнітний момент.

Врахування спін-орбітальної взаємодії призводить до розщеплення триплетних станів з $L\ne0$. Стани ж з $L=0$ розщеплюються тільки зовнішним магнітним полем (наприклад магнітним полем ядра - надтонке розщеплення).

Основні висновки, що можна зробити з попереднього розгляду такі:
- стаціонарні стани атома у електростатичному наближенні можна характеризувати значеннями повного моменту імпульса  $J$, а також орбітального $L$ та спінового $S$ квантових чисел. Це носить назву атомного терма;
- даній електронній конфігурації (заповнення одноелектроних станів) відповідають декілька термів у відповідності до принципу Паулі;
- енергії таких термів відрізняються за рахунок розщеплення обмінного виродження, оскільки вигляд хвильової функції залежить від повного спіна $S$;
- атомний терм з даними $L,S$ вироджений за проекціями $L_z, S_z$ ($(2L+1)(2S+1)$-кратність) розщеплюється або з врахуванням спін-залежних релятивістських взаємодій (спін-орбітальна, або спін-спінова, при цьому $2J+1$-виродженність по проекції повного моменту $M_J$ залишається),  або у зовнішніх полях, що порушують сферичну симетрію гамільтоніану атома (знімається виродженність по $M_J$).
