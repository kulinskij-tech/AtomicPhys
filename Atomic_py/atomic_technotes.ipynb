## Технічне метод забезпечення
Посібник оформлений як колекція <a href="https://en.wikipedia.org/wiki/Project_Jupyter"> Jupyter</a>  notebooks ([version 6.4.13](https://github.com/jupyter/notebook/tree/6.4.x)+[nbextensions](https://jupyter-contrib-nbextensions.readthedocs.io/en/latest/install.html) for LaTeX formula enumeration etc) з використанням <a href="https://www.python.org/">Python</a> пакетів <a href="https://scipy.org"> Matplotlib, NumPy, SciPy, SymPy</a>.
