## Спектр молекул. Адіабатичне наближення.

Молекула є ще більш складною систему ніж атом. 
Нагадаємо перш за все, що щодо природи хімічного зв'язку розрізняють два типи молекул: гомополярні та гетерополярні. Гетерополярні молекули у полярних розчинниках
(наприклад, у воді) розпадаються на іони. Характерним прикладом є молекула NaCL
Схема утворення цієї молекули полягає в тому що в атомі Na є слабко пов'язаний валентний електрон, а в атомі Сl не вистачає одного електрона для утворення стійкої електронної оболонки аргону. При наближенні цих атомів на достатню відстань електрон $Na$ займає вільне місце на оболонці Сl і іони Na$^+$ і Cl$^{-}$, що утворюються, утримуються електростатичними силами. В принципі ця картина зрозуміла і з суто класичної т.з. як притягнення протилежних зарядів. 

Пояснити ж гомополярний звязок що виникає між однаковими атомами типу молекул $H_2, N_2, O_2$ на основі класичної фізики неможливо, оскільки суттєвим чинником є необхідність врахування спіна та принципа Паулі для електронних станів.

Гамільтоніан молекулярної системи складається з взаємодій ядро-ядро, ядро-електрон і електрон-електрон. Однак, той факт, що маси ядер у тисячі разів більше ніж маса електронів, приводить до того, що у нульовому наближенні можна розділити ядерні поступові і електронні степені свободи. Таке наближення, при якому ядра розглядаються як нерухому а електрони рухаються у поле нерухомих ядер, називається __адіабатичним__ (у застосуванні до молекул - наближення __Борна-Оппенгеймера__). В такому наближенні молекулярну систему можна розглядати як складену з двох підсистем: ядерної і електронної. Положення ядер $\mathbf{R}_i\,,i=1,\ldots,N$ розглядаються як параметри від яких залежить спектр енергії електронної підсистеми $\varepsilon(\mathbf{R}_1,\ldots,\mathbf{R}_N)$. В адіабатичному наближенні рівноважна конфігурація ядер для даного стану електронної підсистеми визначається з умови мінімуму для енергії електронної підсистеми:
<!--  -->
\begin{equation}
\label{feq9} \nabla_{\mathbf{R}_{k}}\varepsilon
({\mathbf{R}}_{k}^{0} ) = 0\,\,\,\,,k = 1,\ldots, N
\end{equation}
<!--  -->
Зрозуміло, що при невеликих збудженнях основного стану (рівні з малими квантовими числами) енергія молекули в адіабатичному наближенні може бути представлена як сума енергій електронної підсистеми і енергії руху ядер. 

Аналіз електронних молекулярних термів на основі інтегралів руху можливий тільки для молекул, конфігурації яких мають елементи симетрії (вісі, площини симетрії, віддзеркалення і інш.). Продемонструємо такий підхід на прикладі двухатомних молекул. Такі молекули мають очевидну вісь симетрії. Цьому елементу аксіальної симетрії відповідає збереження проекції моменту імпульсу електронної підсистеми на цю вісь. Він позначається $\Lambda$. Нехтуючи спіновими взаємодіями повний спін електронної підсистеми також є інтегралом руху (згадаємо атом гелію та його синглетні та триплетні стани), тому молекулярні терми характеризуються значенням $S$. Значення проекції моменту $\Lambda = 0,1,2,\ldots$ позначаються по аналогії з атомними термами літерами $\Sigma \,,\,\Pi \,,\,\Delta \,,\,\ldots$ вказуючи мультиплетність $2S + 1$, наприклад: $^{3}\Sigma\,,\,^{1}\Pi$ і інш.. Крім того, електронна густина не змінюється відносно віддзеркалень в площині яка містить вісь
симетрії. При цьому проекція моменту змінює знак (момент імпульса - псевдовектор), тому має місце двократне виродження енергії, оскільки стани з протилежними проекціями мають однакову енергію, виключаючи $\Sigma $ терм ($\Lambda $=0). Однак, хвильова функція такого стану може змінювати знак, тому стани $\Sigma$-терма можуть бути або парними $\Sigma^{+} $ або
непарними $\Sigma^{-}$. Якщо молекула складається з однакових атомів, то з'являється додатковий елемент симетрії - перетворення інверсії координат електронів, відносно центра симетрії (проекція моменту не змінює знак завдяки аксиальності). Цьому перетворенню відповідає ще одне квантове число типа парності: $\Lambda _{g} $ - парні, $\Lambda _{u} $ - непарні терми. Основний стан для
більшості двохатомних молекул є повністю симетричним з рівним 0 повним спіном ${}^{1}\Sigma^{+} $ чи ${}^{1}\Sigma _{g}^{+} $ для однакових атомів. Виключеннями є О$_{2}$ - терм ${}^{3}\Sigma
_{g}^{ -} $ і NO - терм ${}^{\,2}\Pi $.

### Коливальні (вібронні) та обертальні (ротаційні) спектри молекул

Енергія руху ядерної підсистеми складається з енергії коливальних і обертальних ступенів свободи (у випадку малих енергій ці степені свободи можна враховувати як незалежні):

\begin{equation}
\label{feq10} E = \varepsilon _{m} ({\mathbf{R}}_{k}^{0} ) + E_{osc}  +
E_{rot}
\end{equation}

Коливальні рухи можна описувати на основі моделі гармонічного осциллятора застосовуючи відому з класмех концепцію нормальних коливань ($3N-6(5 \text{для лінійних молекул})$):

\begin{equation}
\label{eq_osc} E_{osc} = 
\sum\limits_{s} \hbar\, \omega_{m\,s} (n_{s} + 1 / 2)
\end{equation}

Обчислення енергії обертального руху засновується на розгляді гамільтоніана:

\begin{equation}
\label{feq11} H_{rot} = \frac{M_{1}^{\,2}}{2\,I_{1}} +
\frac{M_{2}^{\,2}}{2\,I_{2}} + \frac{M_{3}^{\,2}}{2\,I_{3}}
\end{equation}

де $1,\,2,\,3$ позначають компоненти відповідних величин на головні вісі інерції (система відліку, яка пов'язана з молекулою).  Обчислення власних значень (\ref{feq11}) може буте здійснено або у випадку квантового ротатора $I_{1} = I_{2} = I_{3} $ або лінійної молекули:

\begin{equation}
\label{feq13} 
H_{rot} = \frac{\mathbf{M}^2}{2\,I} \Rightarrow
E_{rot} = {\frac{{\hbar^{\,2}J(J + 1)}}{{2\,I}}}
\end{equation}

Оцінимо відстань між відповідними рівнями по безрозмірному параметру $m / M$. Очевидно, що відстань між електронними рівнями асимптотично не залежить від цього параметра. Відстань між коливальними рівнями пропорційна $\Delta E_{osc} \sim \omega \sim {\frac{1}{{\sqrt {M}}} }\sim \sqrt {{\frac{{m}}{{M}}}} $. Для обертальних рівнів маємо $\Delta E_{rot} \sim {\frac{1}{{M}}}\sim
{\frac{{m}}{{M}}}$. Таким чином:

\begin{equation}
\label{feq14} \Delta \varepsilon \gg \Delta E_{osc} \gg \Delta
E_{rot}
\end{equation}

Застосування вказаного наближення визначається малістю амплітуди коливань ядер у порівнянні з розмірами молекули:

\begin{equation}
\label{feq15} \frac{{\sqrt {\langle\,r^{\,2}\,\rangle}}}{d}\sim
\left(\frac{m}{M}\right)^{1 / 4} \ll 1
\end{equation}

Таким чином молекулярні спектри разом з переходами між електронними станами (оптичний діапазон) містять також лінії що відповідають переходам між [коливальними станами - коливальна спектроскопія](https://en.wikipedia.org/wiki/Infrared_spectroscopy) (інфрачервоний спектр), [станами обертання - обертальна спектроскопія](https://en.wikipedia.org/wiki/Rotational_spectroscopy) (мікрохвильове випромінювання) та переходами що обумовлені їх взаємодією [ротовібронна спектроскопія](https://en.wikipedia.org/wiki/Rotational%e2%80%93vibrational_spectroscopy).

В результаті молекулярний спектр складається з відносно чітких ліній електронних переходів  між якими йдуть смуги вібраційних переходів  між ними (ще більш щільно) смуги обертальних переходів. Молекулярний спектр - лінійчасто-смугастий на відміну від атомарного (лінійчастий).

Молекулярні коливання взаємодіють з електромагнітним випромінюванням завдяки зміні дипольного моменту молекули. Чим більше зміна дипольного моменту, тим сильніше речовина поглинає падаюче випромінювання. Отже сполуки з полярними зв'язками типа O-H і C-O, мають лінії випромінювання-поглинання у інфрачервоному спектрі за якими їх ідентифікують. З цього зрозуміло, що неполярні  молекули N$_2$ і O$_2$ мають нульовий дипольний момент і є "інфрачервоно пасивними". Правило відбору для коливальних переходів у гармонічному наближенні $\Delta n = \pm 1$. З врахуванням ангармонізмів зміна $n$  можлива з більшою різницею $\pm 2, \pm 3\, \ldots$.

<img src="./figs/fig_rotovibr.png" alt="Image" width="400">

Суто обертальні переходи (без зміни стана електонної системи) здійснюються між станами $|J-J'|=1$ ядерної підсистеми. Відповідні частоти наприклад переходу випромінювання:
$$\omega = \frac{E_{rot}(J+1)-E_{rot}(J)}{\hbar} = \frac{\hbar}{2\,I}\,(J+1)$$

### Природа гомополярного зв'язку
Крім загальної характеристики молекулярних спектрів квантова механіка дозволила з'ясувати механізм утворення молекул на основі поняття валентності. Необхідною умовою утворення молекули є взаємне доповнення електронних оболонок атомів, тобто вільні одночастинкові стани одного атома повинні заповнюватися за рахунок наявності заповнених у іншому. Зрозуміло, що при цьому основну роль знову відіграє принцип Паулі. 

Продемонструємо це на прикладі молекули водню, яка дає приклад гомополярного зв'язку (однакові атоми), на відміну від гетерополярного зв'язку (різні атоми). Розглянемо два атома водню в основному стані ($1s^{1}\,^{\,2}S)$. При їх зближенні, згідно правила складення моментів можуть бути отримані терми основного стана ${}^{1}\Sigma _{g}^{+}  $ чи ${}^{3}\Sigma _{u}^{+}  $. Дійсно, спінова частина хвильової функції синглетного стани є антисиметричною, тому координатна частина повинна бути парною (g) і не мати вузлів. Спінова частина хвильової функції триплетного навпаки симетрична, а координатна антисиметрична (u). Очевидно, що основний стан відповідає терму ${}^{1}\Sigma _{g}^{ + } $. Дійсно, у даній задачі ми маємо справу з типовим, для систем, які складаються з однакових частинок,
обмінним виродженням. Координатні частини хвильових функцій, правильного нульового наближення, які побудовані на незбурених хвильових функціях дорівнюють:

\begin{equation}
\label{feq16}
\begin{array}{l}
 \psi _{\pm}  = C\left( {\varphi _{1} (\mathbf{1})\varphi _{2}
 (\mathbf{2})\pm \varphi _{1} (\mathbf{2})\varphi _{2} (\mathbf{1})}
\right) \\
 C = {\frac{1}{{\sqrt {2(1\pm N^{\,2})}}} } \\
 N = \int {\varphi _{1} ({\mathbf{r}})\varphi _{2}
 ({\mathbf{r}})\,d{\mathbf{r}}} \\
 \end{array}
\end{equation}

Тут враховано те, що хвильові функції основних станів можна обрати дійсними. Симетрична (координатна) хвильова функція відповідає синглетному стану (S=0), антисиметрична - триплетному (S=1). Рівні енергії таких станів визначаються як середні від оператора енергії кулонівської взаємодії:

\begin{equation}
\label{feq17} E_{\pm}  = 2E_{0} + \left( {{\frac{{e^{\,2}}}{{R}}} +
Q} \right)\pm A \mp N^{\,2}{\frac{{Q\pm A}}{{1\pm N^{\,2}}}}
\end{equation}

де

\begin{equation}
\label{feq18}
\begin{array}{l}
 Q(R) = e^{\,2}\int {\varphi _{1}^{\,2}(\mathbf{1})\,\varphi _{2}^{\,2} (\mathbf{2}){\left[ {{\frac{1}{{\vert \mathbf{1} - \mathbf{2}\vert}} } - {\frac{1}{{\vert \mathbf{1} - R\vert}} } -
{\frac{1}{{\vert \mathbf{2} - R\vert}} }}
\right]}\,d\,\mathbf{1}\,d\,\mathbf{2}} \\
 A(R) = e^{\,2}\int {\varphi _{1}(\mathbf{1})\,\varphi _{2}(\mathbf{2}){\left[ {{\frac{1}{{\vert \mathbf{1} - \mathbf{2}\vert}}
} - {\frac{1}{{\vert \mathbf{1} - R\vert}} } - {\frac{1}{{\vert
\mathbf{2} - R\vert}} }} \right]}\varphi _{2} (\mathbf{1})\varphi
_{1} (\mathbf{2})d\,\mathbf{1}\,d\,\mathbf{2}}
\\
 \end{array}
\end{equation}

Другий доданок у дужках (\ref{feq17}) дає ван дер ваальсівську взаємодію (притягнення) на великих відстанях, де обмінні доданки експоненціально малі. Наявність обмінного вкладу є суттєвим на близьких відстанях і приводить до того, що для енергії синглетного терму є глибокий мінімум, який відповідає зв'язаному стану, тобто молекулі водню. Енергія триплетного терму в цієї області монотонна. Тому утворення молекули ${\rm H}_{2}$ обумовлене типово квантовим обмінним ефектом (утворення пари електронів з протилежними спінами).
