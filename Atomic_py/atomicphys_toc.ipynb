## Експериментальна карта курсу

Теоретичні моделі курсу зручно читати разом із експериментами, які зробили відповідні поняття необхідними. Фотоелектричний ефект, випромінювання чорного тіла, Комптон-ефект, дослід Франка-Герца, досліди Резерфорда, Девіссона-Джермера, Штерна-Герлаха, ефекти Штарка і Зеємана та правило Мозлі утворюють історичний міст від класичної фізики до квантового опису атома.


__Мета:__ вивчення основних положень атомної теорії та методів розв'язків задач; формування навичок аналізу квантових явищ.

__Завдання:__ засвоєння основних положень квантової теорії та вдосконалення навичок рішення задач. В результаті вивчення навчальної дисципліни студент повинен:
- Знати: основні положення атомної теорії.
- Вміти:
  - аналізувати атомні явища використовуючи базові положення і основні методи
  - розв'язувати типові задачі
  - представляти результати самостійної роботи із використанням сучасних засобів презентації наукових результатів (Jupyter notebook).
