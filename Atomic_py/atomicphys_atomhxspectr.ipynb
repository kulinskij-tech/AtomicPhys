Рентгенівські промені [Wilhelm Conrad Röntgen](https://www.nobelprize.org/prizes/physics/1901/summary/) дозволяють досліджувати структури атомних масштабів ($\sim 10^{-2} - 10^2\,\overset{\circ}{A}$) і є одним з найефективніших експериментальних інструментів для дослідження квантових властивостей матерії. Як вже зазначалось раніше ці промені утворюються або при гальмуванні енергетичних електронів (гальмівне випромінювання), або при збудженні електронів внутрішних оболоньок (характеристичне випромінювання). Спект гальмівного випромінювання є неперервним і характеризується коротхохвильовою межею - мінімальна довжина хвилі відповідає перетвореню енергії електрона в енергію рентгенівського фотона:
<!--  -->
$$
e\,V = h\,\nu_{max}\Rightarrow \lambda_{min} = \frac{h\,c/e}{V}
\quad \label{eq:xrayv}
$$
<!--  -->
Ця границя не залежить від типу речовини аноду(антикатоду) і визначається тільки різницею потенціалів яка прискорює електрони. Спектр [характеристичного випромінювання](https://en.wikipedia.org/wiki/Characteristic_X-ray) навпаки є лінійчатим, тобто складається з окремих ліній і залежить від речовини так що спектр такого випромінювання є паспортною характеристикою за якою цю речовину можна ідентифікувати.

З цієї т.з рентгенівські і оптичні спектри аналогічні, але між ними є суттєві відмінності:
- рентгенівські спектри різних елементів мають однакову структуру, тоді як структура оптичних спектрів різних елементів суттєво різниться. 
- оптичні спектри поглинання складаються з окремих ліній, що збігаються з лініями випромінювання головної серії відповідного елемента. Рентгенівські спектри поглинання не схожі на рентгенівські спектри випромінювання та складаються з декількох смуг з різким довгохвильовим краєм.
- частота рентгенівського випромінювання в тисячі разів більше, ніж частота оптичного випромінювання. Це означає, що енергія рентгенівського кванта в тисячі разів більше енергії оптичного кванту.

Ці особливості пов'язані з природою переходів між станами які породжують характеристичне випромінювання і сруктурою електронних оболонок викладених у попередніх  розділі. Якщо оптичні переходи повязанані з слабкозвязаними (валентними) електроннними станами зовнішніх оболонок, то характеристичні рентгенівські переходи відбуваються завдяки електронним станам внутрішніх оболонок. Оскільки внутрішні оболоньки практично не змінюються навіть якщо атом елемента входить до якоїсь молекулярної сполуки, то характеристичне рентгенівське випромінювання слугує надійним інструментом ідентифікації хімічної будови речовини.

Електрон (або фотон) з достатньою енергією ($>100 eV$) що зіткається з атомами анода може вибити електрон з однієї з внутрішніх оболонок атома ($K (n=1)$, $L (n=2),$ $M (n=3),\ldots$). Отже, електрони зовнішніх оболонок можуть переходити на це місце і випромінюється квант рентгенівського випромінювання. Виникають відповідно серії $K,L,M,\ldots$. Енергію електронного стана на внутрішній оболонці можна представити у вигляді:
$$E = -\frac{E_0}{n^2}\,(Z-z_{n})^2\,,\,\,\,E_0 = \frac{m\,e^4}{2\,\hbar^2} \Rightarrow \sqrt{\frac{\lambda_0}{\lambda}}\simeq Z-z_n$$
де $z_n(\ll Z)$ це екрануючий фактор, що враховує часткове екранування заряду ядра іншими електронами. Велики значення $Z$ призводять до великих значень енергії переходу. Оскільки 
внутрішні оболонки атомів мають однаковeу будову, поправка $а_п$ для даної оболонки (даного п) мають бути однаковими. Звідси випливає, що всі важкі атоми повинні мати однаково побудовані рентгенівські спектри, і у важчих атомів спектр зміщується у бік
високих частот пропорційно до $Z^2$. Це і було спостережено у експериментах - т.з. правило Мозлі (див. Рис).

Характеристичні рентгенівські переходи відбуваються згідно ''однофотонних'' правил відбору: $\Delta l = \pm 1\,,\,\,\Delta j = -1,0,1$. Наприклад K-серія складається з дублетів $K_{\alpha_1}, K_{\alpha_2}$ від переходів з термів $2 ^{2}P_{1/2}$ $2 ^{2}P_{3/2}$ і $K_{\beta_1}, K_{\beta_2}$ з термів $3 ^{2}P_{1/2}$ $3 ^{2}P_{3/2}$ на $1 ^{2}S_{1/2}$ відповідно. Можливі такох переходи з $N$-станів ($L_{\gamma_{1,2}}$ дублет $4 ^{2}P_{1/2}, 4 ^{2}P_{3/2}$) і т.д.

Рентгенівські спектри поглинання відрізняються від спектрів випромінювання оскільки немає обернення ліній як для оптичних переходів (лінія випромінювання = лінія поглинання). Якщо через шар будь-якого елемента пропустити пучок гальмівного рентгенівського випромінювання (тобто випромінювання, що розкладається в суцільний спектр), то на місці характеристичних ліній не з'являться темні лінії, як це було б у разі оптичних спектрів. Це пояснюється тим, що в речовині, через яку проходить рентгенівський пучок, немає атомів у збуджених станах. Усі внутрішні електронні оболонки заповнені, і в яких немає дірок в які при опроміненні рентгенівськими променями могли б перейти електрони з глибших енергетичних рівнів. Відповідно енергетичний поріг фотоіонізації електрона без пердачі йому значної кінетичної енергії з K-оболонки буде давати максимум в інтенсивності поглинання на відповідній частоті. Далі виникає падіння інтенсивності поглинання і поступове зростання до збудження L оболонки яке складається з послідовної іонізації термів $2 ^{2}S_{1/2}, 2 ^{2}P_{1/2,3/2}$, далі M оболонка (5 термів) і т.д.

## Ефект Оже
Фотографуючи в камері Вільсона фотоелектрони, що звільняються рентгенівськими променями з атомів важких інертних газів (наприклад, криптону), Оже виявив, що в деяких випадках в одній і тій же точці беруть початок два електронні сліду. Один електрон утворюється в результаті звичайної фотоіонізації, інший - внаслідок внутрішнього перерозподілу енергії збудженого атома - ефект Оже. Ефект Оже - це ефект автоіонізації збудженого атома. На відміну від звичайної фотоіонізації, коли поглинений рентгенівський квант призводить безпосередньо до вильоту електрона за межі атома, ефект Оже відбувається у дві стадії. На першій стадії відбувається поглинання рентгенівського кванта, що супроводжується збудженням
атома, а саме звільненням електрона з оболоньки, тобто утворенням дірки у ній. На другій стадії відбувається заповнення цієї дірки електроном, що перескакує з L-шару. Надлишок енергії або виділяється у вигляді кванта характеристичного випромінювання або призводить до викидання одного електрона з лежачих вище оболонок атома. Цей процес і називається ефектом Оже, а викинутий електрон - електроном Оже. Таким чином, дірка в L-шарі зберігається, але виникає друга дірка в одному з вищих шарів. Іншими словами, атом стає двічі іонізованим. Електрони Оже завжди забирають значну частку енергії збудженого атома. Особливо сильно ефект Оже проявляється у тих випадках, коли переходи з випромінюванням фотонів заборонені ($\Delta l = 0$).
