# Ймовірносний зміст квадрата модуля хвильової функції. Обчислення середніх.
<!--  -->
$\newcommand{\ket}[1]{\left.\left\vert\, #1\,\right. \right \rangle}$
$\newcommand{\bra}[1]{\left. \langle \, #1\, \right\vert}$
$\newcommand{\braket}[2]{\langle \,#1 \vert #2 \, \rangle}$
$\newcommand{\av}[3]{\left\langle \,#1 \left \vert\, #2 \,\right\vert #3 \,\right \rangle}$

Згідно статистичної інтерпретації результатів вимірювання положення квантової частинки квадрат модуля хвильової функції за умови нормування:
<!--  -->
\begin{equation}\label{norm}
\int\limits_{V}
{|\,\psi (\mathbf{r},t)\,|^{\,2}} d\,V  = 1 \,.
\end{equation}
<!--  -->
може бути проінтерпретований як густина розподілу ймовірності знаходження частинки у елементі об'єму $dV$
<!--  -->
\begin{equation}
\label{eq4/1} {P_{V} = \int\limits_{V} \rho\,(\mathbf{r},t)}\,dV
\end{equation}
<!--  -->
Таким чином можливо обчислення середніх значень для розподілу положення:
<!--  -->
$$ 
\langle \mathbf{r}\rangle = \int\limits_{V} \mathbf{r}\,|\psi(\mathbf{r},t)|^2\,dV  
$$
<!--  -->
та дисперсії координат $\sigma_{x_{i}} = \sqrt{\langle\Delta x^2_{i}\rangle}$:
<!--  -->
$$ 
\langle \Delta x^2_{i}\rangle = \langle x^2_{i}\rangle - \langle x_{i}\rangle^2\,,\,\,\,\,\, \langle x^2_{i}\rangle = \int\limits_{V} x^2_{i}\,|\psi(\mathbf{r},t)|^2\,dV  
$$

<!--  -->

Дуальний опису стану у термінах вимірення імпульсу частинок знаходить своє математичне відображення у співвідношеннях типу перетворення Фур'є:
<!--  -->
\begin{equation}\label{eq6/1} 
\psi (\mathbf{r},t) = {\int\limits_{\Lambda}{c(\mathbf{p},t)}}\, \psi_{\mathbf{p}} (\mathbf{r},t)d\,\mathbf{p}
\end{equation}
<!--  -->
\begin{equation}\label{eq:psipdevroigl}
\psi_{\mathbf{p}}({\rm r},\,t) =\frac{1}{(2\,\pi\,\hbar )^{3/2}}
 \exp\left(\frac{i\mathbf{p} \cdot \mathbf{r}-i\,E\,t}{\hbar}\right) =
\frac{1}{(2\pi \hbar)^{3/2}}\,\exp\left(i\,\mathbf{k}\cdot\mathbf{r} - i\,\omega(\mathbf{k}) \,t\right)\,,
\end{equation}
<!--  -->
та
<!--  -->
\begin{equation}\label{eq7/1} 
c(\mathbf{p},t) = \int\limits_{V} \psi(\mathbf{r},t)\,\psi^*_{\mathbf{p}}(\mathbf{r},t)d\,V\,.
\end{equation}
<!--  -->
Функція $c(\mathbf{p},t)$ є хвильовою функцією у __імпульсному представленні__. Відповідно $\vert c(\mathbf{p},t)\vert^2$ є густиною розподілу ймовірності значень імпульсу за умови нормування на 1). До того ж як відомо з властивстей перетворення Фур'є:
<!--  -->
$$ \int_{V} |\psi(\mathbf{r},t)|^2\,d\mathbf{r} =  \int_{\Lambda} |c(\mathbf{p},t)|^2\,d\mathbf{p}$$
<!--  -->
Таким чином можливо обчислення середніх значень для розподілу імпульса:
<!--  -->
$$ 
\langle \mathbf{p}\rangle = \int\limits_{\Lambda} \mathbf{p}\,|c(\mathbf{p},t)|^2\,d\mathbf{p}  
$$
<!--  -->
та дисперсії $\sigma_{p_{i}} = \sqrt{\langle\Delta p^2_{i}\rangle}$:
<!--  -->
$$ 
\langle \Delta p^2_{i}\rangle = \langle p^2_{i}\rangle - \langle p_{i}\rangle^2\,,\,\,\,\,\, \langle p^2_{i}\rangle = \int\limits_{\Lambda} p^2_{i}\,|c(\mathbf{p},t)|^2\,d\mathbf{p}  
$$
<!--  -->
До того ж виконується співвідношення невизначенностей Гайзенберга:
<!--  -->
\begin{equation}\label{heiz} 
\sigma_{x}\,\sigma_{p_x} \ge \frac{\hbar}{2}
\end{equation}
<!--  -->
Виходячі із взаємозв'язку координатного та імпульсного представлень для хвильової функції можна отримати наступні формули обчислення середніх:
<!--  -->
\begin{equation}
\label{eqxop}
\begin{array}{l}
\langle x \rangle \, = \int\limits_{V}
\psi^*(\mathbf{r},t) \,x\,\psi
(\mathbf{r},t)\,dV = 
\int\limits_{\Lambda} 
c^*(\mathbf{p},t)\,\left(i\hbar \,\frac{\partial}{\partial \,p_{x}}\right)\,
c(\mathbf{p},t)\,d\mathbf{p} \\
\langle p_{x} \rangle \, = \,\int\limits_{\Lambda}
c^* (\mathbf{p},t)\,p_{x}\,c(\mathbf{p},t)\,d\mathbf{p} = 
\int\limits_{V} \psi^*(\mathbf{r},t)\,\,\left(- i\hbar \,\frac{\partial}
{\partial \,x} \right)\,\psi (\mathbf{r},t)\,d\mathbf{r}\,, \\
\end{array}
\end{equation}
<!--  -->
Це демонструє те що координата та імпульс у квантовій теорії представляються операторами.  Вигляд такого оператора залежить від обранного представлення хвильової функції:
<!--  -->
$$
\begin{array}
\hat{x}\,\psi(x,t) = x\,\psi(x,t)\,,&
\hat{p}_{x}\,\psi(x,t) = -i\,\hbar\,\frac{\partial}{\partial\,x}\psi(x,t)\\
\hat{x}\,c(p,t) = i\,\hbar\,\frac{\partial}{\partial\,p_{x}}\,c(p,t)\,,& \hat{p}_{x}\,c(p,t) = p_{x}\,c(p,t)
\end{array}
$$
